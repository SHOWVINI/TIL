```python
from openai import OpenAI

model_id = "meta-llama/Meta-Llama-3-8B"

# Modify OpenAI's API key and API base to use vLLM's API server.
openai_api_key = "OPENAI_API_KEY"
openai_api_base = "http://127.0.0.1:8000/v1"
client = OpenAI(
    api_key=openai_api_key,
    base_url=openai_api_base,
)
```

```python
prompts = [
    "### Human: 숫자 8과 1233이 2의 거듭제곱인지 확인해줘.### Assistant:",
    "### Human: 75를 1555로 나눈 결과는 무엇인가요?### Assistant:",
]

completion = client.completions.create(model="oasst",
                                       prompt=prompts,
                                       temperature=0.7,
                                       top_p=0.9,
                                       max_tokens=500)
print("답변:", completion)


prompts = [
    "<user>숫자 8과 1233이 2의 거듭제곱인지 확인해줘.</user>\n\n<tools>",
    "<user>75를 1555로 나눈 결과는 무엇인가요?</user>\n\n<tools>",
]

completion = client.completions.create(model="xlam",
                                      prompt=prompts, temperature=0.0, max_tokens=500)
print("답변:", completion)
```
