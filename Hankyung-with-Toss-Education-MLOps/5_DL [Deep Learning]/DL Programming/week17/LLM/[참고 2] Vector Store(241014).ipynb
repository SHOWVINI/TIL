```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.vectorstores import Chroma
from langchain.document_loaders import TextLoader
from langchain.document_loaders import PyPDFLoader
from langchain.embeddings import HuggingFaceEmbeddings
```

```python
!wget https://wdr.ubion.co.kr/wowpass/img/event/gsat_170823/gsat_170823.pdf
```

```python
## pdf 파일로드 하고 쪼개기
loader = PyPDFLoader('gsat_170823.pdf')
pages = loader.load_and_split()
print(len(pages))
```

```python
pages[0]
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=0,
)

splited_docs = text_splitter.split_documents(pages)
len(splited_docs)
```

```python
splited_docs[0]
```

```python
len(splited_docs[0].page_content)
```

# Embedding

```python
model_huggingface = HuggingFaceEmbeddings(model_name="BAAI/bge-m3")
```

# Chroma

```python
# DB 생성 시 chunk, embedding 모델이 들어간다.

db = Chroma.from_documents(
    documents=splited_docs,
    embedding=model_huggingface,
)

print("문서의 개수 : ", db._collection.count())
```

```python
# VectorDB Query

query = "삼성전자의 주요 사업 영역은?"
docs = db.similarity_search(query)

for doc in docs:
    print(doc)
    print("----" * 30)
```

## Chroma 데이터베이스 파일 저장 & 불러오기

```python
# 벡터화된 텍스트를 Chroma DB 파일로 저장

db_toFiles = Chroma.from_documents(
    documents=splited_docs,
    embedding=model_huggingface,
    persist_directory="./samsung_db"
)

print("문서의 수 : ", db_toFiles._collection.count())
```

```python
# 불러오기

db_fromFile = Chroma(
    persist_directory="./samsung_db",
    embedding_function=model_huggingface,
)

docs = db_fromFile.similarity_search(query)
docs[0]
```

## Chroma 문서 추가

저장된 문서들은 각각 고유한 id를 갖습니다. 새로운 문서를 추가할 때, ID를 부여하는 것도 가능

```python
# id를 출력하되, 상위 5개만 출력

db_fromFile._collection.get()["ids"][:5]
```

```python
# 신규 문서 추가 하기
text1 = '''삼성전자가 미국 반도체업체 엔비디아에 고대역폭 메모리(HBM)를 납품하기 위한 테스트를 아직 통과하지 못했다고 로이터통신이 3명의 익명 소식통을 인용해 24일 보도했다.'''
text2 = '''엔비디아발 훈풍으로 전 세계 반도체가 들썩이는 가운데 국내 증시에서는 SK하이닉스 (198,600원 ▼1,400 -0.70%)의 활약이 돋보인다. 엔비디아에 대한 투자자들의 기대치가 높아지고 있어 SK하이닉스 역시 상승세를 이어갈 것이란 전망이 쏟아진다. 다만 SK하이닉스 주가에 상승동력으로 작용했던 엔비디아 실적 재료가 소진된 점은 차익실현 매물이 출현할 가능성을 높인다.'''
```

```python
# 반드시 VectorStore에 추가되어야 할 형식은 Document 형식이어야 한다.
# Document는 metadata와 page_content로 이뤄져 있어야 한다.


from langchain.docstore.document import Document

# metadata : 데이터를 설명하기 위한 데이터 (문서의 출처, 페이지 등) - 메타 데이터가 없더라도 꼭 있는 것처럼 만들어줘야 한다
new_doc = Document(
    page_content=text2,
    metadata={
        "source": "https://news.naver.com",
        "page": 1
    }
)

new_doc
```

```python
new_docs = [ new_doc ]

# 문서 추가하기 : 추가하고 나면 아이디가 나온다

id = db_fromFile.add_documents(
    new_docs,
    ids=["7"]
)

id
```

```python
db_fromFile._collection.get(id)
```

```python
db_fromFile._collection.count()
```

## Chroma 문서 삭제

```python
db_fromFile._collection.delete(ids=id) # id로 삭제
db_fromFile._collection.get(id) # 삭제된 id로 가져왔기 때문에 아무것도 안나옴
```

```python
from langchain.vectorstores import FAISS

faiss_db = FAISS.from_documents(splited_docs, model_huggingface)
```

```python
docs = faiss_db.similarity_search(query)
docs[0]
```

## FAISS 데이터베이스 파일 저장 & 불러오기

```python
faiss_db.save_local("faiss_index")
```

```python
faiss_fromFile = FAISS.load_local(
    "faiss_index",
    model_huggingface,
    allow_dangerous_deserialization=True)

docs = faiss_fromFile.similarity_search(query)
docs[0]
```

## 신규 문서 추가

```python
# 새로운 데이터를 위한 VectorStore를 따로 만든다.
new_db = FAISS.from_texts([text1, text2], model_huggingface)
```

```python
# 새롭게 만들어진 Vector Store를 기존 Vector Store와 병합
faiss_db.merge_from(new_db)
```
