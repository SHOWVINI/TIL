# 머신러닝 강화학습의 원리와 DQN의 작동 방식

---

## 1. 강화학습의 원리

### 1.1 정의

강화학습(Reinforcement Learning)은 학습 주체인 에이전트(Agent)가 환경(Environment)과 상호작용하며, 주어진 목표를 달성하기 위해 보상(Reward)을 최대화하는 방식으로 학습하는 기계학습 분야임. 

지도학습(Supervised Learning)과 달리 정답 레이블이 명시적으로 주어지지 않고, 에이전트는 보상을 기준으로 좋은 행동과 나쁜 행동을 구분하여 스스로 정책을 발전시켜 나감.

### 1.2 구성 요소

강화학습은 다음과 같은 핵심 요소로 구성됨.

| 구성 요소 | 설명 | 예시 |
| --- | --- | --- |
| Agent (에이전트) | 행동하는 주체 (학습하는 인공지능) | 게임 플레이어 |
| Environment (환경) | 에이전트가 상호작용하는 세계 | 게임 맵, 현실 공간 |
| State (상태, s) | 현재 환경의 정보 | 화면의 상황, 센서 데이터 |
| Action (행동, a) | 에이전트가 선택하는 행동 | 이동, 공격, 회피 등 |
| Reward (보상, r) | 행동에 대한 평가 | 성공 시 +1, 실패 시 -1 |
| Policy (정책, π) | 상태에 따라 행동을 결정하는 함수 | π(s) → a |
| Value Function (가치 함수) | 상태 또는 상태-행동의 장기 보상 기대값 | V(s), Q(s, a) |


### 1.3 학습 목표

강화학습의 핵심 목적은 \*\*정책(Policy)\*\*을 학습하여 에이전트의 목표는 **미래의 보상까지 고려한 누적 보상(expected return: 누적 보상의 기대값)**을 최대화하는 행동 전략(policy)을 찾는 것.

이를 위해 시간 차를 고려해 보상을 누적할 때, **감가율 γ (0 < γ ≤ 1)**을 사용.

단기 보상뿐 아니라 미래의 보상을 고려하기 위해 \*\*감가율 (Discount factor, $gamma$ $in$ $[0,1]$)\*\*을 사용함.

감가율이 클수록 미래 보상의 영향력이 커지고, 작을수록 현재 보상 중심으로 학습.

강화학습의 목적 함수는 다음과 같음:

$Gt=rt+γrt+1+γ2rt+2+⋯=k=0∑∞γkrt+k$

강화학습에서의 최종 목표는 최적의 정책 $pi$를 찾아 다음을 최대화하는 것임:

$$
\pi^* = \arg\max_\pi \mathbb{E}[R_t] = \arg\max_\pi \mathbb{E}\left[ \sum_{k=0}^\infty \gamma^k r_{t+k} \right]
$$


### 1.4 학습 전략: 탐험과 이용

강화학습에서 중요한 것은 “새로운 행동을 시도할 것인가, 아니면 지금까지 가장 좋았던 행동을 반복할 것인가”의 선택.

* **탐험 (Exploration)**: 에이전트가 아직 시도하지 않은 행동을 시도하여 더 나은 보상을 찾기 위한 전략
* **이용 (Exploitation)**: 현재까지 학습한 정보를 바탕으로 가장 높은 보상을 줄 것으로 예상되는 행동을 선택하는 전략

에이전트는 이 두 전략 사이에서 균형을 유지해야 하며, 일반적으로 ε-greedy 정책(\*\*$epsilon$-탐욕 정책 ($epsilon$-greedy policy)\*\*)을 사용하여 $epsilon$의 확률로 무작위 탐험을 수행하고, $1 - epsilon$의 확률로 현재 최적의 행동을 선택함.

- ε의 확률로 무작위 행동 (탐험)

- 1-ε의 확률로 현재 최선의 행동 선택 (이용)

### 1.5 알고리즘 유형 분류

| 분류                    | 정의                                      | 대표 알고리즘         |
| --------------------- | --------------------------------------- | --------------- |
| 가치 기반 (Value-based)   | 각 상태-행동 쌍의 가치를 추정하고 그 값을 최대화하는 방향으로 학습함 | Q-Learning, DQN |
| 정책 기반 (Policy-based)  | 정책 자체를 직접 파라미터화하여 최적화를 수행함              | REINFORCE, PPO  |
| 액터-크리틱 (Actor-Critic) | 정책과 가치 함수를 동시에 학습하여 안정성과 수렴 속도를 모두 확보함  | A2C, A3C, DDPG  |


### 1.6 응용 분야

* **게임**: 알파고(바둑), DQN 기반 아타리 게임, 스타크래프트 전투 전략
* **로보틱스**: 로봇 팔의 정밀 제어, 보행 로봇의 균형 유지 및 목표 이동
* **금융**: 주식 매매 시점 결정, 포트폴리오 구성 최적화, 옵션 가격 예측
* **추천 시스템**: 사용자 행동에 따른 실시간 정책 갱신을 통한 개인화 추천
* **산업 제어**: HVAC 시스템의 에너지 최적화, 생산공정 자동 제어

## 2. DQN (Deep Q-Network)의 원리와 동작

### 2.1 개요

DQN은 Q-Learning의 가치 기반 강화학습 방법에 심층 신경망을 결합한 알고리즘으로, DeepMind에서 2015년에 발표됨. 

Q-Learning의 테이블 기반 방식이 고차원 상태 공간에서는 메모리와 계산량의 한계로 인해 적용이 어려운 문제를 해결하기 위해 제안됨. 

DQN은 이미지와 같은 복잡한 입력을 처리하여 Q값을 근사함.


> 근사하다: 정확한 값을 구하기는 어렵지만, 그와 매우 비슷한 값을 구하는 것

신경망은 상태-행동 쌍에 대해 정확한 Q값을 계산하지는 않지만, 매우 비슷한 값을 출력해 주기 때문에 우리는 “신경망이 Q함수를 근사한다”고 말함.


### 2.2 Q-Learning

Q-Learning은 상태 $s$에서 행동 $a$를 선택했을 때 얻는 장기 보상의 기대값인 $Q(s, a)$를 반복적으로 업데이트함. 업데이트 식은 다음과 같음:

$$
Q(s, a) \leftarrow Q(s, a) + \alpha \left[ r + \gamma \max_{a'} Q(s', a') - Q(s, a) \right]
$$

이때 $alpha$는 학습률(learning rate), $gamma$는 미래 보상에 대한 감가율임.

- Q-Learning의 한계

Q-Learning은 모든 상태-행동 쌍을 테이블에 저장해야 함.

하지만 상태가 수천 개, 수백만 개라면 이건 불가능에 가까움.

예: 이미지가 상태라면 100x100픽셀만 되어도 상태의 수는 10,000차원 이상임

- DQN(Deep Q-Network)의 등장

DQN은 Q함수를 **신경망(Deep Neural Network)**으로 근사하는 방식.

즉, 테이블 대신 **딥러닝 모델이 Q값을 예측**하는 구조.


### 2.3 DQN의 핵심 구성 요소

* **Q-Network**: 

상태 $s$를 입력으로 받아 가능한 행동 $a$들에 대한 Q값을 출력하는 심층 신경망

* **Replay Buffer**: 

에이전트의 경험 $(s, a, r, s')$(=이전의 경험: 상태, 행동, 보상, 다음 상태)을 저장하고 무작위 샘플링하여 학습에 활용하는 버퍼 구조. 

샘플 간 상관성을 제거하고 학습의 분산을 낮추기 위한 목적임

* **Target Network**: 

Q-Network의 구조를 복사한 별도 네트워크로, 일정 주기마다 파라미터를 업데이트하여 안정적인 학습을 유도함

### 2.4 학습 절차

1. 에이전트가 현재 상태 $s$에서 정책 $epsilon$-탐욕 방식으로 행동 $a$를 선택함
2. 환경으로부터 보상 $r$과 다음 상태 $s'$를 관찰함
3. 경험 $(s, a, r, s')$을 Replay Buffer에 저장함
4. 미니배치 형태로 경험을 샘플링하여 손실 함수를 정의함:

$$
L = \mathbb{E}\left[ \left( r + \gamma \max_{a'} Q_{\text{target}}(s', a') - Q(s, a) \right)^2 \right]
$$

5. 이 손실을 역전파(Backpropagation)로 최소화하도록 Q-Network를 학습함
6. 일정 간격으로 Target Network의 파라미터를 Q-Network에서 복사하여 동기화함

### 2.5 주요 장점과 한계

* **장점**:

  * CNN 기반 Q-Network를 사용하여 이미지 입력 등의 고차원 상태 처리 가능
  * 경험 재사용을 통해 효율적 학습 가능
  * Target Network 도입으로 학습의 안정성 향상

* **한계**:

  * 연속적 행동 공간 문제 해결에는 한계 존재 (→ DDPG, SAC 등 필요)
  * Q값의 과대추정 문제 발생 가능 (→ Double DQN으로 보완 가능)
  * 리워드 희소 환경에서는 수렴 속도가 느림

### 2.6 응용 사례

* **아타리 게임 마스터링**: DQN은 Breakout, Pong 등에서 인간을 능가하는 성능을 달성함
* **로보틱스 제어**: 시뮬레이션 환경에서 로봇 팔 제어, 장애물 회피 등 구현 가능
* **웹 최적화**: 사용자 행동 데이터를 기반으로 페이지 구성, 광고 배치 등을 실시간으로 학습함
* **스마트 팩토리**: 설비 자동 제어, 생산 속도 조절, 에너지 소비 최적화

# 요약 정리

강화학습은 보상을 기반으로 한 학습 방식으로, 동적인 환경 속에서도 자율적인 의사결정이 가능한 시스템을 설계하는 데 핵심적인 기술임. DQN은 Q-learning의 근사 기법으로서 딥러닝을 도입함으로써 복잡한 환경에서도 학습을 가능하게 하였으며, 게임 AI, 로보틱스, 추천 시스템 등에서 큰 성과를 거둠.

향후에는 DQN을 기반으로 한 다양한 변형(예: Double DQN, Dueling DQN, Prioritized Experience Replay)을 활용하여 더 안정적이고 효율적인 강화학습 모델 개발이 가능함.

| 구분 | 설명 |
| --- | --- |
| 강화학습 | 보상을 통해 행동 전략을 학습 |
| Q-Learning | 상태-행동의 장기 보상 기대값을 테이블로 학습 |
| DQN | Q값을 신경망으로 예측하는 딥러닝 기반 강화학습 |
| 핵심 기능 | Replay Buffer, Target Network, ε-greedy 탐험 |
| 실제 적용 | 게임 AI, 로봇, 자율주행, 광고 추천 등 |

# 간단한 코드 예제

```python
"""
[간단한 DQN 구현 예제 - CartPole 환경]

개요:
- 본 코드는 OpenAI Gym의 CartPole-v1 환경을 대상으로 Deep Q-Network(DQN)를 구현한 예제임.
- Q-Learning의 원리를 딥러닝과 결합하여 고차원 상태공간에서도 학습이 가능하도록 설계함.

구성 단계:

1. 환경 초기화
    - OpenAI Gym을 활용하여 강화학습 환경을 불러옴
    - 상태 공간과 행동 공간의 크기를 확인하고 변수로 저장함

2. DQNAgent 클래스 정의
    - Q-Network 모델 정의 (2층 MLP 사용)
    - 탐험(ε-greedy), 리플레이 버퍼, 감가율, 학습률 등의 하이퍼파라미터 정의
    - 정책 함수(act), 경험 저장(rememeber), 학습 함수(replay) 구현

3. 학습 루프
    - 정해진 에피소드 수만큼 학습 반복
    - 각 에피소드마다:
        - 상태 초기화 후 행동 선택
        - 보상 및 다음 상태 관찰
        - 경험 저장
        - 일정 조건 만족 시 미니배치 학습 수행
    - 학습이 진행될수록 탐험 비율(ε)은 감소

4. 학습 결과 출력
    - 각 에피소드마다 점수(score)와 현재 ε 값 출력
    - 훈련이 반복될수록 점수 증가 및 ε 감소를 통해 학습의 진행 확인 가능

기술 요약:
- Replay Buffer를 사용하여 샘플 간 상관성을 줄이고 학습 안정성 확보
- 신경망으로 Q함수를 근사하여 고차원 상태에서도 Q-Learning 수행 가능
- ε-greedy 정책을 통해 초기에는 탐험 위주, 후반에는 이용 위주 학습 전략 수행

"""
```

```python
import gym
import numpy as np
import random
from collections import deque
import tensorflow as tf
from tensorflow.keras import models, layers, optimizers

# ▶ DQN 에이전트 정의
class DQNAgent:
    def __init__(self, state_size, action_size):
        self.state_size = state_size                         # 상태 공간의 크기 정의
        self.action_size = action_size                       # 행동 공간의 크기 정의
        self.memory = deque(maxlen=2000)                     # 경험 리플레이를 위한 메모리 버퍼
        self.gamma = 0.95                                    # 감가율(미래 보상에 대한 할인율)
        self.epsilon = 1.0                                   # 탐험 비율 초기값
        self.epsilon_min = 0.01                              # 탐험 비율의 최솟값
        self.epsilon_decay = 0.995                           # 탐험 비율 감소율
        self.learning_rate = 0.001                           # 학습률
        self.model = self._build_model()                     # Q-네트워크 모델 생성

    def _build_model(self):
        # 간단한 2층 MLP 신경망 정의
        model = models.Sequential()
        model.add(layers.Dense(24, input_dim=self.state_size, activation='relu'))
        model.add(layers.Dense(24, activation='relu'))
        model.add(layers.Dense(self.action_size, activation='linear'))  # Q값 출력
        model.compile(loss='mse', optimizer=optimizers.Adam(learning_rate=self.learning_rate))
        return model

    def remember(self, state, action, reward, next_state, done):
        # 경험을 메모리에 저장하는 함수
        self.memory.append((state, action, reward, next_state, done))

    def act(self, state):
        # ε-greedy 정책에 따라 행동을 선택
        if np.random.rand() <= self.epsilon:
            return random.randrange(self.action_size)  # 무작위 행동
        q_values = self.model.predict(state, verbose=0)
        return np.argmax(q_values[0])  # 가장 높은 Q값을 가진 행동 선택

    def replay(self, batch_size):
        # 미니배치를 사용해 Q-Network 업데이트
        minibatch = random.sample(self.memory, batch_size)
        for state, action, reward, next_state, done in minibatch:
            target = reward
            if not done:
                # 다음 상태에서 얻을 수 있는 최대 Q값을 반영하여 타깃 계산
                target += self.gamma * np.amax(self.model.predict(next_state, verbose=0)[0])
            target_f = self.model.predict(state, verbose=0)
            target_f[0][action] = target  # 해당 행동에 대한 Q값만 업데이트
            self.model.fit(state, target_f, epochs=1, verbose=0)

        # ε 값을 점차 줄여 탐험을 줄이고 이용을 강화
        if self.epsilon > self.epsilon_min:
            self.epsilon *= self.epsilon_decay

# ▶ 메인 실행 함수
env = gym.make('CartPole-v1')  # 환경 초기화
state_size = env.observation_space.shape[0]  # 상태 공간 크기 (4)
action_size = env.action_space.n             # 행동 공간 크기 (2)
agent = DQNAgent(state_size, action_size)
episodes = 300
batch_size = 32

# 에이전트 학습 루프
for e in range(episodes):
    state = env.reset()
    state = np.reshape(state, [1, state_size])  # 신경망 입력 형식에 맞게 변형

    for time in range(500):  # 한 에피소드의 최대 길이
        # 환경 시각화는 원할 경우 아래 주석 해제
        # env.render()

        action = agent.act(state)  # 행동 선택
        next_state, reward, done, _, _ = env.step(action)  # 환경에 행동 적용
        reward = reward if not done else -10  # 에피소드가 끝났다면 패널티 부여
        next_state = np.reshape(next_state, [1, state_size])

        agent.remember(state, action, reward, next_state, done)  # 경험 저장
        state = next_state

        if done:
            print(f"episode: {e + 1}/{episodes}, score: {time}, epsilon: {agent.epsilon:.2f}")
            break

    # 메모리에 충분한 경험이 쌓이면 학습 수행
    if len(agent.memory) > batch_size:
        agent.replay(batch_size)
```
