# **1. 생성 모델(Generative Models)**

## 정의 및 개념

**생성 모델**은 데이터의 **확률 분포를 학습**하여, **새로운 데이터를 생성할 수 있는 모델**. 

일반적인 머신러닝 모델이 입력 데이터를 받아 **분류(classification)** 또는 **회귀(regression)** 와 같은 예측을 수행하는 것과 달리, 생성 모델은 **실제처럼 보이는 가짜 데이터를 만들어내는 것**이 주된 목적.

### 예시

- 고양이 사진을 학습한 생성 모델은 **실제로 존재하지 않는 새로운 고양이 사진을 생성**할 수 있음.
- 문장을 학습한 생성 모델은 **새로운 자연어 문장**을 생성할 수 있음.

---

## 생성 모델의 종류

| 분류 | 모델 예시 | 특징 |
| --- | --- | --- |
| 확률 기반 | Naive Bayes, Hidden Markov Model | 비교적 단순한 모델, 특정 가정 하에 작동 |
| 에너지 기반 | Boltzmann Machine, RBM | 상태 에너지 함수 기반의 샘플 생성 |
| 오토인코더 기반 | Variational Autoencoder (VAE) | 잠재 공간에서 샘플링, 확률 해석 가능 |
| 적대적 학습 기반 | Generative Adversarial Network (GAN) | 판별자와 생성자의 경쟁 구조 |
| 자기 회귀 기반 | PixelRNN, GPT 등 | 이전 상태를 기반으로 순차적으로 샘플 생성 |

---

## 수학적 정의 (기초 수준)

데이터 분포를 $p_{data}(x)$라고 할 때, 생성 모델은 이와 유사한 확률 분포 $p_{model}(x)$를 학습하여, $x \sim p_{model}(x)$ 와 같은 데이터를 생성하는 것이 목표.

즉, 생성 모델은 다음을 학습:

> 실제 데이터 분포와 구별 불가능할 정도로 유사한 새로운 샘플을 만들어 내는 함수 또는 확률 모델

---

## 주요 응용 분야

- 이미지 생성 (예: 위조 이미지, 예술 작품)
- 텍스트 생성 (GPT 모델 등)
- 음성 합성 (TTS: Text-to-Speech)
- 데이터 증강 (학습용 데이터 생성)
- 의료 영상 생성 및 보완

---

## 생성 모델의 학습 방식

생성 모델은 보통 **비지도 학습** 또는 **자기지도 학습**의 형태로 학습. 

학습 데이터에 정답(label)이 없더라도, 데이터의 구조를 이해하고 샘플을 재구성할 수 있도록 설계됨.

---

## [정리]

- 기존 데이터의 분포를 학습하고, 새로운 샘플을 생성.
- **GAN**, **VAE**, **Autoregressive Models**, **Flow-based Models** 등 다양한 아키텍처가 존재.
- GAN은 생성 모델 중에서도 매우 인기 있고 강력한 구조로, **적대적 훈련(adversarial training)**을 통해 매우 사실적인 결과를 얻음.

# **2. 생성 적대 신경망(GAN)의 원리와 동작**

## GAN(Generative Adversarial Network)의 개념

**GAN**은 2014년 Ian Goodfellow가 제안한 생성 모델로, **두 개의 신경망(G, D)이 서로 경쟁**하며 학습.

| 구성요소 | 설명 |
| --- | --- |
| Generator (G) | 진짜 같은 데이터를 생성하는 네트워크 |
| Discriminator (D) | 입력된 데이터가 진짜(real)인지 가짜(fake)인지 구분하는 네트워크 |



> Generator는 Discriminator를 속이기 위해 점점 더 정교한 데이터를 생성하고, Discriminator는 이를 더 잘 구분하려고 학습.
 
> 이 둘의 **적대적 훈련(adversarial training)** 과정을 통해 생성자는 매우 사실적인 데이터를 생성하는 능력을 갖추게 됨.


---

## GAN의 학습 구조

1. Generator는 **무작위 벡터 z** (노이즈)를 입력으로 받아, 가짜 데이터를 생성함:
    
    $ G(z) \rightarrow \tilde{x} $ 
    
2. Discriminator는 실제 데이터(x)와 생성된 데이터(𝑥̃)를 받아, 진짜인지 가짜인지 확률로 출력함:
    
    $D(x) \rightarrow 1 (진짜)$, $\quad D(\tilde{x}) \rightarrow 0 (가짜)$
    
3. 두 모델은 다음의 목적을 가짐:

    - **Generator**는 Discriminator를 속이도록 학습:
        
        $\min_G \log(1 - D(G(z)))$
        
    - **Discriminator**는 진짜와 가짜를 정확히 분류하도록 학습:
        
        $ \max_D [logD⁡(x)+log⁡(1−D(G(z)))]$
        
- Generator는 `Discriminator를 속이기 위해` 점점 더 사실적인 가짜 데이터를 생성함.
- Discriminator는 `실제와 가짜를 더 잘 구별하기 위해` 학습함.
- 이 경쟁이 이어질수록 Generator는 더 사실적인 데이터를 만들게 됨.


---

## GAN 학습의 특징

- **학습의 불안정성**: 두 모델이 균형을 맞추지 못하면 학습이 실패하거나 발산함.
- **모드 붕괴(Mode Collapse)**: Generator가 매우 유사한 형태의 데이터만 반복적으로 생성하는 현상.
- **데이터 다양성과 품질 간의 트레이드오프** 존재

---

## GAN의 동작 흐름 (Step-by-step)

1. 랜덤 노이즈 벡터 z 생성
2. Generator가 z를 기반으로 가짜 이미지 생성
3. 진짜 이미지와 가짜 이미지를 섞어 Discriminator에게 입력
4. Discriminator는 둘을 판별
5. 판별 결과에 따라 두 모델 각각 손실 계산 및 역전파 수행
6. Generator와 Discriminator를 번갈아가며 반복 학습


#### 비유하자면

- **Generator**는 **위조지폐범**
- **Discriminator**는 **지폐 감별사**

> 위조지폐범은 감별사를 속이기 위해 더 정교한 지폐를 만들고,

> 감별사는 진짜와 가짜를 더 잘 감별하려 함.

> 이 싸움이 계속될수록 위조지폐범은 거의 진짜 같은 지폐를 만들게 됨.

---

## GAN의 대표적 활용 사례

| 분야 | 응용 사례 |
| --- | --- |
| 이미지 생성 | 사람 얼굴, 풍경, 애니메이션 등 |
| 이미지 변환 | 흑백 → 컬러, 낮 → 밤, 여름 → 겨울 등 |
| 예술 창작 | 새로운 그림 스타일 생성 |
| 영상 복원 | 저해상도 → 고해상도, 노이즈 제거 |
| 데이터 증강 | 의학 이미지 생성, 표본 부족 상황 해결 |

# **3. VAE vs GAN: 생성 모델 비교**

| 항목 | VAE (Variational Autoencoder) | GAN (Generative Adversarial Network) |
| --- | --- | --- |
| 생성 방식 | 확률적 샘플링 기반 | 적대적 훈련 기반 |
| 학습 안정성 | 비교적 안정적 | 매우 불안정할 수 있음 |
| 생성 결과 | 흐릿하거나 평균적인 경향 | 선명하고 사실적인 이미지 |
| 활용 목적 | 잠재 공간 탐색, 의미 기반 조작 | 고품질 생성, 이미지 변환 |
| 손실 함수 | Reconstruction Loss + KL Divergence | Binary Cross Entropy 기반 Adversarial Loss |
| 장점 | 해석 가능, 잠재 공간 의미 있음 | 고해상도, 사실적 샘플 생성 |
| 단점 | 생성 이미지 품질 낮음 | 학습 불안정, 모드 붕괴 문제 |

### 결론:

- **VAE**는 확률 분포를 명시적으로 모델링하고, 의미 기반 학습에 유리함.
- **GAN**은 고품질의 실제 같은 데이터를 생성하는 데 더 적합함.

# **4. 발전형 GAN 아키텍처**

GAN의 한계(모드 붕괴, 불안정성)를 극복하기 위한 다양한 변형이 등장.

### 1. DCGAN (Deep Convolutional GAN)

- CNN 기반 GAN
- Convolution layer를 활용하여 **이미지 생성 품질 개선**
- GAN의 입문 구조로 많이 사용

### 2. WGAN (Wasserstein GAN)

- **Loss를 Wasserstein Distance**로 변경
- **Gradient vanishing 문제 완화**
- Discriminator 대신 "critic" 사용
- Lipschitz 조건을 만족시키기 위한 gradient penalty 사용 (WGAN-GP)

### 3. Conditional GAN (cGAN)

- 생성자와 판별자에 **조건 정보(class label 등)**를 함께 입력
- 예: 특정 숫자(3)를 생성하게 하는 GAN
- $G(z, y)$, $D(x, y)$ 구조

### 4. CycleGAN

- **도메인 간 이미지 변환** 가능 (예: 말 ↔ 얼룩말)
- 라벨 없이도 두 이미지 도메인 간 매핑 학습
- Cycle Consistency Loss 사용

### 5. StyleGAN

- NVIDIA에서 개발, **고해상도 얼굴 생성** 가능
- 잠재 벡터의 스타일적 조작이 가능
- Facial attribute control (예: 나이, 표정 조절 등)

---

# **5. GAN 학습에서 자주 겪는 문제**

| 문제 | 설명 | 해결 전략 |
| --- | --- | --- |
| 모드 붕괴 | G가 소수의 결과만 반복적으로 생성 | Mini-batch Discrimination, WGAN |
| Gradient Vanishing | G의 gradient가 너무 작아 학습 불능 | WGAN, Batch Normalization |
| 학습 불균형 | D가 너무 강하거나 G가 너무 약함 | 학습 비율 조정, Regularization |
| 불안정한 Loss | 학습이 반복돼도 Loss가 수렴하지 않음 | WGAN-GP, Spectral Normalization |
