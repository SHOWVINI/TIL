## 1. 군집화(Clustering) 기법

1. **K-means**  
   
   - 각 군집의 중심(centroid)을 반복적으로 업데이트하며 군집을 형성하는 알고리즘임  
   - 사전에 군집 개수 K를 설정해야 하며, 유클리디언 거리 등을 통해 중심으로부터의 오차제곱합(SSE)을 최소화함  
   - 단순하고 계산량이 적으나, 초깃값 및 K 값 선정에 따라 결과 품질이 크게 달라질 수 있음  

      - **개념**: 
         - 데이터를 미리 정한 K개의 그룹(군집)으로 나누는 알고리즘임.

      - **방식**:
         - 초기 단계에서는 K개의 중심점(centroid)을 무작위로 선정함.
         - 각 데이터 포인트를 가장 가까운 중심점에 할당함.
         - 군집 내 데이터의 평균값을 새 중심점으로 갱신함.
         - 수렴할 때까지 반복하는 방식임.

      - **특징**:
         - 단순하고 계산량이 적음.
         - 군집 개수 K를 사전에 지정함.
         - 초기 중심점 선정 및 K 값에 따라 결과 품질이 달라질 수 있음.

2. **계층적 군집화(Hierarchical Clustering)**  

   - 데이터 간 거리나 유사도를 기반으로 트리 형태(덴드로그램)를 생성함  
   - **상향식(Agglomerative)** 방식은 개별 데이터 포인트에서 시작해 가장 유사한 군집끼리 병합하고,  
     **하향식(Divisive)** 방식은 전체 데이터를 하나의 군집으로 본 뒤 점차 세분화해 나가는 접근임  
   - 덴드로그램을 통해 군집 개수를 시각적으로 결정할 수 있음  
   - 대규모 데이터에 대해 계산 복잡도가 높은 단점이 있음  

      - **개념**: 
         - 데이터 간의 유사도(거리)를 기준으로 단계적으로 군집을 합치거나 분할하는 방식임.

      - **방식**:
         - **상향식(Agglomerative)**: 각 데이터를 개별 군집으로 시작하여, 가장 유사한 두 군집을 합치는 방식임.
         - **하향식(Divisive)**: 전체 데이터를 하나의 군집으로 보고, 점진적으로 세분화하는 방식임.

      - **특징**:
         - 덴드로그램(트리 구조)을 통해 군집 개수를 시각적으로 결정할 수 있음.
         - 대규모 데이터에 대해서는 계산 복잡도가 높은 단점이 있음.


3. **DBSCAN (Density-Based Spatial Clustering of Applications with Noise)**  
   - 임의의 점 주변 밀도가 충분히 높은 경우(ε 반경 내 점의 개수가 일정 기준 이상), 해당 점을 ‘핵심 점(core point)’으로 분류하고 연결된 핵심 점들의 집합을 군집으로 정의함  
   - 사전에 군집 개수를 설정할 필요가 없음  
   - 이상치(outlier)나 노이즈를 군집과 분리하여 탐지할 수 있음  
   - ε과 최소 점 개수(minPts) 설정이 까다로울 수 있음  

      - **개념**: 
         - 데이터 포인트 주변의 밀도를 기반으로 군집을 정의하는 알고리즘임.

      - **방식**:
         - 임의의 점을 기준으로 ε 반경 내에 일정 수 이상의 점이 존재하면 해당 점을 핵심 점(core point)으로 판단함.
         - 연결된 핵심 점들을 하나의 군집으로 묶음.
         - 기준에 미달하는 점들은 이상치(노이즈)로 분류함.

      - **특징**:
         - 군집 개수를 사전에 설정할 필요가 없음.
         - 이상치 탐지에 유리함.
         - ε과 최소 점 개수(minPts) 설정이 중요함.

4. **가우시안 혼합 모델(Gaussian Mixture Model, GMM)**  
   - 여러 개의 가우시안(정규 분포)이 혼합된 것으로 데이터를 설명하고, 각 분포의 파라미터를 추정함  
   - Expectation-Maximization(EM) 알고리즘으로 각 데이터가 어느 분포에 속할 확률을 계산함  
   - K-means보다 복잡한 데이터 분포를 모형화하는 데 유리함  
   - 가우시안 가정이 실제 데이터 특성과 맞지 않으면 성능이 떨어질 수 있음  

      - **개념**: 
         - 데이터가 여러 개의 가우시안(정규) 분포의 혼합으로 이루어졌다고 가정하는 모델임.

      - **방식**:
         - EM(Expectation-Maximization) 알고리즘을 사용하여 각 데이터 포인트가 어느 분포에 속할 확률을 추정함.
         - 분포의 평균, 분산 등의 파라미터를 반복적으로 업데이트함.

      - **특징**:
         - 복잡한 데이터 분포 모형화에 유리함.
         - 가우시안 분포 가정이 실제 데이터와 부합하지 않을 경우 성능 저하 가능성이 있음.

## 2. 차원 축소(Dimensionality Reduction) 기법

1. **주성분분석(PCA, Principal Component Analysis)**  
   - 고차원 데이터를 저차원으로 사영(projection)하여 분산(variance)을 최대화하는 방향(주성분)을 찾는 방법임  
   - 선형 변환을 사용하므로, 비선형 구조의 데이터에서는 한계가 있음  
   - 특성 간 상관성을 줄이고 시각화 및 전처리에 유용함  

      - **개념**: 
         - 데이터의 분산을 최대한 보존하는 방향(주성분)을 찾아 고차원 데이터를 저차원으로 사영하는 기법임.
      - **방식**:
         - 데이터의 공분산 행렬 계산 후 고유값분해(또는 SVD)를 수행함.
         - 가장 분산이 큰 주성분들을 선정하여 차원을 축소함.
      - **특징**:
         - 선형 변환 기반임으로 비선형 데이터 구조에서는 한계가 있음.
         - 특성 간 상관성을 제거하고 중요한 정보는 유지함.

2. **오토인코더(Autoencoder)**  
   - 입력을 저차원 잠재 공간(latent space)으로 압축(인코딩)한 뒤 다시 원본 차원으로 복원(디코딩)하는 신경망 구조임  
   - 은닉층의 비선형성을 통해 복잡한 패턴 포착 및 차원 축소가 가능함  
   - 네트워크 구조 및 학습 기법(규제, 드롭아웃 등)에 따라 노이즈 제거와 고유 특성 추출에 효과적임 

      - **개념**: 
         - 입력 데이터를 낮은 차원의 잠재 공간으로 압축한 후, 다시 원본으로 복원하는 신경망 구조임.
      - **방식**:
         - 인코더가 데이터를 압축함.
         - 디코더가 압축된 데이터를 원본과 유사하게 복원하도록 학습함.
      - **특징**:
         - 은닉층의 비선형 활성화 함수를 통해 복잡한 패턴도 포착함.
         - 노이즈 제거 및 이상치 탐지에도 활용 가능함. 

3. **t-SNE (t-distributed Stochastic Neighbor Embedding)**  
   - 고차원 데이터 간 확률적 거리(인접도)를 저차원에서도 유사하게 유지하는 비선형 차원 축소 기법임  
   - 시각화 품질이 우수하여 이미지나 텍스트 임베딩 등을 탐색하는 데 활용도가 높음  
   - 대규모 데이터에서는 계산량이 많고, 퍼플렉시티(perplexity) 등 매개변수 설정에 따라 결과가 좌우됨  

      - **개념**: 
      - 고차원 데이터의 유사도(거리) 정보를 저차원에서도 최대한 보존하도록 변환하는 비선형 차원 축소 기법임.
      - **방식**:
         - 데이터 간의 확률적 유사도를 계산하여 저차원 공간에 반영함.
      - **특징**:
         - 2차원 또는 3차원 시각화에 매우 효과적임.
         - 대규모 데이터에서는 계산 부담이 큼.
         - 퍼플렉시티(perplexity) 등의 매개변수 설정에 따라 결과가 달라짐.

4. **UMAP (Uniform Manifold Approximation and Projection)**  
   - 데이터가 잠재적 매니폴드 구조를 가진다고 가정하고 이를 저차원 공간에 근사화하는 기법임  
   - t-SNE 대비 빠른 속도와 전체 구조(글로벌 구조) 보존 능력이 뛰어남  

      - **개념**: 
         - 고차원 데이터가 잠재적 매니폴드 구조 위에 있다고 가정하고, 이를 저차원에 근사하는 기법임.
      - **방식**:
         - 데이터의 국소 및 글로벌 구조를 고려하여 저차원에 매핑함.
      - **특징**:
         - t-SNE보다 빠른 연산 속도를 가짐.
         - 국소적 구조뿐만 아니라 전체적인(글로벌) 구조도 보존함.

## 3. 연관 규칙 학습(Association Rule Learning)

- 대규모 거래 데이터(장바구니 분석)에서 품목 간 동시 발생 패턴을 찾는 데 활용됨  
- 대표 알고리즘으로 Apriori, FP-Growth 등이 있음  
- 지지도(support), 신뢰도(confidence) 등 지표를 사용해 유의미한 규칙을 탐색함  
- 지도학습과 달리 라벨 없이 데이터 간 연관성을 도출함

## 4. 이상치 탐지(Anomaly Detection)

- 대다수 데이터와는 다른 소수의 극단값(이상치, 이상 패턴)을 식별하는 기법임  
- 군집화(군집 중심에서 크게 벗어난 데이터를 이상치로 간주) 또는 확률 분포 기반(혼합 모델에서 확률이 매우 낮은 지점 판별) 접근이 있음  
- 라벨이 없거나 이상치가 매우 드문 상황에서 주로 활용됨

## 5. 기타 주요 비지도학습 분야

```python
# 한글 안깨지게
import matplotlib as mpl
import matplotlib.pyplot as plt

# Mac 기본 한글 폰트 설정 예시
mpl.rc('font', family='AppleGothic')   # macOS 기본 한글 폰트
plt.rcParams['axes.unicode_minus'] = False  # 음수 기호 깨짐 방지
```

```python
# =============================================================================
# 1. 군집화(Clustering) 기법 실습: K-means, 계층적 군집화, DBSCAN, Gaussian Mixture Model
# =============================================================================
```

```python
import numpy as np
from sklearn.datasets import make_blobs

# 인공 데이터 생성: 300개의 샘플과 클러스터 4개 갖는 데이터
X, y_true = make_blobs(
    n_samples = 300,
    centers = 4,
    cluster_std = 0.60,
    random_state = 42,
)
```

```python
# ----- K-means 클러스터링 -----
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=4, random_state=42)
y_kmeans = kmeans.fit_predict(X) # K-means 모델을 데이터 X에 적합시키고, 각 데이터 포인트의 군집 번호를 예측하여 y_kmeans에 저장

plt.figure(figsize=(6, 4))
plt.scatter(X[:, 0], X[:, 1],  # 데이터 X의 첫 번째와 두 번째 특징(좌표)을 산점도로 그림 ( X는 300개의 샘플(sample)과 2개의 특징(feature)을 가지는 2차원 배열. 즉, X.shape는 (300, 2). 각 행은 하나의 데이터 포인트 (예: [4.7, 2.3]), 각 열은 특징(feature) 또는 **좌표값(x, y)**를 의미. X[:, 0]과 X[:, 1]은 각각? => X[:, 0] : 각 샘플의 첫 번째 특징값 (x좌표), X[:, 1] : 각 샘플의 두 번째 특징값 (y좌표) )
            c=y_kmeans,         # 군집 번호에 따라 색상을 다르게 지정
            s=50,               # 각 점의 크기 설정
            cmap='viridis')     # 색상 맵은 'viridis' 사용

# 군집의 중심점을 빨간색 'X' 마커로 표시
plt.scatter(kmeans.cluster_centers_[:, 0],  # 중심점의 x 좌표
            kmeans.cluster_centers_[:, 1],  # 중심점의 y 좌표
            s=200,                          # 중심점의 크기
            c='red',                        # 색상은 빨간색
            marker='X')                     # 마커 모양은 X
plt.title('K-means 클러스터링 결과')  
plt.xlabel('특징 1')               
plt.ylabel('특징 2')              
plt.show()                        
```

```python
# ----- 계층적 군집화 (Hierarchical Clustering) -----
from scipy.cluster.hierarchy import dendrogram, linkage  # 덴드로그램과 linkage 함수

# linkage 함수를 사용하여 데이터를 계층적으로 군집화
# method='ward'는 분산을 최소화하는 방법을 사용
Z = linkage(X, method='ward')

plt.figure(figsize=(10, 7)) 
dendrogram(Z)               # 계층적 군집화 결과를 덴드로그램 형태로 시각화
plt.title('계층적 군집화 덴드로그램')  
plt.xlabel('샘플 인덱스')       
plt.ylabel('거리')               
plt.show()       

# 계층적 군집화는 결론적으로 큰 하나의 범주에 속해 있는 것인가? 그리고 이 안에서 하나 하나 계층적으로 나눠지는 것이고?
# 계층적 군집화(Hierarchical Clustering)는 처음에는 모든 데이터를 하나의 큰 범주로 간주한 후, 이 범주 안에서 유사한 데이터끼리 점점 묶어가거나 나누어 가며 계층을 만들어 가는 방식

# 계층적 군집화에는 두 가지 방식이 있음:

    # Agglomerative (병합형) – 가장 많이 사용됨
        # 각 데이터를 하나의 군집으로 시작 → 비슷한 것끼리 합쳐서 점점 큰 군집을 형성함
        # (작은 것들을 아래서부터 위로 모아가는 방식)
    # Divisive (분할형)
        # 전체 데이터를 하나의 군집으로 시작 → 점점 쪼개며 분할
        # (큰 집합을 위에서부터 아래로 나누는 방식)
```

```python
# ----- 오토인코더 (Autoencoder) -----
# TensorFlow 및 Keras 관련 모듈 import
import tensorflow as tf
from tensorflow.keras.layers import Input, Dense  # 신경망 레이어 import
from tensorflow.keras.models import Model         # 모델 생성 함수 import

# MNIST 데이터셋 로드: 손글씨 이미지 데이터, 28x28 크기의 이미지
mnist = tf.keras.datasets.mnist
(x_train, _), (x_test, _) = mnist.load_data()  # 학습용, 테스트용 데이터 분리 (라벨은 사용하지 않음)

# 데이터 전처리: 픽셀 값을 0~1 사이로 정규화 후, 1차원 벡터(784 차원)로 변환
x_train = x_train.astype('float32') / 255.  # 학습 데이터 정규화
x_test  = x_test.astype('float32') / 255.    # 테스트 데이터 정규화
x_train = x_train.reshape((len(x_train), -1)) # 학습 데이터를 1차원 벡터로 변환 (28*28=784)
x_test  = x_test.reshape((len(x_test), -1))   # 테스트 데이터를 1차원 벡터로 변환

# 입력 차원과 잠재 공간 차원 설정
input_dim = x_train.shape[1]   # 입력 차원: 784 (28x28)
encoding_dim = 64              # 잠재 공간(인코딩) 차원: 64

# 인코더-디코더 구조로 오토인코더 모델 구성
input_img = Input(shape=(input_dim,))       # 입력 레이어 정의
# 인코더: 784차원 입력 -> 128 노드 -> 64 노드로 차원 축소
encoded = Dense(128, activation='relu')(input_img)  # 첫 번째 은닉층, ReLU 활성화
encoded = Dense(encoding_dim, activation='relu')(encoded)  # 두 번째 은닉층, 64차원으로 축소
# 디코더: 잠재공간(64차원) -> 128 노드 -> 784 노드로 원본 복원
decoded = Dense(128, activation='relu')(encoded)    # 첫 번째 디코더 층, ReLU 활성화
decoded = Dense(input_dim, activation='sigmoid')(decoded)  # 두 번째 디코더 층, sigmoid 활성화로 원본 복원

# 전체 오토인코더 모델 정의: 입력부터 복원된 출력까지 연결
autoencoder = Model(input_img, decoded)
# 모델 컴파일: adam 옵티마이저, 이진 교차 엔트로피 손실 함수 사용
autoencoder.compile(optimizer='adam', loss='binary_crossentropy')

# 오토인코더 학습: x_train을 입력과 출력으로 사용 (비지도 학습)
autoencoder.fit(x_train, x_train,
                epochs=10,            # 10 에포크 동안 학습
                batch_size=256,       # 배치 크기 256
                shuffle=True,         # 데이터를 섞어가며 학습
                validation_data=(x_test, x_test))  # 검증 데이터 지정

# 학습된 오토인코더로 테스트 데이터 복원 (예측)
decoded_imgs = autoencoder.predict(x_test)

# 원본 이미지와 복원된 이미지 비교 시각화 (10개 샘플)
n = 10  # 비교할 샘플 수
plt.figure(figsize=(20, 4))  # 시각화할 그림 크기 설정
for i in range(n):
    # 원본 이미지 출력
    ax = plt.subplot(2, n, i + 1)  # 2행 n열 subplot 생성, 첫 번째 행 사용
    plt.imshow(x_test[i].reshape(28, 28), cmap='gray')  # 1차원 벡터를 28x28 이미지로 재구성 후 출력
    plt.title('원본')       # 제목 설정
    plt.axis('off')         # 축 숨김
    
    # 복원 이미지 출력
    ax = plt.subplot(2, n, i + 1 + n)  # 두 번째 행에 subplot 생성
    plt.imshow(decoded_imgs[i].reshape(28, 28), cmap='gray')  # 복원된 이미지를 28x28으로 재구성 후 출력
    plt.title('복원')       # 제목 설정
    plt.axis('off')         # 축 숨김
plt.show()  # 모든 서브플롯을 포함한 그림 출력
```

```python
# =============================================================================
# 3. 연관 규칙 학습 (Association Rule Learning) 실습: Apriori, FP-Growth
# =============================================================================
```

```python
import pandas as pd  # 데이터 프레임 처리를 위한 pandas import
from mlxtend.preprocessing import TransactionEncoder  # 거래 데이터를 one-hot 인코딩하기 위한 클래스 import
from mlxtend.frequent_patterns import apriori, association_rules  # Apriori 알고리즘과 연관 규칙 추출 함수 import

# 거래 데이터 예시: 각 리스트는 한 거래에서 구매된 품목들을 나타냄
dataset = [
    ['우유', '빵', '버터'],      # 첫 번째 거래
    ['빵', '계란'],            # 두 번째 거래
    ['우유', '빵', '계란', '버터'],  # 세 번째 거래
    ['빵', '버터'],            # 네 번째 거래
    ['우유', '계란']           # 다섯 번째 거래
]

# TransactionEncoder를 사용하여 거래 데이터를 one-hot 인코딩함
te = TransactionEncoder()        # TransactionEncoder 객체 생성
te_array = te.fit(dataset).transform(dataset)  # 데이터셋에 맞춰 학습 후 one-hot 인코딩 수행
df = pd.DataFrame(te_array, columns=te.columns_)  # 인코딩된 데이터를 DataFrame으로 변환

# Apriori 알고리즘 적용: 최소 지지도(min_support)를 0.4로 설정하여 빈발 항목 집합 도출
frequent_itemsets = apriori(df, min_support=0.4, use_colnames=True)
print("빈발 항목 집합:\n", frequent_itemsets)  # 도출된 빈발 항목 집합 출력

# 연관 규칙 추출: 신뢰도(min_threshold)를 0.7로 설정하여 의미 있는 규칙 도출
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)
print("\n연관 규칙:\n", rules)  # 추출된 연관 규칙 출력
```

```python
# GMM을 사용하여 각 데이터 포인트의 로그 확률을 계산하여 이상치 판단
# 낮은 로그 확률 값을 가진 데이터가 이상치로 간주됨

# 각 데이터 포인트에 대한 로그 확률 계산 (score_samples 메서드 사용)
log_prob = gmm.score_samples(X)

# 하위 5% 로그 확률 값을 이상치 임계치로 설정 (np.percentile 사용)
threshold = np.percentile(log_prob, 5)
# 로그 확률이 임계치보다 낮은 데이터를 이상치로 판단
anomalies = X[log_prob < threshold]

# 전체 데이터와 이상치를 시각화
plt.figure(figsize=(6, 4))
plt.scatter(X[:, 0], X[:, 1],  # 전체 데이터 산점도 (정상 데이터)
            c='blue',         # 색상 파란색 지정
            s=50,             # 점의 크기 설정
            label='정상 데이터')  # 범례에 '정상 데이터'로 표시
plt.scatter(anomalies[:, 0], anomalies[:, 1],  # 이상치 데이터 산점도
            c='red',        # 색상 빨간색 지정
            s=50,           # 점의 크기 설정
            label='이상치')   # 범례에 '이상치'로 표시
plt.title('Gaussian Mixture Model을 이용한 이상치 탐지')  # 제목 설정
plt.xlabel('특징 1')            # x축 레이블 설정
plt.ylabel('특징 2')            # y축 레이블 설정
plt.legend()                   # 범례 표시
plt.show()                     # 그래프 출력
```
