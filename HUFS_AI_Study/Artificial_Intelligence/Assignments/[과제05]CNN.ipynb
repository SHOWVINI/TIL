## CNN을 이해하기 위한 사전 개념

### 1. 이미지 데이터의 구조
- **픽셀**: 디지털 이미지는 픽셀로 구성되며, 각 픽셀은 밝기 혹은 색상 정보를 담고 있음
- **채널**: 흑백 이미지는 1채널, 컬러 이미지는 일반적으로 R, G, B의 3채널로 구성됨
- **해상도**: 이미지의 가로 × 세로 크기로, 예: 224×224×3

### 2. 인공신경망 (ANN: Artificial Neural Network)
- **퍼셉트론**: 가장 기본적인 뉴런 단위로 입력값을 받아 가중치를 곱하고, 편향을 더한 뒤 비선형 함수로 출력
- **완전 연결층(Dense layer)**: 모든 입력 뉴런이 모든 출력 뉴런과 연결된 구조
- **한계점**: 이미지처럼 고차원 입력을 처리할 때 파라미터 수가 급증하며, 위치 정보를 반영하지 못하는 한계 존재

### 3. 지역적 연결성과 가중치 공유 개념
- **지역적 수용 영역(Receptive Field)**: 특정 뉴런이 입력의 일부분만 바라보는 구조
- **가중치 공유**: 동일한 필터(커널)를 여러 위치에 적용함으로써 연산 효율성과 특성 추출 가능

## CNN의 기초 개념

### 1. 합성곱 연산(Convolution)
- **정의**: 필터(커널)를 이미지에 슬라이딩하면서 각 영역과의 내적을 통해 특징을 추출하는 연산
- **커널**: 보통 3×3, 5×5 등의 행렬로 구성되며, 학습을 통해 최적화됨
- **특징 맵(Feature Map)**: 필터를 적용한 결과로 생성되는 출력값

### 2. 활성화 함수 (Activation Function)
- **ReLU (Rectified Linear Unit)**: 음수를 0으로 만들고 양수는 그대로 출력하는 함수로, 비선형성 부여
- **역전파에 유리**하며, 학습 속도 향상에 기여함

### 3. 풀링(Pooling)
- **목적**: 특징 맵의 공간적 크기를 줄이고, 중요한 정보만 남기기 위함
- **Max Pooling**: 일정 영역에서 가장 큰 값만 선택
- **Average Pooling**: 일정 영역의 평균값을 선택
- **장점**: 연산량 감소, 과적합 방지, 위치 불변성 제공

### 4. 패딩(Padding)과 스트라이드(Stride)
- **패딩(Padding)**: 입력 가장자리에 0을 추가하여 출력 크기를 조절하거나 정보 손실을 방지
- **스트라이드(Stride)**: 커널이 이동하는 간격을 의미하며, 보통 1 또는 2 사용

### 5. 구조적 특징
- **CNN은 여러 층의 합성곱층, 활성화 함수, 풀링층을 반복적으로 쌓는 구조**
- **마지막에는 완전연결층(Dense Layer)이 존재하며, 최종 분류나 회귀 수행**

## CNN의 심화 개념

### 1. 다양한 커널과 깊이 있는 표현
- **다채널 입력**: 각 필터는 입력의 모든 채널에 대해 각각의 가중치를 학습하며, 결과는 하나의 출력 채널로 통합
- **다수의 필터 사용**: 다양한 패턴(모서리, 곡선 등)을 탐지하기 위해 여러 필터를 사용함

### 2. CNN 아키텍처
- **LeNet-5**: 초창기 CNN 구조로, 숫자 인식에 사용됨
- **AlexNet**: ReLU 도입, GPU 병렬처리, 드롭아웃 등으로 큰 성능 향상
- **VGGNet**: 3×3의 작은 필터를 깊게 쌓아 성능 향상
- **ResNet**: 잔차 연결(Skip Connection)을 도입하여 깊은 네트워크 학습 가능

### 3. 학습 기법
- **드롭아웃(Dropout)**: 학습 중 일부 뉴런을 임의로 제거하여 과적합 방지
- **배치 정규화(Batch Normalization)**: 입력 분포를 정규화하여 학습 속도 향상 및 안정화
- **전이 학습(Transfer Learning)**: 사전학습된 CNN 모델을 가져와 재학습하여 적은 데이터로도 높은 성능 확보 가능

### 4. 시각적 해석
- **필터 시각화**: 학습된 필터의 가중치를 이미지화하여 네트워크가 어떤 특징에 민감한지 확인
- **특징 맵 시각화**: 각 층의 출력값을 시각화하여 추출된 정보 흐름을 파악 가능

### 5. 한계점과 대응 방안
- **데이터 요구량 큼**: 많은 데이터가 필요하며, 이를 위해 증강(Augmentation) 기법 활용
- **계산 자원 소모 큼**: 연산이 많기 때문에 GPU 사용 필수
- **해석 어려움**: 왜 특정 결과가 나왔는지 설명 어려움, 시각화 기법 및 XAI 연구 진행 중

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class SimpleCNN(nn.Module):
    def __init__(self):
        super(SimpleCNN, self).__init__()
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3, padding=1)
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)
        self.conv2 = nn.Conv2d(16, 32, 3, padding=1)
        self.fc1 = nn.Linear(32 * 8 * 8, 128)
        self.fc2 = nn.Linear(128, 10)

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))  # (B, 16, 16, 16)
        x = self.pool(F.relu(self.conv2(x)))  # (B, 32, 8, 8)
        x = x.view(-1, 32 * 8 * 8)  # Flatten
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x

# 모델 생성 및 샘플 입력 테스트
model = SimpleCNN()
input_sample = torch.randn(1, 3, 32, 32)  # (Batch, Channel, Height, Width)
output = model(input_sample)
print(output.shape)  # torch.Size([1, 10])
```
