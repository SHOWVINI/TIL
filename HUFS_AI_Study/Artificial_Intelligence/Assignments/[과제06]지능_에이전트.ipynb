# 지능 에이전트

---

## 1. 지능 에이전트란 무엇인가?

지능 에이전트란,

**스스로 주변 세계를 관찰하고(센싱)**,

**관찰한 내용을 해석해 상황을 이해하고(인식)**,

**상황에 맞는 최선의 행동을 결정하고(추론 및 계획)**,

**직접 행동을 수행하여(액추에이션)**

**목표를 달성하는 자율 시스템**을 의미함.

즉, 단순히 반응하는 것이 아니라, **판단하고 계획하고 실행하는 전체 사이클**을 스스로 수행해야 비로소 "지능적 에이전트"라고 부를 수 있음.

---

## 2. 지능 에이전트의 5대 구성요소

| 구성 요소 | 구체적 역할 | 왜 필요한가? |
| --- | --- | --- |
| 센싱(Sensing) | 주변 환경 정보를 수집 | 세상을 이해하려면 관찰이 선행되어야 함 |
| 인식(Perception) | 수집한 신호를 해석해 의미 있는 정보로 변환 | 관찰만 해서는 의미를 모르기 때문에 해석 필요 |
| 추론 및 계획(Reasoning & Planning) | 상황 분석 후 최적의 행동 결정 | 단순 반응이 아니라 목표 지향적 행동을 위해 필수 |
| 행동 실행(Actuation) | 계획한 행동을 물리적으로 수행 | 생각만 하면 소용없고, 실제 행동이 필요 |
| 지식 베이스(Knowledge Base) | 세상에 대한 추가 지식 저장 및 참고 | 과거 경험이나 외부 지식을 기반으로 더 똑똑하게 판단 가능 |

---

### 이해를 돕는 구체적 예시

> 자율주행차를 예로 들면:
> 
> - **센싱**: 카메라·라이다로 도로 상황을 실시간 촬영
> - **인식**: "앞에 보행자가 있다" 인식
> - **추론 및 계획**: "브레이크를 밟아야 한다" 결정
> - **행동 실행**: 실제로 제동
> - **지식 베이스**: "보행자 우선" 교통 규칙을 학습한 정보 저장

## 3. 지능 에이전트의 분류 (구조별 특징)

| 유형 | 특징 | 장단점 |
| --- | --- | --- |
| 반응형 에이전트 | 즉각적인 반응만 수행 (상황 → 바로 행동) | 빠르지만 복잡한 목표 처리 불가 |
| 목표 지향형 에이전트 | 명확한 목표를 설정하고 그에 맞는 행동 계획 | 목표 달성 능력↑, 계산 복잡도↑ |
| 학습형 에이전트 | 경험을 통해 스스로 성능을 개선 | 유연성↑, 데이터와 시간 소모 발생 |
| 사회적 에이전트 | 다른 에이전트와 협력 또는 경쟁 | 복잡한 문제 해결 가능, 통신 비용 발생 |

### 이해를 돕는 구체적 예시

> 반응형: 모기(자극에 바로 반응)목표 지향형: GPS 네비게이션(목적지 설정 후 경로 탐색)학습형: 알파고(바둑 경험을 통해 스스로 실력 향상)사회적 에이전트: 자율주행차 집단 주행(차들이 서로 조정하며 군집 주행)

## 4. 지능 에이전트 개발 과정 (Life Cycle)

**하나의 지능 에이전트를 만들기 위해 다음 단계를 순차적으로 수행해야 함**

| 단계 | 세부 내용 |
| --- | --- |
| 문제 정의(Problem Definition) | 해결할 과업과 성공 기준을 명확히 설정 |
| 환경 모델링(Environment Modeling) | 에이전트가 관찰하고 조작할 수 있는 세계를 정의 |
| 구조 설계(Agent Architecture Design) | 센서, 인식기, 추론기, 액추에이터, 지식 베이스 연결 설계 |
| 알고리즘 선택(Algorithm Design) | 강화학습, 규칙 기반 추론, 최적화 기법 등 결정 |
| 학습 및 평가(Learning and Evaluation) | 데이터 기반 모델 훈련 및 성능 검증 |
| 디플로이 및 운용(Deployment and Operation) | 실제 환경에 투입하여 지속적인 모니터링 및 개선 |

> 5. 앱으로 배포 및 사용자 피드백 반영하여 업그레이드

# 정리

- 지능 에이전트는 **관찰→이해→판단→행동**의 전체 과정을 자율적으로 수행하는 시스템임.
- 센서, 인식, 추론, 액추에이터, 지식베이스라는 명확한 구성요소를 지님.
- 설계, 학습, 배포라는 일련의 개발 과정을 거쳐 실세계 문제 해결에 기여함.
- 자율성·적응성·목표 지향성은 지능 에이전트를 평가하는 핵심 기준임.

```python
# 간단한 지능 에이전트 예시 (Python)

# 문제

#     센서: 랜덤으로 숫자(0~100) 감지
#     인식: 감지한 숫자가 어떤 범위에 속하는지 분류
#     추론: 숫자가 "낮음", "적당함", "높음" 중 어디에 해당하는지 결정
#     행동: 그에 맞는 메시지 출력

import random

# [1] 센싱(Sensing): 외부 환경(랜덤 수) 감지
def sense_environment():
    number = random.randint(0, 100)
    return number

# [2] 인식(Perception): 감지한 수를 해석
def perceive(number):
    if number < 30:
        return "low"
    elif number < 70:
        return "medium"
    else:
        return "high"

# [3] 추론 및 계획(Reasoning & Planning): 상태에 따라 행동 결정
def plan_action(state):
    if state == "low":
        return "너무 낮습니다. 주의가 필요합니다."
    elif state == "medium":
        return "적당한 수치입니다. 계속 유지하세요."
    else:
        return "수치가 높습니다. 조치를 고려하세요."

# [4] 행동 실행(Actuation): 결정한 행동을 수행
def act(message):
    print(f"[에이전트 메시지] {message}")

# [5] 에이전트 전체 흐름
def simple_agent():
    number = sense_environment()
    print(f"[센싱] 감지된 숫자: {number}")
    
    state = perceive(number)
    print(f"[인식] 현재 상태: {state}")
    
    message = plan_action(state)
    act(message)

# 실행
if __name__ == "__main__":
    simple_agent()
```
