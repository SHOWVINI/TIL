# 1

# 기계학습의 종류와 머신러닝의 지도학습 모델

: 기계학습의 주요 종류와, 그중 머신러닝 지도학습에 해당하는 대표적인 모델과 기술을 정리.

---

## 1. 인공지능을 학습할 때 알아야 할 주요 개념

### 1.1 인공지능(AI)의 개념
인공지능(AI)은 인간의 지적 능력을 모방하거나 확장하기 위해 만들어진 알고리즘과 시스템을 아우르는 광범위한 학문 분야임. 기계학습, 딥러닝 등의 하위 분야가 있으며, 이러한 기술은 데이터에서 패턴을 찾고 예측하거나 의사결정을 자동화함.

### 1.2 기계학습(머신러닝, ML)
기계학습은 인공지능을 실현하기 위해 컴퓨터가 데이터로부터 스스로 학습하도록 하는 알고리즘과 방법론을 의미함. 전통적으로 다음 세 가지 대표적인 학습 방식으로 구분함.
- **지도학습(Supervised Learning)**: 입력 데이터와 그에 대응하는 정답(레이블)이 함께 주어지는 환경에서 학습함.
- **비지도학습(Unsupervised Learning)**: 정답(레이블) 없이 데이터 자체의 구조나 패턴을 찾아냄.
- **강화학습(Reinforcement Learning)**: 에이전트가 환경과 상호작용하며 보상(Reward)을 최대화하는 방향으로 학습함.

### 1.3 딥러닝(Deep Learning)
딥러닝은 인공신경망(Artificial Neural Network)을 심층화하여 많은 은닉층(Hidden Layer)을 쌓은 형태의 모델을 지칭함. 이미지 분류, 음성 인식, 자연어 처리 등 여러 분야에서 우수한 성능을 보이는 핵심 기술임. 주로 대용량 데이터와 GPU(또는 TPU)와 같은 고성능 하드웨어를 사용하여 모델을 훈련함.

### 1.4 기타 중요 기술
- **특징 공학(Feature Engineering)**: 모델이 좋은 예측을 할 수 있도록 중요한 특징(Feature)을 추출하고 전처리하는 과정임.
- **모델 해석 기법(Model Interpretability)**: 예측 결과가 어떻게 도출되었는지 이해하기 위한 방법으로, LIME, SHAP 등이 활용됨.
- **성능 평가와 하이퍼파라미터 튜닝**: 학습된 모델의 성능을 적절하게 평가하고, 하이퍼파라미터를 최적화하여 모델 성능을 극대화하는 과정임.

---

## 2. 머신러닝 지도학습 개요

머신러닝 지도학습은 **입력 데이터**와 해당 **정답(레이블)**이 함께 주어지는 환경에서 학습하는 방법임. 지도학습 모델들은 입력과 레이블 간의 관계를 학습하고, 이를 통해 새로운 데이터에 대해 올바른 결과를 예측함.

지도학습은 분류(Classification)와 회귀(Regression)라는 두 가지 큰 범주로 나뉨.

**분류(Classification)**  
   - 출력이 특정 카테고리나 클래스로 분류되는 문제에 적용됨  
   - 예: 스팸 메일 분류, 이미지를 보고 사람/동물을 분류 등  

**회귀(Regression)**  
   - 출력이 연속적인 수치값을 갖는 문제에 적용됨  
   - 예: 주택 가격 예측, 주가 예측 등  

---


## 3. 대표적인 지도학습 모델

### 3.1 선형 모델 (Linear Models)

1. **선형회귀(Linear Regression)**
   - 입력 변수(특징)와 출력 변수(타깃) 사이의 선형 관계를 가정하는 회귀 모델임  
   - 가중치(회귀계수)와 절편을 통해 데이터를 가장 잘 설명하는 직선(또는 초평면)을 찾음  
   - 잔차제곱합(RSS)을 최소화하기 위해 경사하강법(Gradient Descent)이나 정규방정식(Normal Equation) 등을 사용하여 모델 파라미터를 추정함  
   - 해석이 간단하고 구현이 쉬우며, 선형 관계가 강한 데이터에서 우수한 성능을 보임  

2. **로지스틱 회귀(Logistic Regression)**
   - 출력이 이진 범주(0 또는 1)인 이진분류(Classification)에 주로 사용됨  
   - 이름은 회귀이지만, 실제로는 분류 문제에 특화된 모델임  
   - 시그모이드(로지스틱) 함수를 통해 클래스에 속할 확률을 추정함  
   - 분류 임계값(Threshold)을 조정하여 모델의 민감도(Recall)나 정밀도(Precision)를 조절할 수 있음  

3. **정규화 선형 모델(Regularized Linear Models)**
   - **릿지(Ridge) 회귀**와 **라쏘(Lasso) 회귀**가 대표적임  
   - 모델의 복잡도를 조절하기 위해 가중치에 규제항(Regularization)을 추가함  
   - **릿지**는 \( L_2 \) 규제, **라쏘**는 \( L_1 \) 규제를 사용함  
   - 과적합을 방지하고 특징 선택(Feature Selection)에 도움을 줌  

### 3.2 비선형 모델 (Non-Linear Models)

1. **의사결정나무(Decision Tree)**
   - 분류와 회귀 모두에 사용 가능하며, 트리 형태로 의사결정을 시각화하여 이해하기 쉬움  
   - 트리가 깊어질수록 과적합(Overfitting)이 발생할 수 있으며, 이를 막기 위해 가지치기(Pruning) 등 기법을 활용함  
   - 분산이 큰 데이터나 잡음이 많은 데이터에 대해서는 단독 모델만으로는 성능이 제한적일 수 있음  

2. **서포트 벡터 머신(Support Vector Machine, SVM)**
   - 데이터들을 고차원 공간으로 매핑하여 최대 폭(Margin)의 결정경계(Hyperplane)를 학습함  
   - 커널 트릭(Kernel Trick)을 이용해 비선형 문제도 분류와 회귀에 적용 가능함  
   - 데이터 차원이 높거나 복잡한 경우에도 성능이 좋은 편이지만, 하이퍼파라미터(C, 커널 파라미터) 튜닝이 중요함  

3. **k-최근접 이웃(k-Nearest Neighbors, k-NN)**
   - 예측 시점에 **가장 가까운 k개 이웃의 라벨**을 참고하여 결과를 결정함  
   - 모델 학습 단계가 거의 없으나, 예측 단계에서 모든 학습 데이터를 탐색해야 하므로 계산 비용이 큼  
   - k값이나 거리 측정 방법(유클리디안 거리, 맨해튼 거리 등)에 따라 성능이 크게 좌우됨  

4. **나이브 베이즈(Naive Bayes)**
   - 베이즈 정리를 이용하여 각 특징이 독립적이라고 가정(나이브 가정)함  
   - 가정이 단순하지만 여러 분류 문제(예: 스팸 필터링)에서 효과적임  
   - 조건부 확률을 추정하기 쉽고, 훈련 속도가 빠름  

5. **가우시안 프로세스 회귀/분류(Gaussian Process for Regression/Classification)**
   - 데이터를 확률적 관점에서 처리하여, 함수 자체를 확률분포로 다룸  
   - 정확도가 높은 편이나, 데이터가 많아질수록 계산 비용이 급격히 증가함  

### 3.3 앙상블 학습(Ensemble Learning)

앙상블 학습은 여러 모델을 조합함으로써 **개별 모델**이 내는 예측 결과보다 더 높은 성능을 도출하려는 기법임.

1. **배깅(Bagging)**
   - 학습 데이터를 무작위로 여러 세트로 나누어(부트스트래핑), 각각의 모델을 독립적으로 학습시킴  
   - 예측 시에는 개별 모델의 결과를 평균(회귀) 또는 다수결(분류)로 최종 결정함  
   - 대표적인 예로 **랜덤 포레스트(Random Forest)**가 있으며, 다수의 결정나무를 사용하여 개별 나무의 편향은 낮추고 분산을 줄임  

2. **부스팅(Boosting)**
   - 순차적으로 모델을 학습하여, 이전 모델이 만든 오류를 다음 모델이 개선하도록 하는 방식임  
   - 대표적인 알고리즘으로 **AdaBoost**, **Gradient Boosting**, **XGBoost**, **LightGBM**, **CatBoost** 등이 있음  
   - 과적합 방지를 위해 학습률(Learning Rate), 트리의 깊이, 정규화 파라미터 등을 적절히 조절해야 함  

3. **스태킹(Stacking)**
   - 서로 다른 알고리즘(또는 하이퍼파라미터가 다른 같은 알고리즘)에서 예측한 결과를 최종 메타 모델(Meta-Model)에 입력하여 다시 학습시킴  
   - 베이스 모델들의 예측 결과가 입력이 되므로, 메타 모델은 이 결과들을 종합하여 최종 예측을 수행함  
   - 개별 모델 간의 상관관계를 잘 이용하면 높은 성능을 얻을 수 있으나, 튜닝이 복잡할 수 있음  

### 3.4 신경망 기반 모델 (Neural Network-based Models)

1. **다층 퍼셉트론(MLP, Multi-Layer Perceptron)**
   - 가장 기본적인 형태의 인공신경망으로, 은닉층(Hidden Layer)이 여러 개 있는 형태임  
   - 회귀와 분류 문제 모두에 적용 가능하며, 범용성이 높음  
   - 은닉층의 개수와 뉴런 수 등 하이퍼파라미터 설정이 성능에 영향을 줌  

2. **합성곱 신경망(CNN, Convolutional Neural Network)**
   - 주로 이미지 처리 분야에 특화된 모델이지만, 1D·2D·3D 합성곱을 통해 시계열이나 텍스트 처리에도 적용 가능함  
   - 합성곱(Convolution)과 풀링(Pooling) 연산을 통해 특징을 자동으로 추출함  

3. **순환 신경망(RNN, Recurrent Neural Network)**
   - 시퀀스(Sequence) 데이터 처리에 적합함 (예: 자연어 처리, 시계열 예측 등)  
   - LSTM(Long Short-Term Memory)이나 GRU(Gated Recurrent Unit) 같은 구조를 통해 장기 종속성(Long-term Dependency)을 학습함  

**딥러닝(Deep Learning) 기반 모델**은 일반적으로 **대량의 데이터**와 **고성능 컴퓨팅 자원**을 필요로 하며, 전통적인 머신러닝 모델보다 높은 표현력(Expressive Power)을 가짐  

---

## 4. 지도학습의 필수 기술과 고려사항

### 4.1 데이터 전처리 및 특징 공학(Feature Engineering)

- **결측치 처리**: 제거, 대치(평균값·중앙값·최빈값 등)  
- **이상치 제거**: 통계적 방법(IQR, Z-score 등)으로 극단값을 탐지 후 처리  
- **인코딩**: 범주형 변수 처리(원-핫 인코딩, 라벨 인코딩 등)  
- **스케일링**: Min-Max 스케일링, 표준화(Standardization) 등을 통해 특징 값의 범위를 일관성 있게 맞춤
- **파생변수**: 도메인 지식을 이용한 파생변수 생성 엔지니어링

### 4.2 모델 평가 및 성능 개선

1. **교차검증(Cross-Validation)**
   - 데이터셋을 여러 분할로 나누어 학습과 검증을 반복 수행함  
   - 일반화 성능을 더욱 신뢰도 높게 추정할 수 있음  
   - 대표적으로 K-겹 교차검증(K-Fold)이 많이 사용됨  

2. **평가 지표(Evaluation Metrics)**
   - **분류**: 정확도(Accuracy), 정밀도(Precision), 재현율(Recall), F1 점수, ROC-AUC, 혼동행렬(Confusion Matrix) 등  
   - **회귀**: MSE(평균제곱오차), MAE(평균절대오차), RMSE(제곱근 평균제곱오차), R² 등  
   - 문제의 특성, 오분류 비용, 불균형 데이터 여부 등을 종합적으로 고려해 지표를 선택함  

3. **하이퍼파라미터 튜닝(Hyperparameter Tuning)**
   - 그리드 서치(Grid Search), 랜덤 서치(Random Search), 베이지안 최적화(Bayesian Optimization) 등을 이용함  
   - 학습률, 정규화 계수, 트리 깊이 등 모델별로 중요한 파라미터를 조정하여 최적의 성능을 추구함  

4. **모델 해석(Explainability)**
   - 모델 예측 결과를 이해하거나 신뢰도를 높이기 위해, SHAP, LIME 등의 해석 도구를 사용할 수 있음  
   - 블랙박스 모델인 딥러닝이나 앙상블 모델의 경우에도 해석 기법을 통해 의사결정 과정을 파악할 수 있음

# 2 

# 머신러닝의 대표적인 비지도학습 모델 및 기술 정리

---

데이터 특성이나 문제 유형에 따라 기법 선택 및 파라미터 설정이 달라지므로, 사전 분석과 충분한 검증이 필수적.

## 1. 군집화(Clustering) 기법

1. **K-means**  
   
   - 각 군집의 중심(centroid)을 반복적으로 업데이트하며 군집을 형성하는 알고리즘임  
   - 사전에 군집 개수 K를 설정해야 하며, 유클리디언 거리 등을 통해 중심으로부터의 오차제곱합(SSE)을 최소화함  
   - 단순하고 계산량이 적으나, 초깃값 및 K 값 선정에 따라 결과 품질이 크게 달라질 수 있음  

      - **개념**: 
         - 데이터를 미리 정한 K개의 그룹(군집)으로 나누는 알고리즘임.

      - **방식**:
         - 초기 단계에서는 K개의 중심점(centroid)을 무작위로 선정함.
         - 각 데이터 포인트를 가장 가까운 중심점에 할당함.
         - 군집 내 데이터의 평균값을 새 중심점으로 갱신함.
         - 수렴할 때까지 반복하는 방식임.

      - **특징**:
         - 단순하고 계산량이 적음.
         - 군집 개수 K를 사전에 지정함.
         - 초기 중심점 선정 및 K 값에 따라 결과 품질이 달라질 수 있음.

2. **계층적 군집화(Hierarchical Clustering)**  

   - 데이터 간 거리나 유사도를 기반으로 트리 형태(덴드로그램)를 생성함  
   - **상향식(Agglomerative)** 방식은 개별 데이터 포인트에서 시작해 가장 유사한 군집끼리 병합하고,  
     **하향식(Divisive)** 방식은 전체 데이터를 하나의 군집으로 본 뒤 점차 세분화해 나가는 접근임  
   - 덴드로그램을 통해 군집 개수를 시각적으로 결정할 수 있음  
   - 대규모 데이터에 대해 계산 복잡도가 높은 단점이 있음  

      - **개념**: 
         - 데이터 간의 유사도(거리)를 기준으로 단계적으로 군집을 합치거나 분할하는 방식임.

      - **방식**:
         - **상향식(Agglomerative)**: 각 데이터를 개별 군집으로 시작하여, 가장 유사한 두 군집을 합치는 방식임.
         - **하향식(Divisive)**: 전체 데이터를 하나의 군집으로 보고, 점진적으로 세분화하는 방식임.

      - **특징**:
         - 덴드로그램(트리 구조)을 통해 군집 개수를 시각적으로 결정할 수 있음.
         - 대규모 데이터에 대해서는 계산 복잡도가 높은 단점이 있음.


3. **DBSCAN (Density-Based Spatial Clustering of Applications with Noise)**  
   - 임의의 점 주변 밀도가 충분히 높은 경우(ε 반경 내 점의 개수가 일정 기준 이상), 해당 점을 ‘핵심 점(core point)’으로 분류하고 연결된 핵심 점들의 집합을 군집으로 정의함  
   - 사전에 군집 개수를 설정할 필요가 없음  
   - 이상치(outlier)나 노이즈를 군집과 분리하여 탐지할 수 있음  
   - ε과 최소 점 개수(minPts) 설정이 까다로울 수 있음  

      - **개념**: 
         - 데이터 포인트 주변의 밀도를 기반으로 군집을 정의하는 알고리즘임.

      - **방식**:
         - 임의의 점을 기준으로 ε 반경 내에 일정 수 이상의 점이 존재하면 해당 점을 핵심 점(core point)으로 판단함.
         - 연결된 핵심 점들을 하나의 군집으로 묶음.
         - 기준에 미달하는 점들은 이상치(노이즈)로 분류함.

      - **특징**:
         - 군집 개수를 사전에 설정할 필요가 없음.
         - 이상치 탐지에 유리함.
         - ε과 최소 점 개수(minPts) 설정이 중요함.

4. **가우시안 혼합 모델(Gaussian Mixture Model, GMM)**  
   - 여러 개의 가우시안(정규 분포)이 혼합된 것으로 데이터를 설명하고, 각 분포의 파라미터를 추정함  
   - Expectation-Maximization(EM) 알고리즘으로 각 데이터가 어느 분포에 속할 확률을 계산함  
   - K-means보다 복잡한 데이터 분포를 모형화하는 데 유리함  
   - 가우시안 가정이 실제 데이터 특성과 맞지 않으면 성능이 떨어질 수 있음  

      - **개념**: 
         - 데이터가 여러 개의 가우시안(정규) 분포의 혼합으로 이루어졌다고 가정하는 모델임.

      - **방식**:
         - EM(Expectation-Maximization) 알고리즘을 사용하여 각 데이터 포인트가 어느 분포에 속할 확률을 추정함.
         - 분포의 평균, 분산 등의 파라미터를 반복적으로 업데이트함.

      - **특징**:
         - 복잡한 데이터 분포 모형화에 유리함.
         - 가우시안 분포 가정이 실제 데이터와 부합하지 않을 경우 성능 저하 가능성이 있음.

---

## 2. 차원 축소(Dimensionality Reduction) 기법

1. **주성분분석(PCA, Principal Component Analysis)**  
   - 고차원 데이터를 저차원으로 사영(projection)하여 분산(variance)을 최대화하는 방향(주성분)을 찾는 방법임  
   - 선형 변환을 사용하므로, 비선형 구조의 데이터에서는 한계가 있음  
   - 특성 간 상관성을 줄이고 시각화 및 전처리에 유용함  

      - **개념**: 
         - 데이터의 분산을 최대한 보존하는 방향(주성분)을 찾아 고차원 데이터를 저차원으로 사영하는 기법임.
      - **방식**:
         - 데이터의 공분산 행렬 계산 후 고유값분해(또는 SVD)를 수행함.
         - 가장 분산이 큰 주성분들을 선정하여 차원을 축소함.
      - **특징**:
         - 선형 변환 기반임으로 비선형 데이터 구조에서는 한계가 있음.
         - 특성 간 상관성을 제거하고 중요한 정보는 유지함.

2. **오토인코더(Autoencoder)**  
   - 입력을 저차원 잠재 공간(latent space)으로 압축(인코딩)한 뒤 다시 원본 차원으로 복원(디코딩)하는 신경망 구조임  
   - 은닉층의 비선형성을 통해 복잡한 패턴 포착 및 차원 축소가 가능함  
   - 네트워크 구조 및 학습 기법(규제, 드롭아웃 등)에 따라 노이즈 제거와 고유 특성 추출에 효과적임 

      - **개념**: 
         - 입력 데이터를 낮은 차원의 잠재 공간으로 압축한 후, 다시 원본으로 복원하는 신경망 구조임.
      - **방식**:
         - 인코더가 데이터를 압축함.
         - 디코더가 압축된 데이터를 원본과 유사하게 복원하도록 학습함.
      - **특징**:
         - 은닉층의 비선형 활성화 함수를 통해 복잡한 패턴도 포착함.
         - 노이즈 제거 및 이상치 탐지에도 활용 가능함. 

3. **t-SNE (t-distributed Stochastic Neighbor Embedding)**  
   - 고차원 데이터 간 확률적 거리(인접도)를 저차원에서도 유사하게 유지하는 비선형 차원 축소 기법임  
   - 시각화 품질이 우수하여 이미지나 텍스트 임베딩 등을 탐색하는 데 활용도가 높음  
   - 대규모 데이터에서는 계산량이 많고, 퍼플렉시티(perplexity) 등 매개변수 설정에 따라 결과가 좌우됨  

      - **개념**: 
      - 고차원 데이터의 유사도(거리) 정보를 저차원에서도 최대한 보존하도록 변환하는 비선형 차원 축소 기법임.
      - **방식**:
         - 데이터 간의 확률적 유사도를 계산하여 저차원 공간에 반영함.
      - **특징**:
         - 2차원 또는 3차원 시각화에 매우 효과적임.
         - 대규모 데이터에서는 계산 부담이 큼.
         - 퍼플렉시티(perplexity) 등의 매개변수 설정에 따라 결과가 달라짐.

4. **UMAP (Uniform Manifold Approximation and Projection)**  
   - 데이터가 잠재적 매니폴드 구조를 가진다고 가정하고 이를 저차원 공간에 근사화하는 기법임  
   - t-SNE 대비 빠른 속도와 전체 구조(글로벌 구조) 보존 능력이 뛰어남  

      - **개념**: 
         - 고차원 데이터가 잠재적 매니폴드 구조 위에 있다고 가정하고, 이를 저차원에 근사하는 기법임.
      - **방식**:
         - 데이터의 국소 및 글로벌 구조를 고려하여 저차원에 매핑함.
      - **특징**:
         - t-SNE보다 빠른 연산 속도를 가짐.
         - 국소적 구조뿐만 아니라 전체적인(글로벌) 구조도 보존함.

---

## 3. 연관 규칙 학습(Association Rule Learning)

- 대규모 거래 데이터(장바구니 분석)에서 품목 간 동시 발생 패턴을 찾는 데 활용됨  
- 대표 알고리즘으로 Apriori, FP-Growth 등이 있음  
- 지지도(support), 신뢰도(confidence) 등 지표를 사용해 유의미한 규칙을 탐색함  
- 지도학습과 달리 라벨 없이 데이터 간 연관성을 도출함  

---

## 4. 이상치 탐지(Anomaly Detection)

- 대다수 데이터와는 다른 소수의 극단값(이상치, 이상 패턴)을 식별하는 기법임  
- 군집화(군집 중심에서 크게 벗어난 데이터를 이상치로 간주) 또는 확률 분포 기반(혼합 모델에서 확률이 매우 낮은 지점 판별) 접근이 있음  
- 라벨이 없거나 이상치가 매우 드문 상황에서 주로 활용됨  

---

## 5. 기타 주요 비지도학습 분야

1. **토픽 모델링(Topic Modeling)**  
   - 문서 집합에서 잠재 토픽을 찾는 기법임  
   - Latent Dirichlet Allocation(LDA) 등이 대표적이며, 비지도 방식으로 문서 주제를 추론함  

2. **생성 모델(Generative Models) 일부**  
   - GAN(Generative Adversarial Network)은 명확한 라벨 없이 데이터를 생성해 비지도학습에 해당할 수 있음  
   - Variational Autoencoder(VAE) 등 오토인코더 변형 기법도 비지도학습 범주에 포함 가능함

# 3

# 신경망과 퍼셉트론의 기초 및 심화

## 1. 신경망(Neural Network)의 개념

신경망(Neural Network)은 인간의 뇌 구조를 모방하여 만들어진 인공 지능 모델. 

수많은 뉴런(Neuron)으로 구성되어 있으며, 이 뉴런들은 서로 연결되어 복잡한 연산을 수행. 

입력 데이터를 받아 가중치(Weight)와 편향(Bias)을 통해 가공하고, 활성화 함수(Activation Function)를 거쳐 최종 출력을 생성하는 과정을 거침.

- **뉴런(Neuron)**: 입력을 받아 가중치와 곱한 뒤 모든 입력의 총합에 편향을 더해 활성화 함수를 통과시키는 기본 단위.  
- **활성화 함수**: 입력 값의 총합을 특정 범위나 형태로 변환하는 역할을 하는 함수임. 예: 시그모이드 함수, ReLU 함수 등이 존재.  
- **레이어(Layer)**: 여러 뉴런이 모여 층을 이루며, 입력층(Input Layer), 은닉층(Hidden Layer), 출력층(Output Layer)으로 구성되는 구조.

이런 층별 구조를 통해 비선형적이고 복잡한 함수를 근사할 수 있는 능력을 갖추게 됨.

---

## 2. 퍼셉트론(Perceptron)의 기초

퍼셉트론(Perceptron)은 신경망의 기초가 되는 가장 단순한 형태의 모델. 
1957년 프랭크 로젠블라트(Frank Rosenblatt)가 제안한 알고리즘으로, 단층 퍼셉트론(Single-Layer Perceptron)을 이용하면 선형적으로 구분 가능한 문제를 해결할 수 있음.

### 2.1 퍼셉트론 구조

퍼셉트론은 입력 값들의 선형 조합에 활성화 함수를 적용하여 출력을 결정하는 단순 구조.

1. 여러 개의 입력 $x_1, x_2, \dots, x_n$ 을 가중치 $w_1, w_2, \dots, w_n$ 와 곱한 뒤 합산함.  
2. 편향 $b$ 를 더해 최종 합을 구함.  
3. 활성화 함수를 적용해 이 합이 임계값을 넘으면 1(또는 양의 클래스)로, 그렇지 않으면 0(또는 음의 클래스)로 판단하는 것이 기본 개념.

### 2.2 퍼셉트론 학습
퍼셉트론 학습은 오차(Error)가 발생했을 때, 가중치와 편향을 업데이트하는 방식으로 진행됨.

- **업데이트 규칙**:

  $$
  w_i \leftarrow w_i + \Delta w_i
  $$

  $$
  \Delta w_i = \eta (t - o)x_i
  $$

  여기서 $\eta$ 는 학습률(Learning Rate), $t$ 는 목표 출력, $o$ 는 실제 출력, $x_i$ 는 입력 값임.

이 과정을 반복하면서 퍼셉트론은 선형 분류를 위한 최적의 결정 경계를 찾게 됨.

---
## 3. 퍼셉트론에서 신경망으로의 확장

퍼셉트론 한 개로는 단순히 선형적인 문제만 해결 가능. 

XOR 문제 같은 비선형 패턴은 단층 퍼셉트론만으로는 구분할 수 없는 한계가 존재. 

이를 해결하기 위해 은닉층을 추가한 다층 퍼셉트론(Multi-Layer Perceptron, MLP)이 제안.

### 3.1 다층 퍼셉트론(Multi-Layer Perceptron)

- **은닉층(Hidden Layer)** 을 하나 이상 추가해 입력층과 출력층 사이에 존재.  
- 은닉층의 활성화 함수를 통해 비선형성을 도입하여, XOR와 같은 복잡한 문제도 해결 가능.  
- 오차역전파(Backpropagation) 알고리즘을 통해 학습을 진행.

### 3.2 역전파(Backpropagation) 알고리즘

역전파 알고리즘은 출력층에서 계산된 오차를 거꾸로 전파하여 은닉층과 입력층의 가중치를 재조정하는 학습 방법.

1. **순전파(Forward Propagation)**: 입력층에서 출력층까지 순차적으로 가중치 연산 및 활성화 함수를 적용해 예측 결과를 구함.  
2. **오차 계산(Error Calculation)**: 예측 결과와 실제 결과의 차이를 비용 함수(Loss Function)를 통해 계산함.  
3. **역전파(Backpropagation)**: 출력층에서부터 오차 기여도를 계산해 역방향으로 전달하고, 각 가중치와 편향을 미분 정보(Gradient)에 따라 업데이트함.

이 과정을 반복하면서 모델은 오차를 최소화하도록 가중치를 최적화하는 것.

---
## 4. 응용

신경망 구조를 더욱 깊고 복잡하게 만들어 딥러닝(Deep Learning) 분야로 발전. 
최근에는 이미지 처리, 음성 인식, 자연어 처리 등 다양한 영역에 적용되어 뛰어난 성능을 보이는 추세.

### 4.1 심층 신경망(Deep Neural Network, DNN)

- 은닉층을 다수 포함하고, 일반적으로 수백~수천만 개 이상의 파라미터(Parameter)를 갖는 대규모 모델.  
- **학습 난이도** 가 높아지고, 많은 계산 자원과 데이터가 필요하기 때문에 GPU, 분산 학습 기법 등을 활용.

### 4.2 합성곱 신경망(Convolutional Neural Network, CNN)

- 이미지 인식, 영상 처리 분야에서 탁월한 성능을 보이는 구조.  
- 합성곱(Convolution) 레이어와 풀링(Pooling) 레이어를 통해 공간적 특성을 효율적으로 추출하는 것이 특징.

### 4.3 순환 신경망(Recurrent Neural Network, RNN)

- 시간 순서가 있는 시계열 데이터(예: 음성, 텍스트)를 처리하기에 적합.  
- 내부적으로 순환 구조를 갖추고 있어, 이전 상태의 정보를 기억하거나 반영하는 것이 가능.  
- 장·단기 메모리(Long Short-Term Memory, LSTM), 게이트 순환 유닛(GRU) 등으로 발전해 긴 의존 관계도 학습이 가능.

---
## 5. 정리

신경망과 퍼셉트론은 인공 지능 역사에서 매우 중요한 개념. 

퍼셉트론은 선형 분류의 기초를 제공하지만, 은닉층을 추가하고 역전파 알고리즘을 도입함으로써 복잡한 문제 해결이 가능해짐. 

이를 더 확장해 딥러닝 기법이 발전했고, 현재는 이미지, 음성, 자연어 처리 등 다양한 분야에서 활용되고 있는 추세.
# + 다층 퍼셉트론(MLP) 이론 설명

- 정의

다층 퍼셉트론(Multi-Layer Perceptron, MLP)은 은닉층(Hidden Layer) 을 하나 이상 포함한 피드포워드 신경망. 

퍼셉트론을 층으로 쌓은 구조로, 비선형 문제 해결 능력을 갖춘 모델.

    구조: 입력층 → 은닉층1 → 은닉층2 → ... → 출력층

각 층은 선형 변환(linear) → 비선형 활성화 함수(activation) 를 포함함.

- 특징

은닉층이 1개 이상이기 때문에 XOR 같은 비선형 문제도 학습 가능

역전파(backpropagation) 알고리즘을 통해 파라미터(가중치/편향) 학습

보통 활성화 함수로 ReLU, 출력층에서는 Softmax(분류) 또는 Linear(회귀) 사용

## MLP는 DNN인가? 포함 관계 정리

> **MLP(Multi-Layer Perceptron)는 DNN(Deep Neural Network)의 한 종류**  
> 모든 MLP가 DNN은 아니지만, **은닉층이 2개 이상이면 DNN이라고 부를 수 있음**

---

### 정의 비교

| 용어 | 정의 | 핵심 요소 |
|------|------|------------|
| **MLP (Multi-Layer Perceptron)** | 하나 이상의 **은닉층(hidden layer)** 을 가진 완전 연결 신경망 (Fully Connected Neural Network) | 선형 변환 + 활성화 함수 조합, 피드포워드 구조 |
| **DNN (Deep Neural Network)** | 은닉층이 2개 이상인 **깊은** 인공신경망의 총칭 | 깊이(depth)가 중요한 특징, 다양한 아키텍처 포함 |

---

### 포함 관계 구조도

인공신경망 (Artificial Neural Network)

     ├── 단층 퍼셉트론 (Single-layer Perceptron)

     └── 다층 퍼셉트론 (MLP)

          ├── 얕은 신경망 (1개 은닉층)

          └── 심층 신경망 (2개 이상 은닉층) → DNN


> **즉, MLP는 신경망 구조 중 하나이며**,  
> **은닉층이 2개 이상인 MLP는 DNN으로 간주됨**

---

### 📌 예시 비교

| 모델 구조 | MLP | DNN | 비고 |
|------------|-----|-----|------|
| 은닉층 1개 | ✅ | ❌ | 얕은 MLP |
| 은닉층 3개 | ✅ | ✅ | DNN에 해당 |
| CNN, RNN | ❌ | ✅ | MLP는 아니지만 DNN 범주에 포함될 수 있음 |

---

### 📝 정리

- **MLP는 DNN의 한 형태**  
- **DNN은 더 넓은 개념**으로 CNN, RNN, Transformer 등 다양한 구조를 포함  
- DNN이라는 말은 단순히 “은닉층이 깊다”는 의미이며, 꼭 MLP 형태일 필요는 없음

# 딥러닝?
딥러닝(Deep Learning)은 인공신경망(Artificial Neural Network)을 기반으로 하는 기계학습(Machine Learning)의 한 분야로, 데이터로부터 특징(feature)을 자동으로 학습하여 예측이나 분류 등의 과업을 수행하는 기술임.

### 1. 기본 개념

딥러닝은 인간의 뇌 구조를 모방한 다층 신경망(multi-layer neural network)을 통해 입력 데이터로부터 복잡한 패턴과 표현을 추출함. 

기존의 기계학습 기법이 수작업으로 특징을 설계해야 했던 것과 달리, 딥러닝은 계층적 구조를 통해 특징을 자동으로 추출함.

### 2. 구성 요소

- 입력층(Input Layer): 원시 데이터를 입력받는 계층

- 은닉층(Hidden Layers): 입력을 비선형 함수로 반복 변환하며 특징을 추출하는 계층. 깊은 신경망에서는 이 은닉층이 여러 개 존재함

- 출력층(Output Layer): 분류, 회귀 등 목표 과업의 최종 결과를 출력하는 계층

- 가중치(Weights)와 편향(Biases): 입력과 출력 간의 관계를 조정하는 학습 파라미터

- 활성화 함수(Activation Function): 비선형성을 부여하여 복잡한 표현 학습을 가능하게 하는 함수. 대표적으로 ReLU, sigmoid, tanh 등이 있음

### 3. 학습 방법

딥러닝 모델은 손실 함수(loss function)를 최소화하기 위해 역전파(backpropagation) 알고리즘과 경사하강법(gradient descent) 기반의 최적화 기법을 사용함. 

이 과정을 통해 가중치와 편향을 반복적으로 업데이트하며 모델 성능을 향상시킴.

### 4. 주요 모델 종류

- 합성곱 신경망(CNN, Convolutional Neural Network): 이미지 처리에 특화된 구조로, 지역적인 특징 추출에 강점이 있음

- 순환 신경망(RNN, Recurrent Neural Network): 시계열 데이터나 자연어 처리에서 사용되며, 시점 간 정보 흐름을 유지함

- 트랜스포머(Transformer): 자연어 처리 분야에서 사용되는 구조로, 자기 주의 메커니즘(self-attention)을 통해 장기 의존성을 효과적으로 학습함

- 생성 모델(Generative Models): GAN, VAE 등의 모델이 있으며, 데이터를 생성하거나 보강하는 데 활용됨

### 5. 활용 분야

- 음성 인식

- 이미지 분류 및 객체 탐지

- 자연어 처리(번역, 요약, 질의응답)

- 자율주행

- 의료 영상 분석

- 추천 시스템 등 다양한 산업 및 연구 분야에 적용됨
---
## 1. 딥러닝 기초 개념

### (1) 인공신경망 (Artificial Neural Network, ANN)

- **뉴런(Neuron)**  
  생물학적 뉴런을 모사한 간단한 계산 유닛으로, 입력을 가중합(weighted sum) 형태로 받아 활성화 함수를 거쳐 출력값을 생성하는 구조.

- **레이어 구조**  
  - 입력층(Input Layer): 문제의 특징(feature)을 입력받는 층  
  - 은닉층(Hidden Layer): 입력된 정보를 조합하거나 변환하는 층  
  - 출력층(Output Layer): 최종 예측값을 출력하는 층

### (2) 활성화 함수 (Activation Function)

신경망에 비선형성을 도입하여 학습 및 표현 능력을 높이기 위한 함수.

- 예시:  
  - Sigmoid  
  - Tanh  
  - ReLU  
  - Leaky ReLU  
  - Softmax

### (3) 순전파와 역전파

- **순전파 (Feedforward)**  
  입력 데이터가 네트워크를 따라 한 방향으로 전달되어 출력값을 생성하는 과정

- **역전파 (Backpropagation)**  
  손실 함수로 계산된 오차를 네트워크 뒤쪽부터 전파하면서 각 파라미터의 기울기를 계산하고 가중치와 편향을 업데이트하는 과정

## 2. 딥러닝 심화 개념

### (1) 합성곱 신경망 (Convolutional Neural Network, CNN)

이미지 인식, 영상 분류 등에 특화된 신경망 구조.

- 합성곱 층을 통해 지역적 특징 추출  
- 풀링 층을 통해 차원 축소 및 계산 효율 향상

### (2) 순환 신경망 (Recurrent Neural Network, RNN)

시계열 데이터나 자연어 처리에 적합한 구조.

- 은닉 상태를 통해 과거 정보를 기억  
- LSTM, GRU를 통해 장기 의존성 문제를 완화

### (3) 트랜스포머 (Transformer)

어텐션 메커니즘 기반의 구조로, 문맥 정보를 효과적으로 처리하는 모델.

- 입력 전체에 대한 관계를 고려하여 효율적인 학습 가능  
- GPT, BERT 등의 기반 구조

### (4) 기타 심화 개념

- **정규화 기법 (Regularization)**  
  - 드롭아웃(Dropout): 일부 뉴런을 무작위로 제거  
  - 가중치 감쇠(Weight Decay): L2 정규화를 통해 복잡도 감소

- **배치 정규화 (Batch Normalization)**  
  - 각 배치 단위로 입력을 정규화하여 학습 안정화 및 속도 개선

- **스킵 커넥션 (Skip Connection)**  
  - 깊은 네트워크에서 기울기 소실을 방지하기 위해 입력을 다음 층으로 직접 연결하는 구조 (예: ResNet)

## 3. 딥러닝 학습 과정, 손실 함수, 옵티마이저

### (1) 학습(Training) 과정

1. **순전파(Feedforward)**: 예측값 계산  
2. **손실 함수(Loss function)**: 예측값과 실제 값의 차이 계산  
3. **역전파(Backpropagation)**: 오차 전파 및 기울기 계산  
4. **옵티마이저(Optimizer)**: 가중치와 편향을 업데이트  
5. 이 과정을 여러 번 반복(epoch)하면서 오차를 줄이는 방식으로 모델을 학습

### (2) 손실 함수 (Loss Function)

- **MSE (Mean Squared Error)**  
  회귀 문제에서 사용. 오차 제곱의 평균

- **Cross Entropy Loss**  
  분류 문제에서 사용. 실제 정답과 예측 확률 분포의 차이를 측정

  - Binary Cross Entropy: 이진 분류  
  - Categorical Cross Entropy: 다중 분류

### (3) 옵티마이저 (Optimizer)

- **SGD (Stochastic Gradient Descent)**  
  확률적 경사하강법. 미니배치 단위로 가중치를 업데이트

- **Momentum**  
  기울기의 누적값(속도)을 고려하여 더 빠른 수렴 유도

- **Adam**  
  Momentum과 RMSProp을 결합한 방식. 현재 가장 널리 쓰이는 옵티마이저

- **Adagrad, RMSProp**  
  학습률을 파라미터별로 동적으로 조정

## CNN을 이해하기 위한 사전 개념

### 1. 이미지 데이터의 구조
- **픽셀**: 디지털 이미지는 픽셀로 구성되며, 각 픽셀은 밝기 혹은 색상 정보를 담고 있음
- **채널**: 흑백 이미지는 1채널, 컬러 이미지는 일반적으로 R, G, B의 3채널로 구성됨
- **해상도**: 이미지의 가로 × 세로 크기로, 예: 224×224×3

### 2. 인공신경망 (ANN: Artificial Neural Network)
- **퍼셉트론**: 가장 기본적인 뉴런 단위로 입력값을 받아 가중치를 곱하고, 편향을 더한 뒤 비선형 함수로 출력
- **완전 연결층(Dense layer)**: 모든 입력 뉴런이 모든 출력 뉴런과 연결된 구조
- **한계점**: 이미지처럼 고차원 입력을 처리할 때 파라미터 수가 급증하며, 위치 정보를 반영하지 못하는 한계 존재

### 3. 지역적 연결성과 가중치 공유 개념
- **지역적 수용 영역(Receptive Field)**: 특정 뉴런이 입력의 일부분만 바라보는 구조
- **가중치 공유**: 동일한 필터(커널)를 여러 위치에 적용함으로써 연산 효율성과 특성 추출 가능

## CNN의 기초 개념

### 1. 합성곱 연산(Convolution)
- **정의**: 필터(커널)를 이미지에 슬라이딩하면서 각 영역과의 내적을 통해 특징을 추출하는 연산
- **커널**: 보통 3×3, 5×5 등의 행렬로 구성되며, 학습을 통해 최적화됨
- **특징 맵(Feature Map)**: 필터를 적용한 결과로 생성되는 출력값

### 2. 활성화 함수 (Activation Function)
- **ReLU (Rectified Linear Unit)**: 음수를 0으로 만들고 양수는 그대로 출력하는 함수로, 비선형성 부여
- **역전파에 유리**하며, 학습 속도 향상에 기여함

### 3. 풀링(Pooling)
- **목적**: 특징 맵의 공간적 크기를 줄이고, 중요한 정보만 남기기 위함
- **Max Pooling**: 일정 영역에서 가장 큰 값만 선택
- **Average Pooling**: 일정 영역의 평균값을 선택
- **장점**: 연산량 감소, 과적합 방지, 위치 불변성 제공

### 4. 패딩(Padding)과 스트라이드(Stride)
- **패딩(Padding)**: 입력 가장자리에 0을 추가하여 출력 크기를 조절하거나 정보 손실을 방지
- **스트라이드(Stride)**: 커널이 이동하는 간격을 의미하며, 보통 1 또는 2 사용

### 5. 구조적 특징
- **CNN은 여러 층의 합성곱층, 활성화 함수, 풀링층을 반복적으로 쌓는 구조**
- **마지막에는 완전연결층(Dense Layer)이 존재하며, 최종 분류나 회귀 수행**
---

## CNN의 심화 개념

### 1. 다양한 커널과 깊이 있는 표현
- **다채널 입력**: 각 필터는 입력의 모든 채널에 대해 각각의 가중치를 학습하며, 결과는 하나의 출력 채널로 통합
- **다수의 필터 사용**: 다양한 패턴(모서리, 곡선 등)을 탐지하기 위해 여러 필터를 사용함

### 2. CNN 아키텍처
- **LeNet-5**: 초창기 CNN 구조로, 숫자 인식에 사용됨
- **AlexNet**: ReLU 도입, GPU 병렬처리, 드롭아웃 등으로 큰 성능 향상
- **VGGNet**: 3×3의 작은 필터를 깊게 쌓아 성능 향상
- **ResNet**: 잔차 연결(Skip Connection)을 도입하여 깊은 네트워크 학습 가능

### 3. 학습 기법
- **드롭아웃(Dropout)**: 학습 중 일부 뉴런을 임의로 제거하여 과적합 방지
- **배치 정규화(Batch Normalization)**: 입력 분포를 정규화하여 학습 속도 향상 및 안정화
- **전이 학습(Transfer Learning)**: 사전학습된 CNN 모델을 가져와 재학습하여 적은 데이터로도 높은 성능 확보 가능

### 4. 시각적 해석
- **필터 시각화**: 학습된 필터의 가중치를 이미지화하여 네트워크가 어떤 특징에 민감한지 확인
- **특징 맵 시각화**: 각 층의 출력값을 시각화하여 추출된 정보 흐름을 파악 가능

### 5. 한계점과 대응 방안
- **데이터 요구량 큼**: 많은 데이터가 필요하며, 이를 위해 증강(Augmentation) 기법 활용
- **계산 자원 소모 큼**: 연산이 많기 때문에 GPU 사용 필수
- **해석 어려움**: 왜 특정 결과가 나왔는지 설명 어려움, 시각화 기법 및 XAI 연구 진행 중
