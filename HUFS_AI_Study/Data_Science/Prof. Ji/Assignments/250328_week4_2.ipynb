# Basic Data Visualization (Without pandas)

This notebook demonstrates how to load the Iris dataset with scikit-learn, compute simple statistics with NumPy, and create basic plots with matplotlib. We do **not** use pandas in this example.


## 1. Imports & Dataset Loading

We will:
1. Load the Iris dataset from `sklearn.datasets`.
2. Separate the features (`X`) and labels (`y`).
3. Inspect the shapes and feature names.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris()
X = iris.data  # Shape: (150, 4)
y = iris.target  # Shape: (150,)
feature_names = iris.feature_names
target_names = iris.target_names

print("Data shape (X):", X.shape)
print("Labels shape (y):", y.shape)
print("Feature names:", feature_names)
print("Target names:", target_names)
```

## 2. Basic Statistics

Using NumPy, we can quickly check the mean and standard deviation for each feature. We’ll also look at how many samples there are in each species.

```python
# Calculate the mean and standard deviation for each feature
mean_values = np.mean(X, axis=0)    # use reduction function to calculate mean
std_values = np.std(X, axis=0)      # use reduction function to calculate std

print("Feature means:", mean_values)
print("Feature std dev:", std_values)

# Count how many samples belong to each species
unique_species, counts = np.unique(y, return_counts=True)
print("\nSpecies distribution:")
for species_id, count in zip(unique_species, counts):
    print(f"{target_names[species_id]}: {count} samples")
```

## 3. Data Visualizations
We'll create:
1. A **Histogram** of one feature.
2. A **Scatter Plot** for two features, colored by species.
3. A **Box Plot** to compare one feature across species.


### 3.1 Histogram
Let's plot the distribution of **sepal length** (feature index 0).

```python
print(X)
```

```python
# Histogram of the first feature: Sepal Length

feature_idx = 0

plt.hist(X[:, feature_idx], bins = 15, edgecolor = "black")
plt.title("Histogram of sepal length (cm)")
plt.ylabel("Count")
plt.xlabel(feature_names[0]) 
plt.show()
```

### 3.2 Scatter Plot
We'll create a scatter plot comparing **sepal length** (feature 0) and **sepal width** (feature 1), coloring points by species.

```python
# Create a scatter plot for sepal length vs. sepal width
colors = ['red', 'green', 'blue']

# print(unique_species)

markers = ['o', '^', 's']  # 원, 삼각형, 사각형

for species_id in unique_species:
    # Select rows (samples) that match the current species
    X_species = X[y == species_id] # use logical slicing to select data
    # X에는 타겟(y) 정보가 명시적으로 붙어있지 않은데, X[y == species_id]가 어떻게 동작 가능한가?
    # X와 y는 같은 순서(인덱스 기준)로 정렬되어 있으므로, y를 인덱스로 사용하여 X에서 필터링이 가능.
    plt.scatter(
        X_species[:, 0],  # sepal length
        X_species[:, 1],  # sepal width
        color=colors[species_id],
        marker=markers[species_id],
        label=target_names[species_id],
    ) # plot the data using plt.scatter()

# Add titles, labels and legend
plt.title("sepal length (cm) vs sepal width (cm)")
plt.xlabel(feature_names[0]) 
plt.ylabel(feature_names[1])  
plt.legend()
plt.show()
```

### 3.3 Box Plot
Now, let's visualize the distribution of **petal length** (feature index 2) across the three species.

```python
# Box plot for petal length by species
petal_length_index = 2  # 'petal length (cm)'
box_data = []

for species_id in unique_species:
    # Get all petal length values for this species
    petal_lengths = X[y == species_id, petal_length_index] # use logical slicing to select data
    box_data.append(petal_lengths)

# plot the data using plt.boxplot()
plt.boxplot(box_data, labels=target_names)

# Add titles, labels and legend
plt.title("Petal Length by Species")
plt.ylabel(feature_names[petal_length_index])
plt.xlabel("Species")
plt.show()
```
