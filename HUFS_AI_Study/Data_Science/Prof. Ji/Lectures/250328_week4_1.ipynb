## Python matplotlib

### 1. plt.plot() for plots

```python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3, 4]) # plt.plot([1, 2, 3, 4]) 이 코드에서 [1, 2, 3, 4]는 y값에 해당하며, x값은 자동으로 0부터 시작하는 정수 시퀀스(즉, [0, 1, 2, 3])로 설정
plt.ylabel('y label')
plt.xlabel('x label')
plt.show() # 이 코드는 코랩에서는 없어도 됨
```

```python
plt.plot([0.1, 0.2, 0.3, 0.4],[1, 2, 3, 4])
# [0.1, 0.2, 0.3, 0.4]: x값
# [1, 2, 3, 4]: y값
plt.ylabel('y label')
plt.xlabel('x label')
```

### 2. Use broadcasting to plot a function

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(10)      # 0에서 9 사이의 정수값을 생성함
plt.plot(x,x**2)       # x^2 함수를 plot
# 즉, y = x² 함수의 그래프를 그리는 것
# x축: 0부터 9까지
# y축: 각 x값의 제곱
# 선 그래프 형태로, 2차 함수의 곡선이 그려짐
```

```python
x = np.arange(10)  # 0에서 9 사이의 정수값을 생성함
plt.plot(x, x**2)     #  x^2 함수를 plot
plt.axis([0, 100, 0, 100]) # 범위를 지정함
# plt.axis([x_min, x_max, y_min, y_max])
```

### 3 line styles of plot()

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(-20, 20)      # -20에서 20 사이의 수를 1의 간격으로 생성.
y1 = 2*x                    # 2*x를 원소로 가지는 y1 함수
y2 = (1/3)*x**2 + 5         # (1/3)*x**2 + 5를 원소로 가지는 y2 함수
y3 = -x**2 - 5              # -x**2 - 5를 원소로 가지는 y3 함수
# 빨강색 점선, 녹색 실선과 세모기호, 파랑색 별표와 점선으로 함수를 표현

print(y1)
print("\n---------------------------------------------------------\n")
print(y2)
print("\n---------------------------------------------------------\n")
print(y3)
print("\n---------------------------------------------------------\n")

plt.plot(x, y1, 'g--', x, y2, 'r^-', x, y3, 'b*:')
plt.axis([-30, 30, -30, 30]) # 그림을 그릴 영역을 지정함
```

### Exercise: draw a sine and cosine function

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, np.pi * 2, 100) #2*pi는 360도에 해당함

plt.plot(x, np.sin(x), 'r-') # 사인함수는 빨강 실선으로
plt.plot(x, np.cos(x), 'b:') # 코사인함수는 파랑 점선으로

# plt.plot(x, np.sin(x), 'r^-') 
# plt.plot(x, np.cos(x), 'b*:') 
```

### 4. Title and labels of a figure

```python
x = np.linspace(0, np.pi * 2, 100)

plt.title('Sin cos curve')
plt.plot(x, np.sin(x), 'r-', label='sin curve')
plt.plot(x, np.cos(x), 'b:', label='cos curve')
plt.xlabel('x value')
plt.ylabel('y value')
plt.legend()
```

### 5. Object-oriented interface: figure, axes

```python
fig, ax = plt.subplots(2, 2) # 2행 2열의 그림을 그리는 공간을 만든다
# 3행 2열로 바꿔보기 f, ax = plt.subplots(nrows=3, ncols=2, figsize=(8, 6))

X = np.random.randn(100)     # 정규 분포를 가지는 데이터
Y = np.random.randn(100)     # 정규 분포를 가지는 데이터
ax[0, 0].scatter(X, Y)       # 산점도 그림

X = np.arange(10)            # 0에서 9 사이의 연속값
Y = np.random.uniform(1, 10, 10) # 균일분포값 생성
ax[0, 1].bar(X, Y)           # 막대 차트 그림

X = np.linspace(0, 10, 100)
Y = np.cos(X)
ax[1, 0].plot(X, Y)          # 실선으로 함수를 그림

Z = np.random.uniform(0, 1, (5, 5))
ax[1, 1].imshow(Z)           # 분포를 2D 이미지로 그림

ax[1,0].set_xlabel("ㅋㅋ")
```

### Scatter plots and bar graphs

```python
N = 30                     # 전체 점의 개수
x = np.random.rand(N)      # 점들의 임의의 x 좌표
y = np.random.rand(N)      # 점들의 임의의 y 좌표
colors = np.random.rand(N) # 랜덤한 색상
area = (30 * np.random.rand(N))**2 # 반지름이 0에서 15 포인트 범위

plt.scatter(x, y, s=area, c=colors, alpha=0.5)
# s의 역할: 점의 면적 (size)
```

```python
x = np.arange(3)
years = ['2010', '2011', '2012']
domestic = [6801, 7695, 8010]
foreign = [777, 1046, 1681]

plt.bar(x, domestic)
plt.bar(x, foreign)
plt.xticks(x, years)
```

```python
x = np.arange(3)
years = ['2010', '2011', '2012']
domestic = [6801, 7695, 8010]
foreign = [777, 1046, 1681]

plt.bar(x, domestic, width=0.25)
plt.bar(x + 0.3, foreign, width=0.25)
plt.xticks(x, years)
```

### Pie charts and heatmaps

```python
import matplotlib.pyplot as plt
import numpy as np

data = [5, 4, 6, 11]
clist = ['cyan', 'gray','orange', 'red']
plt.pie(data, autopct = '%.2f', colors=clist)
```

```python
import matplotlib.pyplot as plt
import numpy as np

data = [5, 4, 6, 11]
clist = ['cyan', 'gray','orange', 'red']
explode = [.06, .07, .08, .09]
plt.pie(data, autopct = '%.2f%%',
        colors=clist, labels=clist, explode=explode)
```

### Normal distribution using histogram

```python
import numpy as np
import matplotlib.pyplot as plt

f1 = np.random.randn(100000)
plt.hist(f1, bins=200, color='red', alpha=.7, label='avg = 0, std = 1')
plt.axis([-8, 8, -2, 2500])
plt.legend()
```

```python
f1 = np.random.normal(loc=0, scale=1, size=100000)
f2 = np.random.normal(loc=3, scale=3, size=100000)

plt.hist(f1, bins=200, color='red', alpha=.4,
         label='avg = 0, std = 1')
plt.hist(f2, bins=200, color='green', alpha=.4,
         label='avg = 3, std = 3')
plt.axis([-8, 8, -2, 2500])
plt.legend()
```

```python
f1 = np.random.normal(loc=0, scale=1, size=10000)
f2 = np.random.normal(loc=3, scale=.5, size=10000)

plt.hist(f1, bins=200, color='red', alpha=.7, label='loc = 0, scale = 1')
plt.hist(f2, bins=200, color='blue', alpha=.5, label='loc = 3, scale = .5')
```
