### Correlation Coefficient

```python
import numpy as np   # Repeated content, omitted going forward

np.random.seed(85)   # Using seed 85 to ensure reproducible results
x = np.arange(0, 10) # Generating 0 through 9
y1 = x * 2           # Result of multiplying x by 2
np.corrcoef(x, y1)
```

```python
x = np.arange(0, 10) # Generating 0 through 9
y2 = x ** 3          # x cubed
np.corrcoef(x, y2)
```

```python
x = np.arange(0, 10)                     # Generating 0 through 9
y3 = np.random.randint(0, 100, size=10)  # random integers between 0 and 100
np.corrcoef(x, y3)
```

```python
result = np.corrcoef((x, y2, y3))
print(result)
```

```python
import matplotlib.pyplot as plt

plt.imshow(result)
plt.colorbar()
```

### Anscombe’s quartet dataset

```python
import seaborn as sns

anscombe = sns.load_dataset("anscombe")
print(anscombe.head())
```

```python
anscombe
```

```python
anscombe.info()
```

```python
anscombe.groupby('dataset').describe()
```

```python
sns.lmplot(x="x", y="y", data=anscombe[anscombe["dataset"] == "I"],
           ci=None, scatter_kws={"s": 80})
```

```python
sns.lmplot(x="x", y="y", data=anscombe[anscombe["dataset"] == "II"],
           ci=None, scatter_kws={"s": 80})
# lmplot is a function in seaborn that fits a linear regression model to the data
print(np.corrcoef(anscombe[anscombe["dataset"] == "II"]['x'], anscombe[anscombe["dataset"] == "II"]['y']))
```

### non linear function

```python
import seaborn as sns

anscombe = sns.load_dataset("anscombe")
sns.lmplot(x="x", y="y", data=anscombe[anscombe["dataset"] == "II"],
           order=2, ci=None, scatter_kws={"s": 80})
# order=2 indicates a quadratic fit
```

```python
sns.lmplot(x="x", y="y", data=anscombe[anscombe["dataset"] == "III"],
           ci=None, scatter_kws={"s": 80})
```

```python
sns.lmplot(x="x", y="y", data=anscombe[anscombe["dataset"] == "III"],
           robust=True, ci=None, scatter_kws={"s": 80})
```
