### Series and DataFrame

```python
import pandas as pd
import numpy as np

income = {'Jan' : 9500, 'Feb': 6200, 'Mar': 6050, 'Apr': 7000}
income_se = pd.Series(income)   #월을 인덱스로 하는 매출 시리즈
print('Income')
print(income_se)
```

```python
month_se = pd.Series(['Jan', 'Feb', 'Mar', 'Apr'])
income_se = pd.Series([9500, 6200, 6050, 7000])
expenses_se = pd.Series([5040, 2350, 2300, 4800])
df = pd.DataFrame( {'Month': month_se, 'Income': income_se,
                   'Expenses' : expenses_se})
df    # print(df)를 이용해서 출력형식을 비교해 보자
```

```python
print(df)
```

```python
# 판다스 Series를 이용하여 최대 수익 월을 출력하기
m_idx = np.argmax(income_se)  # 넘파이의 argmax() 사용
print('Month with the greatest income:', month_se[m_idx])
```

```python
print('Max monthly income:', income_se.max(),', Mean monthly income:', income_se.mean())
```

### loc and iloc

```python
import pandas as pd

path = 'https://github.com/dongupak/DataML/raw/main/csv/'
file = path+'vehicle_prod.csv'
```

```python
df.iloc[4]  # this results in a Series
```

```python
# ✅ 차이 요약
# 항목	        iloc()	                        loc
# 기준	        정수 위치 (0부터 시작)	        인덱스 레이블 또는 열 이름
# 인덱스 타입	  정수(int)만 가능	            문자열, 숫자 등 레이블 기반
# 슬라이싱	     Python 스타일 (end 제외)	   Pandas 스타일 (end 포함)

# df.iloc (Integer-location based indexing): 파이썬의 리스트처럼 0부터 시작하는 정수 인덱스를 사용
# df.loc (Label-based indexing): 레이블(이름 또는 인덱스 이름) 기반으로 데이터를 선택


# df.loc[...]과 df.iloc[...] 모두 기본적으로는 "행 기준 인덱싱"
# 즉, 단일 인자([])를 사용할 경우에는 "행"을 선택

    # 열(column)을 선택하고 싶다면?

    # df['열이름']: 열 단독 선택
    # df.loc[:, '열이름'] 또는 df.iloc[:, 열번호]: 열 선택 (명시적)
    # df.loc[행, 열]: 행과 열 동시에 지정
```

```python
df.loc[['US','Korea']] 
```

```python
df.loc['US','Korea'] # Error


# df.loc['Korea']         # ✅ 실행됨 → Series 반환
# df.loc['US', 'Korea']   # ❌ 오류 발생

# df.loc[행, 열]의 형태
# 첫번째 코드는 'Korea'라는 이름을 가진 행을 선택하며, 결과는 Series
# df.loc[['US', 'Korea']]  # ✅ 여러 행 선택, DataFrame 반환
```

### Adding a column to the data

```python
import pandas as pd

path = 'https://github.com/dongupak/DataML/raw/main/csv/'
file = path+'vehicle_prod.csv'
df = pd.read_csv(file, index_col=0)   

df['total'] = df.sum(axis = 1) # add a new column 'total' to df
print(df)
```

```python
# you can also use .sum and .mean methods of DataFrame like this:
df['total'] = df[['2007', '2008', '2009', '2010', '2011']].sum(axis=1)
df['mean'] = df[['2007', '2008', '2009', '2010', '2011']].mean(axis=1)
print(df)
```

```python
df.drop('2007', inplace=True, axis=1)
print(df)  # we still have the same mean and total values from above     
```

```python
# Recalculate the total and mean values after dropping the column (they don't automatically update)
df['total'] = df[['2008', '2009', '2010', '2011']].sum(axis=1)
df['mean'] = df[['2008', '2009', '2010', '2011']].mean(axis=1)
print(df)      
```

### inplace=True for renewing the DataFrame

```python
print(d_df)         # d_df dataframe does not change
```

```python
d_df.drop('B', axis=1, inplace=True) # inplace = True
print(d_df)   # d_df is changed
```

### Grouping the data

```python
import pandas as pd

path = 'https://github.com/dongupak/DataML/raw/main/csv/'
weather_file = path + 'weather.csv'

weather = pd.read_csv(weather_file, encoding='CP949')

weather['month'] = pd.DatetimeIndex(weather['일시']).month
print(weather)
```

```python
# numeric_only = True discards non-numeric columns when calculating the mean
# group by 'month'
monthly_means = weather.groupby('month').mean(numeric_only = True)
print(monthly_means)    
```

### concat() for merging dataframes

```python
import pandas as pd

df_1 = pd.DataFrame( {'A' : ['a10', 'a11', 'a12'],
                      'B' : ['b10', 'b11', 'b12'],
                      'C' : ['c10', 'c11', 'c12']} ,
                      index = ['가', '나',  '다'] )

df_2 = pd.DataFrame( {'B' : ['b23', 'b24', 'b25'],
                      'C' : ['c23', 'c24', 'c25'],
                      'D' : ['d23', 'd24', 'd25']} ,
                      index = ['다', '라',  '마'] )
```

```python
df_1
```

```python
df_2
```

```python
df_3 = pd.concat( [df_1, df_2])
print(df_3)
```

```python
df_4 = pd.concat( [df_1, df_2], join='inner')
print(df_4)
```
