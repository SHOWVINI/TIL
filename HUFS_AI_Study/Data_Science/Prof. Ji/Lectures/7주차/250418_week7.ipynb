# Gradient Descent - Coded Example
We demonstrates **gradient descent** to minimize the function:

\[ y = (x - 3)^2 \]

- The function has a minimum at \( x = 3 \)
- Gradient descent is used to find this minimum step-by-step


```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
```

## Define the Function and Its Gradient

```python
# Define the function y = (x - 3)^2
def f(x):
    return (x - 3) ** 2

# Define its gradient (derivative)
def grad_f(x):
    return 2 * (x - 3)
```

## Initialize Parameters

```python
# Initialize x, learning rate, and number of epochs
x = 0           # starting point
learning_rate = 0.1 # 적절한 학습률 설정이 중요: 진동 | 발산하여 값이 수렴할 수 없음
epochs = 25

# Lists to keep track of x and y values
x_history = [x]
y_history = [f(x)]
```

## Gradient Descent Loop

```python
for _ in range(epochs):
    grad = grad_f(x)
    x = x - learning_rate * grad  # update rule
    x_history.append(x)
    y_history.append(f(x))
```

## Visualize the Gradient Descent Steps

```python
x_vals = np.linspace(-1, 7, 200)
y_vals = f(x_vals)

plt.figure(figsize=(8, 5))
plt.plot(x_vals, y_vals, label="y = (x - 3)^2")
plt.scatter(x_history, y_history, color='red', label="Gradient Descent Steps")
plt.plot(x_history, y_history, linestyle='--', color='red')
plt.title("Gradient Descent Visualization")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

# Gradient Descent in 2D
This part demonstrates **gradient descent** on a 2D function:

\[ f(x_1, x_2) = (x_1 - 2)^2 + 3(x_2 + 1)^2 \]

The goal is to find the minimum using gradient descent and visualize the **curved descent path**.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Define the Function and Its Gradient

```python
def f(x1, x2):
    return (x1 - 2)**2 + 3 * (x2 + 1)**2

def grad_f(x1, x2):
    df_dx1 = 2 * (x1 - 2)
    df_dx2 = 6 * (x2 + 1)
    return df_dx1, df_dx2
```

## Initialize Variables and Perform Gradient Descent

```python
x1, x2 = -3.0, 4.0  # initial point
learning_rate = 0.2
epochs = 30
history = [(x1, x2)]

for _ in range(epochs):
    df_dx1, df_dx2 = grad_f(x1, x2)
    x1 -= learning_rate * df_dx1
    x2 -= learning_rate * df_dx2
    history.append((x1, x2))
```

## Plot Contour with Descent Path

```python
X1 = np.linspace(-5, 5, 100)
X2 = np.linspace(-5, 5, 100)
X1, X2 = np.meshgrid(X1, X2)
Z = f(X1, X2)

hx, hy = zip(*history)

plt.figure(figsize=(8, 6))
plt.contour(X1, X2, Z, levels=30, cmap='viridis')
plt.plot(hx, hy, marker='o', color='red', linestyle='--', label='Descent Path')
plt.title("Gradient Descent on $f(x_1, x_2) = (x_1 - 2)^2 + 3(x_2 + 1)^2$")
plt.xlabel("$x_1$")
plt.ylabel("$x_2$")
plt.legend()
plt.grid(True)
plt.show()
```
