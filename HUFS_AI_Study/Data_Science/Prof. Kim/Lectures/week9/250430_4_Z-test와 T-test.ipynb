# 📊 Z-test와 T-test: **‘평균이 다른지 알아보는 도구’**

## 왜 필요한가?

핵심은 *“평균 비교”*라는 점임

### 👉 상황 예시

- 새로 개발한 약이 기존 약보다 효과가 **좋은지 알고 싶음**
- A반과 B반의 평균 시험 점수가 **정말로 차이 나는지 확인하고 싶음**
- 어떤 공장에서 만든 제품 무게가 **정해진 기준(예: 100g)**을 잘 지키고 있는지 확인하고 싶음

👉 이럴 때 우리는 단순히 평균만 보고 판단하면 안 됨

👉 왜냐하면 **우연히** 평균이 다르게 나올 수도 있기 때문

👉 그래서 **통계적으로 의미 있는 차이인지** 검정하는 과정이 필요함

---

## 🔍 Z-test와 T-test는 뭐하는 도구인가?

- 두 방법 모두 **평균 차이가 '우연인지' 아니면 '진짜 차이인지' 판단하는 도구**
- 말 그대로 “A와 B의 평균이 다르다”는 주장을 **통계적으로 증명하기 위한 검정 도구**

---

## 🎯 Z-test와 T-test의 공통 목적

> “두 평균이 다르다고 말할 수 있는 ‘충분한 증거’가 있나요?”를 판단하는 것
> 

이 질문에 대해:

- **Z-test**: 모집단 정보가 충분할 때 답하는 방식
- **T-test**: 표본만 가지고 답해야 할 때 사용하는 방식

---

## 🧪 Z-test: 정보를 많이 알고 있을 때 쓰는 도구

- 모집단의 평균과 **표준편차(흩어짐 정도)**까지 **정확히 알고 있을 때** 사용
- 즉, **기준이 명확한 경우**: 예를 들어 “이 제품 무게는 100g이어야 한다” → 실제 무게와 비교할 때

---

## 🔬 T-test: 정보를 적게 알고 있을 때 쓰는 도구

- 모집단의 표준편차를 모를 때
- **소수의 표본만** 가지고 평균 차이를 검정할 때
- 예를 들어 “우리 반과 옆 반의 평균 점수가 차이 나는가?”를 **시험 점수 10명씩만 보고 판단**할 때

---

## 📌 정말 쉬운 비교 정리

| 구분 | Z-test | T-test |
| --- | --- | --- |
| 사용 목적 | 평균 차이가 **의미 있는 차이인지** 확인 | 평균 차이가 **의미 있는 차이인지** 확인 |
| 언제 사용? | 모집단 정보가 충분할 때 (특히 표준편차 알고 있을 때) | 표본만 있고, 모집단 표준편차를 모를 때 |
| 예시 | 공장 제품 무게가 100g 기준과 다른지 확인 | 두 반의 시험 평균 점수가 다른지 확인 |

---

## 🔁 요약: Z-test와 T-test는 왜 존재할까?

- 두 방법 모두 **“평균 차이 검정”이라는 같은 목표를 가짐**
- 상황에 따라 **모델을 골라 쓰는 것**임
    - 많은 정보가 있으면 Z-test
    - 정보가 적고 표본만 있으면 T-test

---

```python
# 필요한 패키지 설치

from scipy.stats import shapiro  
# scipy 라이브러리에서 샤피로-윌크 정규성 검정 함수를 불러오는 명령문
# 샤피로 검정은 데이터가 정규분포를 따르는지를 확인하는 데 사용되며, p-value를 통해 판단함
# 통계적 가설 검정을 진행하기 전, 정규성 가정을 확인하는 데 필수적으로 사용됨

from statsmodels.stats.weightstats import ztest as ztest  
# statsmodels 라이브러리에서 Z-검정을 위한 ztest 함수를 불러오며, 이름을 ztest로 지정함
# 평균 비교 시 모집단 분산을 알고 있을 때 사용하는 검정 방식으로, 정규분포 가정하에 평균 차이를 비교함

import scipy.stats  
# scipy.stats 전체를 불러오는 명령문으로, 다양한 통계 검정 함수들을 사용할 수 있게 함
# 예: t-test, chi-square test, anova, correlation 등 다수 포함됨

from scipy import stats  
# scipy.stats와 동일한 모듈을 다른 방식으로 불러오는 명령문
# stats.ttest_ind, stats.pearsonr 등 통계 분석 함수 사용을 가능하게 함
# 기존 `import scipy.stats`와 기능 중복이 있으므로, 둘 중 하나만 사용하는 것이 일반적임

from scipy.stats import ttest_ind  
# 두 독립 표본 간 평균 차이를 검정하는 독립표본 t-검정 함수(ttest_ind)를 별도로 불러오는 명령문
# 두 그룹의 평균이 통계적으로 유의하게 다른지를 검정할 때 사용됨
# 예: 실험군 vs 대조군, 남성 vs 여성 등

import pandas as pd  
# 테이블 형태의 데이터를 다루기 위한 pandas 라이브러리를 불러오는 명령문
# 데이터프레임을 통해 데이터 조작, 분석, 결측값 처리 등에 활용됨

import seaborn as sns  
# seaborn 라이브러리를 불러오는 명령문
# 통계적 시각화를 직관적이고 세련된 형태로 표현할 수 있도록 도와주는 고수준 시각화 도구임

import matplotlib.pyplot as plt  
# matplotlib의 pyplot 모듈을 불러오는 명령문
# 그래프를 직접 그릴 수 있는 저수준의 시각화 기능을 제공함 (선, 막대, 산점도 등)

plt.rcParams['figure.dpi'] = 300  
# 시각화 이미지의 해상도를 300 DPI(Dots Per Inch)로 설정하는 명령문
# 고화질 이미지를 출력하여 인쇄나 리포트 작성 시 시각적 품질을 향상시킴
```

### 정규성 검정 요약

| 항목 | 내용 |
| --- | --- |
| **검정 이름** | Shapiro-Wilk 정규성 검정 |
| **귀무가설(H₀)** | 데이터는 정규분포를 따른다 |
| **대립가설(H₁)** | 데이터는 정규분포를 따르지 않는다 |
| **해석 기준** | p-value > 0.05 → 정규성 있음 인정 (H₀ 기각 못 함) |

### 정규성 검정이 필요한 이유 요약

| 항목 | 설명 |
| --- | --- |
| **t-test, z-test의 기본 가정** | 모집단(또는 표본)의 **정규분포** 가정을 전제로 함 |
| **정규성 미충족 시 문제점** | p-value 신뢰성 저하, 제1종 오류 또는 제2종 오류 확률 증가 |
| **정규성이 중요해지는 이유** | 평균(mean) 기반 검정은 분포가 비대칭이거나 이상치에 민감함 |
| **정규성 없을 경우 대안** | 비모수 검정 (Wilcoxon, Mann-Whitney 등) 사용 |

```python
# 데이터 정규성 검정

# shapiro() 함수는 Shapiro-Wilk 검정을 수행하며, 정규성 여부를 판단하는 데 사용됨
# 출력값은 (검정 통계량 W, p-value) 형태의 튜플로 반환됨
# p-value > 0.05이면 '정규분포를 따른다'는 귀무가설을 기각할 수 없음 → 정규성 가정 충족으로 판단함

print(shapiro(df['TypeA_before']))  
# TypeA_before 데이터가 정규분포를 따르는지를 검정함

print(shapiro(df['TypeA_after']))  
# TypeA_after 데이터에 대해 동일한 정규성 검정을 수행함
# 사전/사후 비교와 같은 구조에서는 각각의 시점에 대해 정규성 검정을 따로 수행해야 함
# 이 결과에 따라 대응표본 t-test 또는 비모수검정(Wilcoxon) 여부를 결정할 수 있음

print(shapiro(df['TypeB_before']))  
# TypeB_before 골프공 타입의 사전 데이터에 대한 정규성 검정을 수행함

print(shapiro(df['TypeB_after']))  
# TypeB_after 골프공 타입의 사후 데이터에 대한 정규성 검정을 수행함

print(shapiro(df['TypeC_before']))  
# TypeC_before 골프공 타입의 사전 측정값에 대한 정규성 검정을 수행함

print(shapiro(df['TypeC_after']))  
# TypeC_after 골프공 타입의 사후 측정값에 대한 정규성 검정을 수행함
```

### 바틀렛 검정 요약

| 항목 | 내용 |
| --- | --- |
| **검정 목적** | 여러 그룹 간 분산이 같은지(등분산성)를 검정함 |
| **사용 조건** | 각 그룹의 데이터가 정규분포를 따를 때 사용함 |
| **귀무가설(H₀)** | 모든 그룹의 분산이 같다 |
| **p-value 해석** | p > 0.05 → 등분산성 인정 / p ≤ 0.05 → 등분산성 기각(→ 이분산성 존재) |

```python
# 데이터 등분산성 검정

stats.bartlett(df['TypeA_before'], df['TypeA_after'],
               df['TypeB_before'], df['TypeB_after'],
               df['TypeC_before'], df['TypeC_after'])  
# 각 그룹 간 분산이 동일한지(등분산성, Homogeneity of Variance)를 검정하는 바틀렛(Bartlett) 검정을 수행함
# 입력된 각 시리즈는 하나의 그룹을 의미하며, 총 6개의 그룹(타입A~C의 사전/사후 데이터)을 대상으로 검정함

# 바틀렛 검정은 다음과 같은 통계적 가설을 전제로 함:
#   - 귀무가설(H₀): 모든 그룹의 분산이 동일함 (등분산이다)
#   - 대립가설(H₁): 적어도 하나의 그룹의 분산이 다름 (이분산이다)

# 반환값은 (검정 통계량, p-value)의 튜플이며, 해석 기준은 다음과 같음:
#   - p-value > 0.05: 등분산성 가정이 충족됨 (귀무가설 기각 불가)
#   - p-value ≤ 0.05: 등분산성이 성립하지 않음 (귀무가설 기각 → 이분산)

# 등분산성은 ANOVA나 독립표본 t-test 수행 전 필수로 확인해야 하는 조건임
# 이 검정 결과에 따라 → 등분산 가정 하의 분석(t-test, ANOVA) 또는 비등분산 처리(Welch t-test 등) 여부를 결정함
```

## ✅ 평균 차이 검정 전 필수 절차

1. **정규성 검정 (Shapiro-Wilk 등)**
2. **등분산성 검정 (Bartlett, Levene 등)**
3. **t-test, z-test 등 실제 평균 비교 검정**

---

## 1️⃣ 왜 먼저 정규성 검정을 하는가?

- **t-test나 z-test는 "정규분포를 따른다"는 전제를 가짐**
- 데이터가 정규분포를 따르지 않으면, 이들 검정의 **신뢰성**과 **유의성 판단**이 무너짐
- 따라서, **모수 검정(t/z-test)을 쓸 수 있는지를 확인하기 위한 1차 관문**이 정규성 검정임

> 📌 정규성이 안 맞으면?
> 
> 
> → **비모수 검정** (Mann-Whitney, Wilcoxon 등)으로 전환
> 

---

## 2️⃣ 왜 정규성 이후에 등분산성 검정을 하는가?

- **등분산성 검정(Bartlett, Levene)은 '여러 그룹의 분산이 같은지'를 확인함**
- 이 역시 t-test, ANOVA와 같은 모수 검정에서 전제 조건임
- 하지만 **등분산성 검정은 정규분포라는 가정 하에 더 정확하게 작동함**
    
    → 특히 Bartlett 검정은 정규성 위반 시 민감하게 잘못된 결론을 유도함

## 3️⃣ 등분산성 검정은 왜 중요한가?

등분산성은 평균 비교 검정 시, **각 그룹이 가진 정보(분산)의 균형을 가정하는 조건**입니다.

| 조건 | 이유 및 영향 |
| --- | --- |
| **등분산 유지** | → 고전적 t-test, ANOVA 사용 가능 |
| **등분산 깨짐 (이분산)** | → Welch t-test, Brown-Forsythe test 등 **보정된 검정** 사용 필요 |
| **등분산성 무시하고 분석** | → **신뢰구간 왜곡**, **Type I/II 오류율 증가**, 결론 신뢰도 하락 |

---

## 🔁 종합 순서 정리

| 단계 | 내용 | 도구 |
| --- | --- | --- |
| ① | 정규성 검정 | Shapiro-Wilk, Kolmogorov |
| ② | 등분산성 검정 | Bartlett (정규), Levene (비정규) |
| ③ | 평균 차이 검정 | t-test, z-test, Welch, ANOVA 등 |

```python
# Z-test (TypeA_before와 TypeA_after 간 평균 차이 검정)

ztest(df['TypeA_before'], x2=df['TypeA_after'], value=0, alternative='two-sided')  
# 두 그룹(TypeA_before vs TypeA_after)의 평균 차이가 통계적으로 유의미한지를 검정하기 위한 Z-test 수행 명령문
# Z-test는 다음과 같은 조건 하에서 사용하는 평균 차이 검정 방법임:
#   ✅ 모집단 분산(또는 표준편차)을 알고 있거나 → 표본 수가 매우 큰 경우 (n > 30)
#   ✅ 데이터가 정규분포를 따른다는 가정이 충족될 경우
# 본 예시는 '두 집단의 평균 차이가 0이 아니다'라는 양측 검정을 수행함

# 인자 설명:
# df['TypeA_before'] → 첫 번째 표본 (사전 테스트 결과)
# x2=df['TypeA_after'] → 두 번째 표본 (사후 테스트 결과)
# value=0 → 두 표본의 평균 차이에 대한 귀무가설: μ₁ - μ₂ = 0
# alternative='two-sided' → 양측 검정, 즉 '두 그룹의 평균이 다르다'를 검정함

# 반환 결과는 (Z 통계량, p-value)의 튜플임
#   - Z 통계량: 표본 평균 차이가 얼마나 유의미하게 큰지를 나타내는 표준 점수
#   - p-value: 귀무가설(평균 차이 = 0)을 기각할 확률 수준
# p < 0.05이면 귀무가설을 기각하고, 두 평균 간에 유의미한 차이가 있다고 판단함
```

| 항목 | 값 | 의미 |
| --- | --- | --- |
| **Z 통계량** | -1.2314 | 표본 평균 차이가 표준 오차 기준으로 -1.23 표준편차 떨어져 있음 |
| **p-value** | 0.2182 | 귀무가설(두 그룹 평균이 같다)을 기각할 확률 수준 |

### 해석 순서

#### ① **귀무가설(H₀)**:

- `TypeA_before`와 `TypeA_after`의 **평균이 같다**
    
    (즉, **변화가 없다**)
    

#### ② **대립가설(H₁)**:

- `TypeA_before`와 `TypeA_after`의 **평균이 다르다**

#### ③ **p-value = 0.2182**

- 유의수준(α) = 0.05 기준에서
    
    → **p > 0.05** 이므로 **귀무가설을 기각할 수 없음**
    

> 🔹 즉, 두 그룹 간 평균 차이는 통계적으로 유의하지 않다는 해석 가능
> 

---

### 결론

- **TypeA 골프공의 사전/사후 측정값 평균은 유의하게 다르지 않음**
- 이 변화는 **우연에 의한 차이일 가능성**이 높음
- 따라서, 해당 실험 조건에서 **TypeA의 변화 효과는 통계적으로 검증되지 않음**

### 세 검정의 차이 요약

| 검정 종류 | 대립가설(H₁) | 사용 목적 |
| --- | --- | --- |
| 양측검정 | 평균이 서로 다름 (μ₁ ≠ μ₂) | **방향 지정 없이 단순한 평균 차이 판단** |
| 단측검정 (smaller) | μ₁ < μ₂ | **A가 B보다 작다고 판단하고 싶은 경우** |
| 단측검정 (larger) | μ₁ > μ₂ | **A가 B보다 크다고 판단하고 싶은 경우** |

```python
# Z-test (TypeA_before와 TypeB_before 간 평균 차이 검정)

# 양측검정
print(ztest(df['TypeA_before'], x2=df['TypeB_before'], value=0, alternative='two-sided'))  
# 두 독립된 집단(TypeA_before vs TypeB_before)의 평균이 서로 다르다는 가설을 검정하는 명령문
# 귀무가설(H₀): 두 집단의 평균이 같다 (μ₁ - μ₂ = 0)
# 대립가설(H₁): 두 집단의 평균이 다르다 (μ₁ ≠ μ₂)
# → 차이의 방향을 특정하지 않고 "그저 다르다"는 전제 하에 검정함
# p-value가 0.05보다 작으면 귀무가설 기각 → 유의미한 평균 차이가 있다고 판단함

# 단측검정(왼꼬리검정)
print(ztest(df['TypeA_before'], x2=df['TypeB_before'], value=0, alternative='smaller'))  
# 귀무가설(H₀): 두 집단의 평균이 같다 (μ₁ - μ₂ = 0)
# 대립가설(H₁): TypeA_before의 평균이 TypeB_before보다 **작다** (μ₁ < μ₂)
# → 특정 방향(왼쪽 꼬리, 음의 방향)으로 차이가 있다고 가정함
# p-value가 0.05보다 작으면, TypeA가 TypeB보다 유의하게 작다고 판단함

# 단측검정(오른꼬리검정)
print(ztest(df['TypeA_before'], x2=df['TypeB_before'], value=0, alternative='larger'))  
# 귀무가설(H₀): 두 집단의 평균이 같다 (μ₁ - μ₂ = 0)
# 대립가설(H₁): TypeA_before의 평균이 TypeB_before보다 **크다** (μ₁ > μ₂)
# → 특정 방향(오른쪽 꼬리, 양의 방향)으로 차이가 있다고 가정함
# p-value가 0.05보다 작으면, TypeA가 TypeB보다 유의하게 크다고 판단함
```

### Z-test 결과 요약

| 검정 종류 | Z 통계량 | p-value | 해석 요약 |
| --- | --- | --- | --- |
| **양측검정** | -2.7895 | **0.0053** | 평균 차이 있음 (유의함) |
| **단측검정 (smaller)** | -2.7895 | **0.0026** | A가 B보다 작음 (유의함) |
| **단측검정 (larger)** | -2.7895 | **0.9974** | A가 B보다 크다는 증거 없음 (기각 못함) |

### 대응표본 t-test를 사용하는 이유

| 비교 유형 | 설명 |
| --- | --- |
| **대응표본 t-test** | **같은 대상의 전후 비교**, 예: Before vs After |
| 독립표본 t-test | 서로 다른 두 집단 비교, 예: A그룹 vs B그룹 |

```python
# 대응표본 t검정 수행

scipy.stats.ttest_rel(df['TypeA_before'], df['TypeA_after'])  
# scipy 라이브러리의 ttest_rel() 함수를 이용하여 대응표본(쌍체) t-검정을 수행함
# 대응표본이란 → 동일한 대상에게 두 번 측정한 데이터 (ex. 전/후, 처리 전/후, 좌/우 등)
# 여기서는 TypeA 골프공을 같은 조건에서 "Before"와 "After"로 측정한 결과 비교에 해당함

# ttest_rel() 함수는 다음과 같은 가설을 검정함:
#   ● 귀무가설(H₀): 두 시점의 평균 차이가 0이다 (즉, 변화가 없다)
#   ● 대립가설(H₁): 두 시점의 평균 차이가 0이 아니다 (즉, 변화가 있다)

# 입력 인자 설명:
# - df['TypeA_before']: 실험 전 측정값
# - df['TypeA_after']: 실험 후 측정값
# 두 시리즈는 반드시 **길이가 같고 1:1로 대응되는 구조**여야 함

# 반환값은 (t-통계량, p-value)의 튜플로 구성됨:
#   - t-통계량: 평균 차이 / 표준 오차 → 표본 차이가 얼마나 큰지를 나타내는 통계량
#   - p-value: 귀무가설을 기각할 확률 수준

# 해석 기준:
#   - p-value < 0.05 → 귀무가설 기각 → 평균 차이가 유의함
#   - p-value ≥ 0.05 → 귀무가설 채택 → 평균 차이가 유의하지 않음
```

## 결과 요약

| 항목 | 값 | 해석 |
| --- | --- | --- |
| **t 통계량** | -1.2214 | 평균 차이가 작고 통계적으로 유의하지 않다는 방향성 |
| **p-value** | 0.2278 | 귀무가설 기각 불가 (유의수준 0.05 기준) |
| **자유도(df)** | 49 | 샘플 수가 50개 → df = n - 1 = 49 |

→ 즉, TypeA의 Before와 After 사이의 평균 차이는 통계적으로 유의하지 않음

다시 말해, 관찰된 평균 변화는 우연일 가능성이 높다는 해석임

```python
# 독립표본 t검정 수행

ttest_ind(df['TypeA_before'], df['TypeB_before'], equal_var=False)  
# 서로 다른 집단 간 평균 차이를 검정하는 '독립표본 t-검정(independent t-test)'을 수행하는 명령문
# TypeA_before와 TypeB_before는 서로 독립된 두 집단(즉, 서로 다른 골프공 유형)에 해당함
# → 같은 실험 조건에서 두 골프공 타입 간의 평균 성능(예: 거리)의 차이가 유의미한지를 검정함

# equal_var=False는 두 집단의 분산이 서로 같지 않다고 가정하고 Welch의 t-검정을 수행하겠다는 설정임
# 등분산성 검정을 통해 이분산이라고 판단된 경우, 반드시 equal_var=False로 설정해야 함
# 이 설정은 분산이 다른 두 집단 간 평균 차이에 대한 보다 정확한 추정을 가능하게 함

# 반환값은 다음과 같음:
# - statistic: t-검정 통계량 → 표본 간 평균 차이가 표준 오차 대비 얼마나 큰지를 나타내는 수치
# - pvalue: 귀무가설을 기각할 확률 수준
#   → p-value < 0.05 이면 두 집단 간 평균이 유의미하게 다르다고 판단함

# 귀무가설(H₀): 두 집단의 평균이 같다 (μ₁ = μ₂)
# 대립가설(H₁): 두 집단의 평균이 다르다 (μ₁ ≠ μ₂)
# 이 검정은 양측 검정이 기본으로 설정되며, 방향성이 없는 평균 차이를 검정함
```

### 결과 요약

| 항목 | 값 | 해석 요약 |
| --- | --- | --- |
| **t 통계량** | -2.7895 | 두 집단 평균 차이가 음의 방향으로 유의미함 |
| **p-value** | 0.0064 | 귀무가설 기각 (유의수준 0.05 기준) |
| **자유도(df)** | 약 96.24 | Welch’s 방식으로 자동 계산된 자유도 |

p-value = 0.0064 < 0.05 → 귀무가설 기각

따라서, TypeA_before와 TypeB_before 간 평균 차이는 통계적으로 유의미함
