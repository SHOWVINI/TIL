# 📊 카이제곱 검정 (Chi-Square Test) – 교차분석 완전 정복

## ✅ 카이제곱 검정이란?

- *카이제곱 검정(χ² test)**은 **범주형 데이터(명목형 데이터)** 간의 **연관성(독립성)**을 분석하기 위한 통계 기법
- 특히 **두 변수 간의 관계가 우연인지, 실제로 연관이 있는지**를 판단하기 위해 사용
- 대표적인 용도가 **교차분석(Cross Tabulation Analysis)**

## 🧠 카이제곱 검정은 어떤 상황에서 사용하는가?

### 👉 예시 상황

- 성별(남/여)과 제품 구매 여부(구매/미구매)가 **관련 있는지** 알고 싶은 경우
- 직업군(학생/회사원/자영업)과 SNS 사용률(사용/비사용)이 **서로 관련 있는지** 판단하는 경우

👉 이런 경우는 **평균 비교가 아닌, 빈도(횟수)를 비교**하는 것이 핵심

👉 두 변수 사이의 **독립성 여부**를 통계적으로 검정하는 목적

## 🔍 가설 설정

| 구분 | 가설 내용 |
| --- | --- |
| **귀무가설(H₀)** | 두 변수는 서로 독립이다 (관련 없음) |
| **대립가설(H₁)** | 두 변수는 서로 독립이 아니다 (관련 있음) |

## 📐 분석 방식 개요

1. **관찰값(Observed)**: 실제 수집된 데이터의 빈도
2. **기대값(Expected)**: 두 변수가 독립이라면 나올 것으로 기대되는 빈도
3. **카이제곱 통계량(χ²)**: 관찰값과 기대값 사이의 차이를 계산

χ2=∑(관찰값−기대값)2기대값\chi^2 = \sum \frac{(관찰값 - 기대값)^2}{기대값}

χ2=∑기대값(관찰값−기대값)2

👉 이 값이 **클수록** 두 변수 사이에 관계가 있을 가능성이 높다는 의미

## 🧪 교차표(Cross Tab) 예시

| 성별 \ 구매여부 | 구매함 | 구매 안 함 | 합계 |
| --- | --- | --- | --- |
| 남성 | 30 | 70 | 100 |
| 여성 | 60 | 40 | 100 |
| 합계 | 90 | 110 | 200 |

→ 위와 같은 표를 바탕으로 **각 셀의 기대값**을 계산하고, 카이제곱 검정 실시

## 🔍 기대값 계산 방식

예: 남성-구매함 셀의 기대값

(행합계×열합계)전체 합계=100×90200=45\frac{(행 합계 \times 열 합계)}{\text{전체 합계}} = \frac{100 \times 90}{200} = 45

전체 합계(행합계×열합계)=200100×90=45

→ 이렇게 각 셀에 대해 기대값을 계산하여 χ² 통계량을 구함

## 📊 분석 결과 해석

- **p-value < 0.05** → 귀무가설 기각 → **두 변수는 관련 있음**
- **p-value ≥ 0.05** → 귀무가설 채택 → **두 변수는 독립**

## 🔎 사용 조건 (주의사항)

1. **자료는 범주형이어야 함** (예: 성별, 직업, 선호도 등)
2. 기대값이 **5 미만인 셀이 너무 많을 경우** 정확도가 떨어짐
    
    → 이 경우 **Fisher's Exact Test** 등을 사용

```python
# 필요한 패키지 설치 및 불러오기

import scipy.stats  
# scipy 라이브러리의 stats 모듈 전체를 불러오는 명령문
# t-검정, 카이제곱 검정, 정규성 검정 등 다양한 통계 분석 기능을 제공함

from scipy.stats import chi2_contingency  
# stats 모듈에서 교차분석(범주형 자료 간 연관성 검정)에 사용되는 chi2_contingency() 함수를 직접 불러오는 명령문
# 이 함수는 카이제곱 독립성 검정을 수행하며, 입력된 교차표(빈도표)를 기반으로 두 변수 간 연관성 존재 여부를 검정함
# 주로 범주형 자료에 대해 "두 변수 간의 독립 여부"를 확인하는 데 사용됨

import pandas as pd  
# 데이터프레임 기반의 자료 처리 및 분석을 위한 pandas 라이브러리를 불러오는 명령문
# csv 파일 읽기, 표 형태 데이터 정리, 그룹 연산 등의 작업에 필수적으로 사용됨

import matplotlib.pyplot as plt  
# 데이터 시각화를 위한 matplotlib 라이브러리의 pyplot 모듈을 불러오는 명령문
# 선 그래프, 막대 그래프, 히스토그램 등 다양한 형태의 시각적 표현을 지원함
# 분석 결과를 그래프로 보여주어 직관적인 인사이트 도출에 활용됨
```

| 라이브러리 | 역할 |
| --- | --- |
| `scipy.stats` | 통계 검정 (정규성, 독립성, 평균 비교 등) |
| `chi2_contingency` | 범주형 변수 간 독립성(관련성) 검정 |
| `pandas` | 데이터 읽기 및 테이블 형태 가공 |
| `matplotlib.pyplot` | 시각화 (막대그래프 등) |

```python
# 데이터 불러오기
# https://www.kaggle.com/datasets/sewonghwang/smoker
df = pd.read_csv("../../../main/datasets/smoker.csv")

# 데이터 샘플 확인
df.head()
```

```python
# 항목 별 집계

df.groupby(['sex','smoke'])['smoke'].count()  
# DataFrame(df)에서 'sex'와 'smoke' 두 범주형 변수를 기준으로 그룹화(groupby)하는 명령문
# 즉, 성별(sex)과 흡연 여부(smoke)의 모든 조합에 대해 데이터를 나누는 구조임

# 그 후 각 그룹 내에서 'smoke' 열의 값을 기준으로 데이터 개수를 세는 명령문
# .count()는 해당 그룹에 속한 행의 수(관측값 개수)를 반환함
```

```python
# 카이제곱 검정용 교차표 생성

crosstab = pd.crosstab(df.sex, df.smoke)  
# pandas의 crosstab() 함수를 사용하여 교차표(빈도표)를 생성하는 명령문
# 행(index)은 'sex'(성별), 열(columns)은 'smoke'(흡연 여부)로 지정함
# 각 셀에는 해당 성별-흡연여부 조합에 해당하는 관측값(빈도 수)이 정수로 채워짐
# 결과적으로 두 범주형 변수 간의 분포 구조를 2차원 표 형태로 나타냄

crosstab  
# 생성된 교차표를 출력하여 확인하는 명령문
# 이 표는 카이제곱 독립성 검정에 사용할 입력 데이터로 활용되며,
# '성별(sex)'과 '흡연 여부(smoke)'가 서로 독립적인지(무관한지)를 판단하는 데 쓰임
```

```python
# 성별 별 흡연자 수 시각화

%matplotlib inline  
# 주피터 노트북 환경에서 작성된 코드일 경우,
# 시각화 결과(그래프)를 코드 셀 아래에 바로 표시하기 위한 매직 명령어
# 주피터가 아닌 다른 환경(Python script, IDE 등)에서는 생략 가능함

crosstab.plot(kind='bar', figsize=(10,5))  
# 앞서 생성한 교차표(crosstab)를 수직 막대그래프로 시각화하는 명령문
# kind='bar'는 범주형 변수를 막대그래프 형태로 나타내겠다는 설정임
# x축에는 'sex' (성별: male, female), 
# 각 그룹의 막대는 'smoke' (흡연 여부: Non-Smoker, Smoker)에 해당하는 값의 길이를 나타냄
# figsize=(10,5)는 그래프 크기를 가로 10인치, 세로 5인치로 설정하는 명령문

plt.grid()  
# 그래프에 격자선을 추가하여 눈금 해석을 용이하게 하는 명령문
# 특히 막대 높이를 정확히 파악하거나 비교할 때 시각적 도움을 줌
```

### 📊 해석 가이드

1. **카이제곱 통계량 (Chi square)**
    - 출력된 chi2 값은 관측 빈도와 기대 빈도 간 차이가 어느 정도인지를 나타내며,
    - 값이 클수록 두 변수 간에 차이가 크며, 독립성을 부정할 근거가 됨.
2. **p-value**
    - 출력된 p-value는 귀무가설(두 변수는 서로 독립이다)을 기각할 확률을 나타내며,
    - 일반적으로 p-value가 0.05 미만인 경우 → 귀무가설 기각 → 두 변수 간에 통계적으로 유의한 연관성이 있음을 의미함.

```python
# 카이제곱 독립성 검정 수행

chiresult = chi2_contingency(crosstab, correction=False)  
# chi2_contingency() 함수는 주어진 교차표(crosstab)를 이용해
# 두 범주형 변수 간 독립성(혹은 연관성)의 유의성을 검정하는 통계 방법(카이제곱 독립성 검정)을 수행함
# correction=False는 연속성 보정(Yates correction)을 적용하지 않겠다는 설정임
# 반환값은 튜플로 구성되며, 순서대로 다음 정보를 포함함:
#   1. chi2: 검정 통계량 (Chi-square statistic)
#   2. p: p-value (유의확률)
#   3. dof: 자유도 (degrees of freedom)
#   4. expected: 각 셀에서 기대되는 빈도수 (expected frequencies under H₀)
# 이 검정을 통해 '성별(sex)'과 '흡연 여부(smoke)' 간의 연관성이 존재하는지를 통계적으로 판단할 수 있음

print('Chi square: {}'.format(chiresult[0]))  
# chi2_contingency() 함수의 반환값 중 첫 번째 항목(chi2 통계량)을 포맷팅하여 출력함
# chi2 통계량은 관측된 교차표와 기대치 간의 차이를 제곱합한 값을 의미하며,
# 값이 클수록 두 변수 간에 독립적이지 않다는 증거가 됨

print('P-value: {}'.format(chiresult[1]))  
# chi2_contingency() 함수의 반환값 중 두 번째 항목(p-value)을 포맷팅하여 출력함
# p-value는 귀무가설(두 변수는 서로 독립임)을 기각할 수 있는 확률적 기준으로, 
# 일반적으로 p-value가 0.05보다 작으면 귀무가설을 기각하고, 변수 간 연관성이 있다고 판단함
```

### 카이제곱 독립성 검정 결과 해석표

| 항목 | 값 | 해석 |
| --- | --- | --- |
| **검정 방법** | 카이제곱 독립성 검정 (Chi-square test of independence) | 두 범주형 변수(성별과 흡연 여부)가 서로 독립인지 확인하는 검정 |
| **카이제곱 통계량 (χ²)** | 7.81 | 관측값과 기대값 사이의 차이 정도를 수치화한 값값이 클수록 두 변수 간 독립이 아닐 가능성이 큼 |
| **자유도 (df)** | 1 | (행 수 - 1) × (열 수 - 1) → 2×2 표이므로 1 |
| **p-value** | 0.0052 | 귀무가설(두 변수는 독립이다)을 기각할 수 있는 확률 |
| **유의수준 (α)** | 0.05 | 일반적인 통계 분석 기준값 |
| **판단** | 0.0052 < 0.05 → **귀무가설 기각** | 두 변수는 독립이 아님 |
| **결론** | 성별(sex)과 흡연 여부(smoke) 간에 **통계적으로 유의한 연관성** 존재 |  |
