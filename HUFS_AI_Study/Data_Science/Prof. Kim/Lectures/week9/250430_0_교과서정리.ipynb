# 📘 12.1 통계 기반 분석 방법론

## 1. 장 전체 개요

이 장은 데이터 분석을 위한 대표적인 **통계적 방법론**들과 **머신러닝 기법**을 비교하며 소개하는 내용을 담고 있음. 여기서 말하는 통계 기반 분석 방법론이란, 데이터에 내재된 관계를 수학적으로 모델링하고 해석하는 기법들을 의미함. 주요 방법론으로는 다음이 포함됨:

- 주성분 분석(PCA)
- 요인 분석
- Z-test, T-test
- ANOVA (분산분석)
- 카이제곱 검정

또한, 머신러닝과 통계 모델의 **사상적 차이**, **분석 목적**, **활용 맥락**에 따라 두 방법론을 구분하여 설명함.

## 2. 통계 모델과 머신러닝 모델의 개요

### ✅ 통계 모델(Statistical Models)

- **목적**: 현상을 설명하거나 해석하기 위한 도구
- **특징**:
    - 모형의 논리적 해석 가능성 중시
    - 오차와 불확실성에 대한 고려
    - 변수 간 관계 해석 중심
- **활용 예시**: 가설 검정, 변수 간 상관성 검토, 요인 추출 등

### ✅ 머신러닝(Machine Learning)

- **목적**: 예측 정확도 극대화
- **특징**:
    - 데이터 기반 자동 학습
    - 모델 복잡도 허용 가능
    - 해석보다는 성능 중시
- **활용 예시**: 이미지 분류, 사용자 추천, 매출 예측 등

> ⚠️ 통계 모델과 머신러닝은 완전히 별개가 아니며, 기본 원리는 통계에서 파생됨. 따라서 통계 이론에 대한 기반 지식을 습득하는 것이 머신러닝 실무 성과 향상에 기여함.
> 

## 3. 데이터 분석 방법론 분류 기준

### 기준 1: 종속 변수 유무

- **지도 학습(Supervised Learning)**: 정답(라벨)이 존재함 → 회귀, 분류
- **비지도 학습(Unsupervised Learning)**: 정답 없음 → 군집, 차원 축소
- **강화 학습(Reinforcement Learning)**: 보상 피드백 기반 학습

### 기준 2: 독립/종속 변수의 척도 수준

- **질적(범주형)**: 명목형, 순서형
- **양적(수치형)**: 구간형, 비율형

분석 목적과 변수 형태에 따라 적용 가능한 방법론이 달라짐.


## 4. 분석 방법론 개요표 설명

| 구분 | 독립 변수 | 종속 변수 | 방법론 (통계 모델) | 용도 |
| --- | --- | --- | --- | --- |
| 통계 모델 | 질적 | 질적 | 교차분석, 스피어만 상관 | 연관성 분석 |
| 통계 모델 | 질적 | 양적 | Z-test, T-test | 가설 검정 |
| 통계 모델 | 질적 | 양적 | ANOVA, MANOVA | 분산 분석 |
| 통계 모델 | 양적 | 양적 | 피어슨 상관 | 연관성 분석 |

| 구분 | 독립 변수 | 종속 변수 | 방법론 (머신러닝) | 용도 |
| --- | --- | --- | --- | --- |
| 지도학습 | 질적/양적 | 질적 | 로지스틱 회귀, 분류나무, 랜덤포레스트 등 | 분류 분석 |
| 지도학습 | 질적/양적 | 양적 | 선형회귀, 회귀나무, 신경망 등 | 예측 분석 |
| 비지도학습 | 질적/양적 | 없음 | k-means, PCA, 요인분석, SOM 등 | 군집/차원축소 분석 |
| 강화학습 | 없음 | 없음 | Model-free RL, Model-based RL | 보상 기반 학습 |

→ 동일 방법론이 질적/양적 모두에 사용될 수 있는 경우도 있음 (예: 회귀, k-NN 등)


## 5. 각 학습 형태별 분석 방식

### ▶ 지도학습(Supervised Learning)

- **정의**: 입력 데이터와 함께 해당 정답이 주어지는 학습 방식
- **분류 분석 (Classification)**: 종속변수가 질적 척도일 경우 → 예: 고객 이탈 여부
- **회귀 분석 (Regression)**: 종속변수가 양적 척도일 경우 → 예: 아파트 가격 예측

### ▶ 비지도학습(Unsupervised Learning)

- **정의**: 정답 없이 데이터 내 숨겨진 구조 파악
- **차원축소(Dimensionality Reduction)**:
    - 주성분분석(PCA), 요인분석
    - 목적: 학습 성능 향상을 위한 전처리
- **군집분석(Clustering)**:
    - k-means, SOM
    - 목적: 유사 특성 그룹화 (예: 고객 세분화)
- **연관규칙(Association Rule)**:
    - 장바구니 분석, 추천시스템에 활용
    - ex) A를 구매한 고객이 B도 구매할 확률

### ▶ 강화학습(Reinforcement Learning)

- **정의**: 보상/벌을 통해 최적의 행동 전략 학습
- **예시**: 알파고의 바둑 경기 학습, 로봇 제어 등

# 📘 12.2 주성분 분석 (Principal Component Analysis, PCA)

## 1. 주성분 분석(PCA)의 개념

주성분 분석(PCA)은 다차원 데이터를 보다 **간결하게 표현하기 위한 차원 축소 기법**임. 여러 개의 독립 변수들 중에서 데이터의 분산을 가장 잘 설명하는 **새로운 성분(주성분)**을 추출하여, 전체 변수를 소수의 핵심 변수로 요약함.

- **목적**: 정보 손실을 최소화하면서 변수의 수를 줄이는 것
- **이점**:
    - 차원의 저주(Curse of Dimensionality) 방지
    - 모델 단순화
    - 해석 용이성 향상



## 2. 차원의 저주란 무엇인가?

**차원의 저주**란, 변수(차원)가 많아짐에 따라 분석을 위해 필요한 데이터 건수도 급격히 증가하게 되는 현상을 의미함.

예를 들어, 1개의 변수당 30개의 관측치가 필요하다고 가정하면, 20개의 변수는 최소 600개의 관측치가 필요하게 됨.

또한 차원이 커질수록 다음과 같은 문제가 발생함:

- **과적합(Overfitting)** 위험 증가
- **다중공선성(Multicollinearity)** 문제 발생 가능

이런 문제들을 예방하기 위해 **차원 축소(PCA)**가 활용됨.



## 3. PCA의 핵심 원리

PCA는 데이터의 분산을 **최대한 보존하는 새로운 축(성분)**을 찾는 것이 핵심임.

1. 원 데이터의 변수들로부터 **새로운 주성분 축**을 찾음
2. 이 축을 기준으로 데이터를 투영하여, 기존 변수 대신 새 성분으로 표현함
3. 보통 **설명력이 높은 상위 몇 개 성분만 선택**하여 차원을 줄임

> 예: 원래 9개의 변수 → 설명력이 80% 이상인 2개의 주성분으로 요약 가능
> 



## 4. 주성분의 수와 설명력

- PCA는 데이터의 차원 수만큼의 **주성분**을 생성함
- 그러나 설명력이 높은 주성분 몇 개만 선택하여 사용함
- *설명력(Explained Variance)**이란, 해당 주성분이 전체 데이터 분산 중 얼마나 많은 비율을 차지하는지를 나타냄

예시:

- 제1주성분의 설명력: 45.4%
- 제2주성분의 설명력: 18%
- 누적 설명력: 63.4% → 대부분의 정보 요약 가능



## 5. 요약 정리

| 항목 | 내용 |
| --- | --- |
| 목적 | 변수 수 축소 및 정보 손실 최소화 |
| 주요 개념 | 분산 최대화 축 찾기, 상위 주성분 선별 |
| 전제 조건 | 정규 분포, 변수 간 독립성, 양적 변수 |
| 장점 | 모델 단순화, 시각화 용이, 다중공선성 문제 해결 |
| 사용 전 데이터 전처리 | 정규화 또는 표준화 필수 |
| 주성분 개수 결정 | 설명력 기준 (예: 누적 설명력 80% 이상 확보 시 종료) |

# 📘 12.3 공통 요인 분석 (Common Factor Analysis, CFA)

## 🧩 1. 개념 정리

### ✅ 1.1 요인 분석이란?

- 여러 관측 변수를 요약하여 그 배후에 존재하는 **숨겨진 공통된 요인**(latent factor)을 찾고자 하는 기법
- 요인 분석의 목적: **데이터 축소**와 **요인 구조 규명**

### ✅ 1.2 요인 분석의 종류

| 분류 | 설명 |
| --- | --- |
| 주성분 분석 (PCA) | 전체 분산을 기준으로 성분을 축출함 |
| 공통 요인 분석 (CFA) | 공통 분산(상관관계 기반)만 고려하여 요인을 추출함 |
| 탐색적 요인 분석 (EFA) | 요인 간 관계를 **미리 정하지 않고** 탐색함 |
| 확인적 요인 분석 (CFA) | 요인 구조를 **미리 가정**하고 검증함 |

**중요**: 여기서의 CFA는 *공통 요인 분석 (Common Factor Analysis)*를 뜻함. *확인적 요인 분석 (Confirmatory Factor Analysis)*와 혼동 주의.


## ⚙️ 2. PCA와 CFA의 차이

| 구분 | PCA | CFA (공통 요인 분석) |
| --- | --- | --- |
| 분산 기준 | 전체 분산 | 공통 분산 |
| 목적 | 데이터 요약, 축소 | 잠재 요인 구조 해석 |
| 요인 추출 방식 | 주성분(설명력 순) | 요인 간 우열 없음, 구조 중심 |
| 결과 해석 | 전체 변수 설명력 기준 | 변수 묶음 간 구조 파악 |
| 적합도 평가 | 없음 | Bartlett 검정, KMO 필요 |


## 🧪 3. 공통 요인 분석 절차

### 3.1. 데이터 전처리

- 문자형 변수 제거
- 정규화 또는 표준화 적용 (MinMaxScaler 등)

### 3.2. 요인 분석 가능성 진단

| 검정 | 목적 | 판단 기준 |
| --- | --- | --- |
| Bartlett의 구형성 검정 | 상관관계 존재 여부 검정 | p-value < 0.05 → 적합 |
| KMO 검정 | 상관 구조의 설명 가능성 판단 | 0.6 이상 → 적합 |

### 3.3. 요인 수 결정

- **고유값(Eigenvalue) ≥ 1**인 요인 수 선택
- **누적 설명력 ≥ 60%** 확보
- **Scree Plot** 시각화로 elbow point 확인

## 🔍 4. 요인 분석 결과 해석

### ✅ 4.1 요인 적재값 (Factor Loading)

- 요인과 변수 간 상관 정도
- 절댓값 |0.3| 이상이면 유의미한 관계
- |0.5| 이상이면 해당 요인을 대표하는 주요 변수로 간주

예:

| 변수명 | F1 | F2 |
| --- | --- | --- |
| 담배 소비량 | 0.84 | -0.003 |
| 건강식 소비량 | 0.09 | 0.77 |

→ F1: 기호식품 소비 요인, F2: 건강식 소비 요인

### ✅ 4.2 시각화 (Heatmap)

- 요인 적재값을 색상으로 표현
- 절댓값 기반 시각화로 상관 강도 직관적으로 확인 가능


## 🧪 5. 실습 요약 (NBA 데이터)

1. 문자형 변수 제거 후 정규화
2. Bartlett test: p-value = 0.0 → 분석 적합
3. KMO test: 0.70 → 준수한 수준
4. 고유값 분석: 4개의 요인 결정
5. 각 요인과 변수 간 적재값 확인
6. 시각화로 요인별 특징 변수 도출


## ✅ 최종 요약

| 항목 | 설명 |
| --- | --- |
| CFA의 목적 | 유사한 성격의 변수들을 묶어 잠재 요인을 도출 |
| 적합성 확인 | Bartlett, KMO 검정을 통해 상관 구조 평가 |
| 요인 수 결정 | 고유값 ≥ 1, 누적 설명력 ≥ 60%, Scree plot |
| 결과 해석 | 요인 적재값을 기준으로 요인별 의미 해석 |

# 📘 12.4 다중공선성 해결과 섀플리 밸류 분석

## 📌 1. 다중공선성(Multicollinearity)의 개념

- **정의**: 다중공선성이란 **독립변수들 사이의 상관관계가 높은 현상**을 의미함.
- **문제 발생**: 독립변수들이 서로 선형적 관계를 가지면 회귀분석의 전제인 독립성 가정을 위배하게 됨.
- **결과적 문제점**:
    - 회귀계수의 통계적 유의성이 감소함.
    - 해석의 어려움과 추정치의 불안정성이 증가함.
    - 예측성능이 과장되어 나올 수 있음.

> 📌 예시: '마신 술의 양'과 '혈중 알코올 농도'는 서로 강한 상관관계를 가지므로 함께 회귀모델에 넣을 경우 다중공선성이 발생함.
> 

## 📌 2. 다중공선성 판별 방법

### 2.1 상관계수 분석

- 독립변수 간의 상관계수 절댓값이 **0.7 이상**일 경우 다중공선성이 의심됨.
- 변수 개수가 많을 경우 해석의 어려움이 존재함.

### 2.2 회귀분석 내 t-값 확인

- **결정계수(R²)는 높지만**, 회귀계수의 **t-값이 작을 경우** 다중공선성 가능성 존재.
- **t-값 기준** (신뢰수준 95%):
    - |t| ≥ 1.96 → 유의
    - |t| < 1.96 → 다중공선성 의심

### 2.3 VIF(Variance Inflation Factor) 확인

- **정의**: 특정 변수 Xₖ가 다른 독립변수들에 의해 얼마나 잘 설명되는지를 수치화한 값.
- **수식**: VIFₖ = 1 / (1 - R²ₖ)
- **해석 기준**:
    - VIF ≥ 5 → 다중공선성 의심
    - VIF ≥ 10 → 다중공선성 있음
- √VIF는 해당 변수의 표준오차가 증가한 배수를 의미함.
    - 예: VIF = 16 → 표준오차가 4배 증가함.

## 📌 3. 다중공선성 해결 방법

| 방법 | 설명 |
| --- | --- |
| 변수 제거 | VIF는 높고 종속변수와의 상관성은 낮은 변수를 제거함. 단, 이론적 중요성을 고려함. |
| 표본 수 증가 | 데이터 수를 늘리면 추정 안정성이 향상됨. 현실적 어려움 존재함. |
| 변수 가공 | 로그 변환, 표준화, 정규화, 범주화 등을 통해 상관관계를 약화시킴. |
| PCA(주성분 분석) 활용 | 상관성이 높은 변수를 하나의 주성분으로 축약함. 해석이 어려울 수 있음. |
| 변수 선택 알고리즘 사용 | 전진 선택, 후진 제거, 단계적 선택 등의 자동화된 방법 사용 가능함. |

## 📌 4. 섀플리 밸류 분석 (Shapley Value Analysis)

### 4.1 개념

- **정의**: 변수의 조합별 기여도를 기반으로, **각 변수의 순수한 설명력(기여도)을 정량화**하는 방법.
- **유래**: 안정적 결혼 문제로 유명한 로이드 섀플리의 협력게임 이론 기반 기법.

### 4.2 계산 예시

### 설명력 조합 예시:

| 변수 조합 | 설명력 |
| --- | --- |
| x₁ | 0.15 |
| x₂ | 0.21 |
| x₃ | 0.17 |
| x₁ + x₂ | 0.32 |
| x₁ + x₃ | 0.29 |
| x₂ + x₃ | 0.31 |
| x₁ + x₂ + x₃ | 0.35 |

### x₁의 섀플리 밸류 계산:

1. 단일 투입: 0.15
2. 2개 조합에서의 평균 기여도:
    - x₁ + x₂ 조합에서 x₁ 기여도: 0.32 - 0.21 = 0.11
    - x₁ + x₃ 조합에서 x₁ 기여도: 0.29 - 0.17 = 0.12
    - 평균: (0.11 + 0.12) / 2 = 0.115
3. 3개 조합에서의 기여도:
    - x₁ + x₂ + x₃에서 x₁ 기여도: 0.35 - 0.31 = 0.04
4. 최종 섀플리 밸류:
    - (0.15 + 0.115 + 0.04) / 3 = **0.101**

> 즉, x₁의 순수한 평균 설명력은 0.101이며, 이 값이 섀플리 밸류가 됨.
> 

## 📌 5. 섀플리 밸류의 의미와 활용

- **모든 변수의 섀플리 밸류 합 = 전체 모델 설명력**
- **해당 변수의 실제적 기여도 정량화 가능**
- **변수 중요도 판단, 변수 선택, 모델 해석력 향상에 매우 유용함**

| 항목 | 요약 설명 |
| --- | --- |
| 다중공선성 | 독립변수들 간의 높은 상관관계 |
| 문제점 | 회귀계수 불안정, 해석력 저하 |
| 판별 기준 | 상관계수, t-값, VIF 등 |
| 해결법 | 변수 제거, 가공, PCA, 알고리즘 선택 등 |
| 섀플리 밸류 | 각 변수의 순수 기여도 정량화 지표 |

# 📘 12.5 데이터 마사지와 블라인드 분석

## 🧪 12.5.1 데이터 마사지 (Data Massage)

### ✅ 정의

- **데이터 마사지**란 데이터 분석 결과가 예상하거나 의도한 방향과 다를 경우, **데이터의 배열이나 관점을 바꾸어 해석이 달라지도록 유도하는 행위**를 의미함.
- 이는 **데이터의 수치 자체를 조작하는 것과는 구별**되며, **해석의 방향을 변경하는 간접적 조작**에 해당함.
- 하지만 분석가의 **주관적 판단이 개입되므로 지양해야 할 행위**에 해당함.

### 🔎 대표적인 데이터 마사지 기법

| 구분 | 설명 |
| --- | --- |
| 편향된 전처리 | 이상치, 결측치를 의도적으로 조정하여 특정 그룹에 유리하게 처리함. |
| 매직 그래프 | 시각화에서 눈속임 기법을 사용해 차이를 과장하거나 축소함. |
| 분모 변경 | 비율의 분모를 바꿔 해석을 유도함. |
| 데이터 누락 | 특정 연도나 데이터를 의도적으로 제외하여 결과를 왜곡함. |
| 파라미터 변경 | 머신러닝 모델의 설정을 반복적으로 바꾸어 의도한 결과를 도출함. |
| 심슨의 역설 | 세부 집단 비중의 차이를 활용해 전체 통계를 왜곡함. |

### 📉 예시 설명

### ① **매직 그래프**

- 그래프에서 y축의 간격을 비정상적으로 조정해 실제보다 차이가 커 보이거나 작아 보이게 함.
- 예: FPS(초당 프레임 수) 1 차이를 강조하기 위해 y축 최소값을 83으로 설정함 → 큰 차이처럼 보이게 만듦.

### ② **분모 변경**

- 위스키 숙성 시 증발 비율을 "전체 대비"가 아닌 "남은 양 대비"로 설정하면 138%처럼 과장 표현 가능함.

### ③ **의도적 누락 및 가공**

- 2014~2021년 물가 상승률 중 일부 연도만 선택하거나 평균 구간을 조정하여 ‘상승세’ 혹은 ‘하락세’를 인위적으로 만들어낼 수 있음.

### ④ **모델 파라미터 변경**

- 머신러닝 알고리즘의 하이퍼파라미터를 반복적으로 조정해 원하는 패턴이나 결과를 유도함.
- 데이터의 복잡성을 이용한 교묘한 결과 조정의 예에 해당함.

## 🧭 12.5.2 블라인드 분석 (Blind Analysis)

### ✅ 정의 및 목적

- **블라인드 분석**은 데이터 마사지를 방지하기 위한 분석 설계 방식으로, **분석 과정에서 분석가의 주관이나 기대가 결과에 영향을 미치는 것을 최소화**하고자 함.
- **기존 가설이나 지식, 선입견으로부터 해석이 왜곡되지 않도록** 통계적 중립성을 지키기 위한 분석 방법임.

### 🧠 왜 필요한가?

| 요소 | 설명 |
| --- | --- |
| 인지적 편향 | 분석가의 기존 경험이나 기대가 무의식적으로 해석에 영향을 주는 현상 |
| 확증 편향 | 기존 믿음을 뒷받침하는 증거만 수용하고 반대 증거는 무시하는 경향 |
- 이러한 편향을 제거하지 않으면, 중요한 변수를 무시하거나 중요하지 않은 변수에 과도한 의미를 부여할 수 있음.

### 🧪 블라인드 분석 수행 방법

- **변수 명칭과 의미를 숨긴 채** 분석을 수행함.
- 종속변수(Y)와 독립변수(X) 사이의 관계만 파악하고, **해당 변수의 실제 이름이나 의미는 보지 않음.**
- 분석 결과만 보고 통계적으로 유의미한 변수나 패턴을 해석함.
- 이후 변수의 이름을 확인하고 결론을 도출함.


### 🧊 예시: 블라인드 테스트처럼 생각

- 음료 브랜드의 **고객 선호도 조사** 시 브랜드를 가린 채 맛만 비교하는 테스트를 수행함.
- 마찬가지로, **변수의 이름을 가리고 결과만으로 판단**하여 분석가의 인지 편향을 최소화함.

### 🧩 블라인드 분석의 한계와 주의점

- **완벽한 편향 제거는 불가능함.**
- 오히려 **유의미한 변수를 놓치거나 잘못된 결론에 도달할 가능성**도 존재함.
- 분석 결과가 기대와 다를 경우 **'실패'라고 단정짓지 않고 재검토할 수 있도록 도와주는 보조 수단**으로 이해해야 함.

## ✅ 요약 정리

| 항목 | 요약 설명 |
| --- | --- |
| 데이터 마사지 | 데이터 해석 방향을 의도적으로 유도하는 행위 |
| 주요 기법 | 편향된 전처리, 그래프 왜곡, 분모 변경, 데이터 누락 등 |
| 블라인드 분석 | 변수의 의미를 가리고 분석하여 편향을 줄이는 방법 |
| 목적 | 분석가의 인지적 편향과 확증 편향을 방지함 |
| 주의 | 과도한 의존은 오히려 해석력을 저하시킬 수 있음 |

# 📘 12.6 Z-test와 T-test

## 1. 개념 개요

- **Z-test와 T-test**는 “두 집단 간 평균값 차이가 통계적으로 유의미한지” 검정하기 위한 통계 기법임.
- 이 절에서 배우는 검정 기법들은 모두 **귀무가설을 기각할 수 있는가**를 판별하기 위한 방법임.

### 예시 상황

- 예: A지역 고객 평균 매출이 67,000원, B지역은 68,500원일 때, 이 1,500원 차이가 우연인지 아닌지를 검정하려면 Z-test 또는 T-test를 사용함.

## 2. Z-test vs T-test vs ANOVA 차이

| 구분 | 사용 목적 | 사용 조건 |
| --- | --- | --- |
| **Z-test** | 평균 차이 분석 (주로 단일/두 집단) | 표본 크기 ≥ 30, 모분산을 알고 있어야 함 |
| **T-test** | 평균 및 비율 차이 분석 (단일/두 집단) | 표본 크기 < 30 혹은 분산 미상 시 사용 가능 |
| **ANOVA** | 2개 이상 집단의 평균/비율 차이 분석 | 집단 수 ≥ 2개, 표본 크기 ≥ 30 |
- **Z-test와 T-test는 2개 집단 비교용**, **ANOVA는 2개 이상 집단 비교용**임.

## 3. T-test 상세 조건 및 수식

### 사용 조건

- 데이터는 **정규 분포**를 따라야 함.
- 두 집단은 **등분산** 조건을 충족해야 함. → `bartlett test` 사용

### 단일 표본 T-test 수식

t=Xˉ−μs/nt = \frac{\bar{X} - \mu}{s / \sqrt{n}}

t=s/n

Xˉ−μ

- Xˉ\bar{X}Xˉ: 표본 평균
- μ\muμ: 모평균 (귀무가설 기준 평균)
- sss: 표준편차
- nnn: 표본 크기

## 4. 실제 적용 예시 – 단일 집단 평균 차이 검정

### 시나리오

- **C 쇼핑몰 고객**의 기존 월 평균 매출: 42,000원
- 마케팅 프로그램 적용 후: 43,000원
- **표본 수**: 100명
- **표준편차**: 6,250
- **표준오차**: 625

- H0H_0H0: 두 평균이 같다 (차이 없음)
- H1H_1H1: 두 평균이 다르다 (차이 있음) → 양측 검정 사용

### 검정 통계량 계산

t=43000−42000625=1.6t = \frac{43000 - 42000}{625} = 1.6

t=62543000−42000=1.6

- → **t값 1.6은 임계값보다 작기 때문에 귀무가설 채택**

## 5. Z-test와 T-test 공통 비교

- **Z-test 수식** (모표준편차를 알고 있을 때 사용):

Z=Xˉ−μσ/nZ = \frac{\bar{X} - \mu}{\sigma / \sqrt{n}}

Z=σ/n

| --- | --- | --- |

# 📘 12.7 ANOVA (Analysis of Variance)

## 1. 개념 개요

- *ANOVA(분산분석)**는 **세 개 이상의 집단 평균 간 차이가 유의미한지 검정**하는 통계 기법임.
- **T-test는 두 집단의 평균 차이**를 비교하지만, **ANOVA는 세 집단 이상일 경우**에 적합함.
- 예를 들어 A, B, C 세 집단이 있을 때, 각각 A-B, B-C, A-C의 t-test를 개별 수행하면 1종 오류(잘못된 기각)의 누적 확률이 증가함. 이를 방지하기 위해 ANOVA를 사용함.

## 2. 분석 원리

- ANOVA는 **집단 간 평균값 차이**를 **집단 내 분산과 비교**하여 판단함.
- 사용하는 분포는 **F 분포**이며, 이는 항상 양수이며 오른쪽으로 꼬리가 긴 형태를 가짐.
- **F 검정 통계량 = 집단 간 분산 / 집단 내 분산**
- 이 비율이 클수록 **집단 간 차이가 크다**고 판단할 수 있음.

### 용어 정의

| 용어 | 설명 |
| --- | --- |
| **집단 간 분산 (MS_between)** | 각 집단 평균이 전체 평균에서 얼마나 떨어져 있는지를 나타내는 분산 |
| **집단 내 분산 (MS_within)** | 각 집단 내부 관측값들이 자기 집단 평균에서 얼마나 떨어져 있는지를 나타내는 분산 |
| **F 값** | MS_between / MS_within |

## 3. 가설 설정

- **귀무가설(H₀)**: 집단 간 평균에 차이가 없다. (모든 집단 평균이 동일하다.)
- **대립가설(H₁)**: 적어도 하나의 집단 평균이 다르다.

## 4. ANOVA 유형

| 유형 | 설명 |
| --- | --- |
| **일원 분산분석 (One-Way ANOVA)** | 독립변수가 하나일 때 사용 (예: '지역'만 분석) |
| **이원 분산분석 (Two-Way ANOVA)** | 독립변수가 두 개일 때 사용 (예: '지역', '연령대') |
| **다원 분산분석 (N-Way ANOVA)** | 독립변수가 셋 이상일 때 사용 |
- 본 강의에서는 **일원 분산분석**을 중심으로 다룸.

## 5. 변수 조건

| 변수 종류 | 조건 |
| --- | --- |
| **독립변수** | 범주형 (ex. 연령대, 지역) |
| **종속변수** | 연속형 (ex. 판매량, 점수, 거리) |
- 독립변수가 연속형이면 **회귀분석**, 종속변수가 범주형이면 **교차분석**으로 전환해야 함.

## 6. ANOVA 공식

> F = MS_between / MS_within
> 
- SS_total: 전체 제곱합 = SS_between + SS_within
- df_total = n - 1
- df_between = k - 1
- df_within = n - k

| 분산의 출처 | 제곱합(SS) | 자유도(df) | 평균제곱(MS) | F값 |
| --- | --- | --- | --- | --- |
| 집단 간 | SS_between | k - 1 | SS_between / (k - 1) | F = MS_between / MS_within |
| 집단 내 | SS_within | n - k | SS_within / (n - k) |  |
| 총합 | SS_total | n - 1 | - | - |

## 7. 예시 데이터: SNS 사용 횟수

- 20대, 30대, 40대 각 연령대 3명씩의 **SNS 앱 실행 횟수** 측정
- 전체 평균: 15회

### 집단 간 분산 (SS_between):

각 집단 평균과 전체 평균 차이의 제곱에 인원 수를 곱해 더함

→ SS_between = 63

### 집단 내 분산 (SS_within):

각 집단 내부의 편차 제곱합

→ SS_within = 66

### F 통계량:

- MS_between = 63 / 2 = 31.5
- MS_within = 66 / 6 = 11
- **F = 31.5 / 11 ≈ 5.73**

### 유의수준 검정:

- F 임계값 5.14보다 크므로 → **귀무가설 기각**, 집단 간 평균 차이가 유의함.

## 8. 사후 검정 (Post-hoc Test)

- ANOVA는 집단 간 차이 존재 여부만 알려줌 → 어떤 집단 간에 차이가 있는지 알 수 없음.
- 이를 확인하기 위해 **사후검정(Tukey HSD 등)** 수행함.

| 사후검정 방법 | 조건 |
| --- | --- |
| **Tukey's HSD** | 각 집단의 샘플 수가 같을 때 |
| **Scheffé 검정** | 샘플 수가 다를 때 |

### Tukey 검정 결과 해석 예시

| 비교 조합 | 평균 차이 | p값 | 유의성 여부 |
| --- | --- | --- | --- |
| A vs B | 5.14 | 0.0129 | ✅ 차이 있음 |
| A vs C | 1.90 | 0.5374 | ❌ 없음 |
| B vs C | -3.24 | 0.9965 | ❌ 없음 |

→ **A와 B 사이에만 통계적으로 유의한 차이**가 존재함

## 🔚 정리

- **ANOVA는 세 집단 이상에서 평균 차이를 검정하는 데 필수적**임.
- **F 분포 기반 통계량**을 통해 차이의 유의성을 판단함.
- **사후 검정**을 통해 실제 차이가 발생한 집단을 식별할 수 있음.
- 실습에서는 `scipy.stats`, `statsmodels.ols`, `pairwise_tukeyhsd` 등 다양한 패키지를 활용할 수 있음.

# 📘 12.8 카이제곱 검정 (Chi-square test)


## 1️⃣ 개념 개요

- 카이제곱 검정은 **범주형 변수 간의 연관성(독립성 여부)**을 분석하기 위한 통계 기법임.
- **명목형(예: 성별, 흡연 여부)** 또는 **서열형(예: 학력 수준)** 변수들이 대상이며, **연속형 변수**는 적절한 구간으로 나눈 뒤 범주형으로 변환하여 사용함.
- 교차표(Crosstab)를 기반으로 두 변수의 조합에 따라 관측 빈도와 기대 빈도를 계산하고, 두 값 간의 차이를 기반으로 **귀무가설(H₀: 두 변수는 독립임)**을 검정함.

> ✅ 본 검정은 Pearson의 상관계수처럼 연관성의 ‘강도’를 수치로 나타내지 않고,
> 
> 
> **단순히 두 변수 간에 통계적으로 의미 있는 연관이 있는지를 판단**하는 데 초점이 있음.
> 

## 2️⃣ 분석 절차 요약

1. **데이터 구성**: 두 개의 범주형 변수 준비
2. **교차표 생성**: 범주별 관측 빈도 산출
3. **기대 빈도 계산**: 각 셀의 기대 빈도 계산
4. **카이제곱 통계량 계산**: χ2=∑(관측−기대)2기대\chi^2 = \sum \frac{(관측 - 기대)^2}{기대}χ2=∑기대(관측−기대)2
5. **자유도 계산**: (행개수−1)×(열개수−1)(행 개수 - 1) \times (열 개수 - 1)(행개수−1)×(열개수−1)
6. **유의수준과 비교**: χ2\chi^2χ2 값이 임계값을 초과하는지 확인 → 귀무가설 기각 여부 판단


## 3️⃣ 기대빈도 계산 방식

|  | 흡연 O | 흡연 X | 행 합 |
| --- | --- | --- | --- |
| 남성 | A | B | A+B |
| 여성 | C | D | C+D |
| 열 합 | A+C | B+D | N=A+B+C+D |
- 셀별 기대 빈도 계산법:기대(A)=N(A+B)×(A+C)기대(B)=N(A+B)×(B+D)... etc.
    
    기대(A)=(A+B)×(A+C)N기대(A) = \frac{(A + B) \times (A + C)}{N}
    
    기대(B)=(A+B)×(B+D)N... etc.기대(B) = \frac{(A + B) \times (B + D)}{N} \quad \text{... etc.}
    


## 4️⃣ 통계량 계산 공식

χ2=∑(관측빈도−기대빈도)2기대빈도\chi^2 = \sum \frac{(관측빈도 - 기대빈도)^2}{기대빈도}

χ2=∑기대빈도(관측빈도−기대빈도)2

- 모든 셀에 대해 위 공식을 적용 후 합산
- 자유도: df=(행개수−1)×(열개수−1)df = (행 개수 - 1) \times (열 개수 - 1)df=(행개수−1)×(열개수−1)
- 유의수준 α = 0.05 → 해당 자유도 기준의 임계값과 비교하여 유의성 판단
