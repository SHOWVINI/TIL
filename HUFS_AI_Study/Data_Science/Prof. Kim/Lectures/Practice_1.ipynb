```python
import requests
from bs4 import BeautifulSoup

url = 'https://n.news.naver.com/mnews/article/023/0003893596'

response = requests.get(url)

if response.status_code == 200:
    html = response.text
    soup = BeautifulSoup(html, 'html.parser')
    #title = soup.select_one('h2.media_end_head_headline').text
    title = soup.select_one('h2#title_area').text
    print("뉴스 제목:",title)
else :
    print(response.status_code)
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time
import os

# 임시 사용자 데이터 디렉토리 설정
user_data_dir = os.path.join(os.getcwd(), "chrome_user_data")  # 현재 디렉토리에 사용자 데이터 저장

# Selenium 옵션 설정
options = webdriver.ChromeOptions()
options.add_argument(f"--user-data-dir={user_data_dir}")  # 사용자 데이터 충돌 방지
options.add_argument("--headless")  # 브라우저 없이 실행
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

# Chrome WebDriver 실행
service = Service(ChromeDriverManager().install()) # This line is replaced by the code above
driver = webdriver.Chrome(service=service, options=options)

try:
    # 웹사이트 이동
    driver.get("https://n.news.naver.com/mnews/article/023/0003893596")
    time.sleep(2)  # 페이지 로딩 대기

    # 뉴스 제목 가져오기
    news_title = driver.find_element("css selector", "h2.media_end_head_headline").text
    print("뉴스 제목:", news_title)

except Exception as e:
    print("에러 발생:", e)

finally:
    driver.quit()  # WebDriver 종료
```
