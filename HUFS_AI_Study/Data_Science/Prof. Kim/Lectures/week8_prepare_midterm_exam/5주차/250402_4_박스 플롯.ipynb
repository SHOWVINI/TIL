```python
# Profit 변수로 기본 가로 세로 박스 플롯 시각화

# 세로 박스 플롯
plt.figure(figsize=(8, 6))
# 출력할 그래프의 가로 너비를 8인치, 세로 높이를 6인치로 설정함
# 수직 방향의 박스 플롯에 적절한 비율을 지정함

sns.boxplot(y='Profit', data=df)
# seaborn 라이브러리의 boxplot 함수를 사용하여 수직 방향 박스 플롯을 생성함
# y축에 'Profit' 변수를 할당하여 이익 분포를 상자 그림으로 표현함
# 박스 플롯은 중앙값, 사분위수, 최대/최소 값, 이상치를 시각적으로 요약함

plt.show()
# 지정된 박스 플롯을 실제로 출력함
# Jupyter 환경에서는 자동 출력되기도 하나, 명시적으로 호출하여 확실히 렌더링함

# 가로 박스 플롯
plt.figure(figsize=(8, 2))
# 출력할 그래프의 가로 너비를 8인치, 세로 높이를 2인치로 설정함
# 수평 박스 플롯에 적절한 형태로 구성함

sns.boxplot(x='Profit', data=df)
# x축에 'Profit' 변수를 할당하여 수평 방향의 박스 플롯을 생성함
# 동일한 데이터를 수직이 아닌 가로 방향으로 표현하여 공간 활용과 해석 방향을 전환함
# 수평 플롯은 범위나 이상치를 시각적으로 빠르게 파악하는 데 유리함

plt.show()
# 수평 박스 플롯을 출력함
```

```python
# State 구분에 따른 Profit 박스 플롯 시각화

plt.figure(figsize=(8, 5))
# 전체 그래프의 크기를 가로 8인치, 세로 5인치로 설정함
# 카테고리(State) 축이 가로 방향이므로 충분한 가로 공간 확보를 위함

sns.boxplot(x="State", y="Profit", data=df)
# seaborn의 boxplot 함수를 사용하여 카테고리별 박스 플롯을 생성함
# x축에 'State'를 배치하여 각 주별 그룹을 형성함
# y축에 'Profit'을 할당하여 각 그룹의 이익 분포를 시각화함
# 각 박스는 해당 주에서의 이익 값의 중앙값, 사분위수, 이상치를 요약함
# 카테고리별 비교를 통해 지역 간 이익 편차 및 분포 특성을 파악함

plt.show()
# 구성된 박스 플롯을 출력하여 시각적으로 표현함
```

```python
# 평균, 데이터 포인트 포함한 박스 플롯 시각화

sns.boxplot(
    x="State", y="Profit", 
    showmeans=True, 
    boxprops={'facecolor':'None'}, 
    data=df
)
# x축에 'State'를, y축에 'Profit'을 할당하여 주(State)별 이익 분포를 시각화함
# showmeans=True: 각 박스 내부에 평균값을 점 형태로 표시함
# boxprops={'facecolor':'None'}: 박스 내부를 투명 처리함
# 박스를 투명하게 설정함으로써 이후 겹쳐 그릴 stripplot의 데이터 포인트가 가려지지 않도록 함
# 데이터프레임 df를 기반으로 시각화를 구성함


sns.stripplot(
    x='State', y='Profit', 
    data=df, 
    jitter=True, 
    marker='o', 
    alpha=0.5,
    color='black'
)
# 각 데이터 포인트를 점(marker) 형태로 흩뿌리는 방식으로 시각화함
# jitter=True: x축 기준으로 약간의 위치 흔들림을 추가하여 점들이 겹치지 않도록 분산시킴
# marker='o': 원형 마커로 지정하여 시각적으로 부드럽게 표현함
# alpha=0.5: 반투명(50%)으로 설정하여 데이터 밀집 구간도 확인 가능하게 처리함
# color='black': 모든 점을 검정색으로 지정하여 기본 박스와 명확히 구분되도록 함

plt.show()
```
