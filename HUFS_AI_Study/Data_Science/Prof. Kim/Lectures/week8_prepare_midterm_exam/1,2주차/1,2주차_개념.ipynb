# 1. 통계학의 이해

## 1.1 왜 통계학을 알아야 할까?
통계학은 데이터 사이언스에서 핵심적인 역할을 하며, 의사결정 과정에서 중요한 정보와 인사이트를 제공함.  
예를 들어, 마케팅 캠페인의 효과를 측정하거나, 제품 품질을 관리할 때 통계학적 방법을 통해 데이터를 분석하고 예측할 수 있음.  
통계학을 알면 데이터에서 패턴을 추출하고, 데이터의 정확성 및 신뢰도를 평가할 수 있음.

## 1.2 머신 러닝과 전통적 통계학의 차이

### 전통적 통계학
- 주로 데이터를 수집하고 이를 기반으로 가설을 설정하고 검증하는 방식임.  
- 예: 특정 치료법의 효과 검증, 모집단의 특성 추정  
- 주요 기법: 회귀분석, 분산 분석(ANOVA), 카이제곱 검정 등

### 머신 러닝
- 데이터에서 자동으로 패턴을 학습하여 예측 모델을 생성하는 방식임.  
- 통계학은 가설 설정과 검증에 중점을 두고, 머신 러닝은 예측 정확도 향상에 중점을 둠  
- 예: 스팸 이메일 분류 모델

## 1.3 통계학의 정의와 기원
- 통계학: 데이터를 수집, 정리, 분석하여 유용한 정보를 도출하는 학문  
- 기원: 17세기부터 발전, 고대 정부와 군대에서도 데이터 분석을 정책 결정에 사용  
- 현재: 경제, 의료, 사회과학, 공학 등 다양한 분야에 활용

## 1.4 기술 통계와 추론 통계

### 기술 통계
- 데이터를 요약하고 주요 특성을 추출하는 방법  
- 예: 시각화, 평균, 중앙값, 표준편차 등

### 추론 통계
- 표본을 통해 모집단의 특성을 추론하는 방법  
- 예: 설문조사 결과를 바탕으로 전체 집단 추정  
- 확률 이론 기반, 통계적 가설 검정 포함

# 2. 모집단과 표본 추출

## 2.1 모집단과 표본, 전수 조사와 표본 조사
- **모집단**: 연구 대상이 되는 전체 집합  
  예: "한국에 거주하는 모든 대학생"
- **표본**: 모집단에서 일부를 선택하여 분석하는 집합  
  - 표본은 모집단을 대표해야 하며, 이를 통해 모집단 특성을 추정

## 2.2 표본 조사의 이유와 데이터 과학적 용법
- 전체 모집단을 조사하는 것은 비용, 시간, 물리적 제약 등으로 어려움  
- **표본 조사**는 일부 데이터를 분석하여 전체를 추론하는 방식  
  예: 대규모 설문조사에서 몇 천 명만 조사하여 전체를 추정

## 2.3 표본 추출에서 나타나는 편향의 종류
- **편향(Bias)**: 표본이 모집단을 정확히 반영하지 못하는 현상
- 주요 편향 유형:
  - **확증 편향**: 원하는 결론을 뒷받침하는 데이터만 선택
  - **기준점 편향**: 처음 접한 정보를 기준으로 판단
  - **선택 지원 편향**: 특정 표본이 왜곡된 방식으로 선택됨

## 2.4 인지적 편향의 종류
- **확증 편향**: 기존 믿음을 강화할 데이터만 선택
- **기준점 편향**: 초기 정보에 지나치게 의존
- **선택 지원 편향**: 선택 후 그 선택이 옳았다고 믿게 됨

## 2.5 머신 러닝 모델 측면의 편향과 분산
- **편향(Bias)**: 일관되게 잘못된 예측을 하는 경향  
  - 과소적합(underfitting)의 원인이 됨
- **분산(Variance)**: 훈련 데이터에 민감하게 반응하여 결과가 불안정  
  - 과적합(overfitting)의 원인이 됨
- 머신 러닝에서는 이 둘 사이의 **균형(Bias-Variance Tradeoff)** 이 중요함

# 3. 변수와 척도

## 3.1 변수의 종류

### 연속형 변수 (Continuous Variable)
- 수치형 데이터로, 실수 값을 가질 수 있음
- 예: 나이, 온도, 가격

### 범주형 변수 (Categorical Variable)
- 숫자가 아닌 범주에 속하는 값을 가지는 변수
- 예: 성별, 국가, 교육 수준

## 3.2 변수 관계의 종류

### 양의 상관관계 (Positive Correlation)
- 한 변수가 증가하면 다른 변수도 함께 증가하는 경향
- 예: 키와 체중

### 음의 상관관계 (Negative Correlation)
- 한 변수가 증가하면 다른 변수는 감소하는 경향
- 예: 나이와 유연성

## 3.3 척도의 종류

### 명목척도 (Nominal Scale)
- 범주 구분은 있으나 순서 없음
- 예: 혈액형 (A, B, AB, O)

### 서열척도 (Ordinal Scale)
- 범주 간에 순서 있음, 간격은 일정하지 않음
- 예: 학위 (학사 < 석사 < 박사)

### 구간척도 (Interval Scale)
- 순서와 간격이 있음, 절대적인 0은 없음
- 예: 온도 (섭씨, 화씨)

### 비율척도 (Ratio Scale)
- 구간척도 특성 + 절대적인 0 존재
- 예: 몸무게, 나이

# 4. 데이터의 기술 통계적 측정

## 4.1 중심 경향의 측정

- **평균 (Mean)**: 데이터의 모든 값을 합산한 후 개수로 나눈 값  
  - 장점: 전체적인 중심 경향을 잘 나타냄  
  - 단점: 극단값(이상치)에 민감함

- **중앙값 (Median)**: 데이터를 크기순으로 정렬했을 때 중앙에 위치한 값  
  - 장점: 극단값의 영향을 받지 않음  
  - 평균이 왜곡될 때 대안으로 적합

- **최빈값 (Mode)**: 가장 많이 나타나는 값  
  - 주로 범주형 데이터에서 유용하게 사용

## 4.2 분산과 표준 편차

- **분산 (Variance)**: 데이터가 평균을 중심으로 얼마나 퍼져 있는지를 나타내는 지표  
\\[
\\text{분산} = \\frac{1}{N} \\sum_{i=1}^{N}(x_i - \\mu)^2
\\]
  - 값이 클수록 데이터가 넓게 퍼져 있음

- **표준 편차 (Standard Deviation)**: 분산의 제곱근으로, 실제 단위 기준에서 변동성 판단 가능  
  - 표준 편차가 작으면 평균 근처에 데이터가 밀집  
  - 표준 편차가 크면 퍼짐이 심함

## 4.3 산포도와 범위, 사분위수, 변동 계수

- **산포도 (Range)**: 최대값 - 최소값  
  - 데이터의 전체 범위를 간단히 파악 가능

- **사분위수 (Quartiles)**: 데이터를 네 개의 동일한 부분으로 나눈 경계값  
  - Q1: 하위 25%  
  - Q2: 중앙값  
  - Q3: 상위 25%

- **변동 계수 (Coefficient of Variation, CV)**:  
  \\[
  CV = \\frac{표준편차}{평균}
  \\]
  - 단위가 다른 데이터 간의 상대적 변동성 비교 시 사용

## 4.4 왜도와 첨도

- **왜도 (Skewness)**: 데이터 분포의 비대칭성
  - 왜도 = 0: 완전 대칭  
  - 음의 왜도: 왼쪽으로 치우침  
  - 양의 왜도: 오른쪽으로 치우침

- **첨도 (Kurtosis)**: 분포의 뾰족함 정도  
  - 첨도 ↑: 중심부에 값이 집중되고 꼬리가 두꺼움 (뾰족한 분포)  
  - 첨도 ↓: 평평한 분포

# 5. 확률과 확률 변수

## 5.1 확률의 기본 개념
- 확률은 **특정 사건이 발생할 가능성**을 수치로 표현한 것
- 항상 0과 1 사이의 값을 가짐
  - 0: 사건이 절대 발생하지 않음
  - 1: 사건이 반드시 발생함

## 5.2 확률의 종류

### 조건부 확률 (Conditional Probability)
- 특정 조건 하에서 사건이 일어날 확률
- 예: 어떤 날의 기상 조건이 주어졌을 때 비가 올 확률

### 결합 확률 (Joint Probability)
- 두 사건이 동시에 발생할 확률
- 예: 학생이 수학과 과학 시험 모두 통과할 확률

## 5.3 분할과 베이지안 이론

### 분할 (Partitioning)
- 모집단을 의미 있는 부분으로 나누어 분석
- 예: 연령대, 성별, 지역 등을 기준으로 분할 분석

### 베이지안 이론 (Bayesian Theory)
- **사전 확률(prior probability)**에 새로운 정보를 반영하여 **사후 확률(posterior probability)**을 갱신하는 방식
- \\[
  P(A|B) = \\frac{P(B|A) \\cdot P(A)}{P(B)}
  \\]
- 확률 추론을 반복적으로 갱신하는 데 사용

## 5.4 확률 분포

### 확률 분포 (Probability Distribution)
- 확률 변수의 가능한 값과 그 확률 간의 관계를 나타내는 함수

### 이산 확률 분포 (Discrete Probability Distribution)
- 확률 변수가 특정한 값을 가질 수 있음
- 예: 주사위 눈금 (1~6)

### 연속 확률 분포 (Continuous Probability Distribution)
- 확률 변수가 연속적인 값을 가질 수 있음
- 예: 키, 체중, 온도 등 실수값을 가지는 데이터

# 6. 데이터 분석 및 기법

## 6.1 데이터 분석의 전반적 과정

데이터 분석은 일반적으로 다음의 6단계로 이루어짐:

1. **문제 정의**
   - 분석의 목적과 해결해야 할 문제를 명확히 설정

2. **데이터 수집**
   - 문제 해결을 위해 필요한 데이터를 다양한 경로에서 수집

3. **데이터 정제**
   - 결측치, 이상치, 오류 데이터 등을 처리하여 분석 가능한 상태로 정리

4. **탐색적 데이터 분석 (EDA: Exploratory Data Analysis)**
   - 데이터의 분포, 관계 등을 시각화 및 통계적 방법으로 탐색
   - 변수 간 상관관계, 이상치 유무 확인

5. **모델링**
   - 예측 또는 분류를 위한 수학적 모델을 구축
   - 선형회귀, 결정트리, SVM, 딥러닝 등 다양한 기법 적용

6. **결과 해석 및 활용**
   - 모델 성능 평가 (정확도, 정밀도, 재현율 등)
   - 비즈니스 인사이트 도출 및 의사결정에 활용

## 6.2 데이터 분석 방법론

### CRISP-DM (Cross Industry Standard Process for Data Mining)
- 데이터 마이닝 표준 프로세스
- 단계:
  1. 비즈니스 이해
  2. 데이터 이해
  3. 데이터 준비
  4. 모델링
  5. 평가
  6. 배포

### SAS SEMMA
- SAS에서 제시한 데이터 분석 방법론
- 단계:
  1. **Sample**: 데이터 샘플링
  2. **Explore**: 데이터 탐색 및 시각화
  3. **Modify**: 변수 선택 및 변환
  4. **Model**: 모델링
  5. **Assess**: 모델 성능 평가

# 7. 가설 검정 (Hypothesis Testing)

## 7.1 귀무가설과 대립가설

- **귀무가설 (Null Hypothesis, H₀)**  
  - 기본적으로 받아들이는 가설  
  - 예: "두 그룹 간 평균에 차이가 없다", "효과가 없다"

- **대립가설 (Alternative Hypothesis, H₁)**  
  - 귀무가설에 반대되는 주장  
  - 예: "두 그룹 간 평균에 차이가 있다", "효과가 있다"

## 7.2 가설 검정의 절차

1. **귀무가설과 대립가설 설정**  
   - 분석 목적에 따라 두 가설을 수립

2. **유의수준 설정 (Significance Level)**  
   - 보통 0.05 또는 0.01 사용

3. **검정 통계량 계산**  
   - 표본 데이터를 바탕으로 검정에 사용할 통계량 계산

4. **p-value 계산**  
   - 귀무가설이 맞을 때 현재 데이터보다 극단적인 결과가 나올 확률

5. **결론 도출**  
   - p-value < 유의수준이면 귀무가설 기각, 대립가설 채택

## 7.3 가설 검정의 유의수준과 p-value

- **유의수준 (α)**  
  - 오류를 감수할 최대 허용 확률  
  - 예: α = 0.05 → 5% 확률로 잘못된 결론 허용

- **p-value**  
  - 관측된 결과가 귀무가설 하에서 얼마나 극단적인지를 나타냄  
  - p < α: 귀무가설 기각  
  - p ≥ α: 귀무가설 채택

## 7.4 1종 오류와 2종 오류

- **1종 오류 (Type I Error)**  
  - 귀무가설이 참인데도 기각하는 오류  
  - False Positive (실제로는 차이가 없는데 있다고 판단)

- **2종 오류 (Type II Error)**  
  - 대립가설이 참인데도 귀무가설을 기각하지 않는 오류  
  - False Negative (실제로는 차이가 있는데 없다고 판단)
