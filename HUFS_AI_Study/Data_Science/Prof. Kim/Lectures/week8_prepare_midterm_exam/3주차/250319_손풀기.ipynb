```python
# def matmult(a, b):
#     c = np.zeros((a.shape[0], b.shape[1]), dtype=int)
#     for i in range(a.shape[0]):
#         for j in range(b.shape[1]):
#             for k in range(a.shape[1]): # m by n, n by k: n = n
#                 c[i, j] += a[i,k] * b[k,j]
#     return c

def matmult(a, b):
    c = np.zeros((a.shape[0], b.shape[1]), dtype=int)
    for i in range(a.shape[0]):
        c[i,:] = (a[i,:]*b.T).sum(axis=1)
    return c
```

# ✅ `pd.concat()` vs `pd.merge()` 차이점 정리

## 📊 차이점 요약 표

| 함수         | 기본 방향                | 목적                                     | 키워드 기준 병합 여부     | 사용 예시                             |
|--------------|-------------------------|------------------------------------------|----------------------------|----------------------------------------|
| `pd.concat()` | 상하 (`axis=0`) 또는 좌우 (`axis=1`) | 단순하게 DataFrame을 이어붙이는 것       | ❌ 키 기반 병합 아님       | 여러 DataFrame을 순서대로 합칠 때     |
| `pd.merge()`  | 좌우 (`axis=1`)         | 공통된 key(열)를 기준으로 병합            | ✅ SQL처럼 key 기준 병합   | 공통 열(예: ID, 이름 등) 기준으로 병합할 때 |

---


```python
df_1 = pd.DataFrame( 
    {
        'A' : ['a10', 'a11', 'a12'],
        'B' : ['b10', 'b11', 'b12'],
        'C' : ['c10', 'c11', 'c12']
    },
        index = ['가', '나',  '다'] 
)

df_2 = pd.DataFrame( 
    {
        'B' : ['b23', 'b24', 'b25'],
        'C' : ['c23', 'c24', 'c25'],
        'D' : ['d23', 'd24', 'd25']
    },
    index = ['다', '라',  '마'] 
)
```

```python
df_3 = pd.concat([df_1, df_2])
df_3
```

```python
# on='B'로 지정한 열은 조인 키이므로 공통 열로 병합되며 하나로 유지됨.
# on으로 **지정하지 않은 공통 열(C 등)**은 양쪽에서 같은 이름이지만 다른 값일 수 있기 때문에 충돌을 방지하기 위해 C_x, C_y로 자동 분리.

# ✅ LEFT OUTER JOIN
# 기준: df_1
# df_1의 모든 행을 유지하고, df_2에서 'B' 값이 일치하는 경우에만 해당 데이터를 병합
# df_1에 존재하는 B 값들(b10, b11, b12)은 df_2에 존재하지 않으므로 병합할 데이터가 없어 NaN 채워짐
print('left outer \n', df_1.merge(df_2, how='left', on='B'))
print("--------------------------------------------------------")

# ✅ RIGHT OUTER JOIN
# 기준: df_2
# df_2의 모든 행을 유지하고, df_1에서 'B' 값이 일치하는 경우에만 해당 데이터를 병합
# df_2의 B 값들(b23, b24, b25)은 df_1에 존재하지 않으므로 병합할 데이터가 없어 NaN 채워짐
print('right outer \n', df_1.merge(df_2, how='right', on='B'))
print("--------------------------------------------------------")

# ✅ FULL OUTER JOIN
# 기준: 양쪽 모두
# df_1 또는 df_2 중 하나라도 'B' 값이 존재하면 모두 포함
# 어느 한쪽에만 존재하는 값은 상대쪽의 컬럼이 NaN으로 채워짐
# 결과적으로 b10 ~ b25까지 총 6개의 행이 포함됨
print('full outer \n', df_1.merge(df_2, how='outer', on='B'))
print("--------------------------------------------------------")

# ✅ INNER JOIN
# 기준: 양쪽 공통
# df_1과 df_2 양쪽에 모두 존재하는 'B' 값만 병합
# df_1의 B 값: b10, b11, b12 / df_2의 B 값: b23, b24, b25
# 공통되는 B 값이 없으므로 결과는 빈 DataFrame
print('inner \n', df_1.merge(df_2, how='inner', on='B'))
```

```python
df_3 = df_1.merge(df_2, how='outer', on='B')
print(df_3)
```

```python
df_3 = df_1.merge(df_2, how = 'outer',
                  left_index = True, right_index = True )
print(df_3)
```

| 구분            | `on='B'` 방식                            | `left_index=True, right_index=True` 방식          |
|-----------------|-------------------------------------------|--------------------------------------------------|
| 🔍 병합 기준     | 열(Column) `'B'`의 값                    | 행 인덱스(Index)                                 |
| 📌 전제 조건     | 양쪽에 `'B'`라는 이름의 열이 존재해야 함 | 양쪽에 일치 가능한 인덱스가 존재해야 함         |
| 💡 특징          | SQL의 `ON column` 기반 JOIN과 유사        | SQL의 `ON index` 기반 JOIN과 유사                |
| 🧪 결과 예측     | `'B'` 값이 같은 행끼리 병합               | 인덱스가 같은 행끼리 병합                        |
| 🚨 주의          | 중복된 `'B'` 값이 있으면 Cartesian Product 발생 | 중복된 인덱스 있으면 동일하게 곱집합 발생     |
