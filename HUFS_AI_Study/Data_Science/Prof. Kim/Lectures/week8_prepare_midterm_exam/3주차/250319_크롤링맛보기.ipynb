```python
import requests                         # HTTP 요청을 보내기 위한 requests 라이브러리 임포트
from bs4 import BeautifulSoup           # HTML 파싱을 위한 BeautifulSoup 클래스 임포트

url = 'https://n.news.naver.com/mnews/article/023/0003893596'  # 크롤링할 대상 뉴스 기사 URL 지정

response = requests.get(url)            # 지정한 URL에 대해 GET 요청을 전송하여 서버로부터 응답 객체 수신

if response.status_code == 200:        # 응답 코드가 200(정상 응답)인 경우에만 다음 절차 진행
    html = response.text               # 응답 객체에서 HTML 텍스트 추출
    soup = BeautifulSoup(html, 'html.parser')  # HTML 문서를 파싱하여 BeautifulSoup 객체로 변환

    # title = soup.select_one('h2.media_end_head_headline').text  
    # 과거 버전에서 사용하던 선택자 예시로, 현재 페이지 구조에서는 동작하지 않을 수 있음

    title = soup.select_one('h2#title_area').text  
    # id가 title_area인 h2 태그에서 텍스트만 추출하여 제목으로 사용
    print("뉴스 제목:", title)           # 추출한 뉴스 제목을 출력
else:                       # 응답 코드가 200이 아닌 경우 (예: 404, 500 등)
    pass                 # 에러 처리 로직이 없는 상태로 넘어감 (실전에서는 로그 기록 또는 예외 처리 권장)
```

```python
from selenium import webdriver                                 
# 웹 브라우저 자동화를 위한 selenium 모듈에서 webdriver 클래스 임포트
from selenium.webdriver.chrome.service import Service          
# ChromeDriver의 실행 서비스를 위한 Service 클래스 임포트
from webdriver_manager.chrome import ChromeDriverManager       
# Chrome 드라이버를 자동으로 설치/업데이트하는 매니저 임포트
import time                                                    
# 로딩 대기 등을 위한 시간 제어용 time 모듈 임포트
import os                                                      
# 경로 관련 작업을 위한 os 모듈 임포트

# 임시 사용자 데이터 디렉토리 설정
user_data_dir = os.path.join(os.getcwd(), "chrome_user_data")  
# 현재 작업 디렉토리에 "chrome_user_data" 폴더를 생성하여 사용자 데이터 저장 경로로 설정

# Selenium 옵션 설정
options = webdriver.ChromeOptions()                            
# 크롬 실행 옵션을 설정하기 위한 ChromeOptions 객체 생성
options.add_argument(f"--user-data-dir={user_data_dir}")       
# 사용자 데이터 디렉토리를 지정하여 기존 세션과 충돌 방지
options.add_argument("--headless")                             
# 브라우저 GUI 없이 백그라운드에서 실행되도록 설정
options.add_argument("--no-sandbox")                           
# 리눅스 환경에서 권한 관련 이슈 방지를 위한 설정
options.add_argument("--disable-dev-shm-usage")                
# 공유 메모리 제한 문제를 피하기 위한 설정

# Chrome WebDriver 실행
service = Service(ChromeDriverManager().install())             
# ChromeDriverManager를 통해 드라이버를 자동 설치하고 서비스 객체로 생성
driver = webdriver.Chrome(service=service, options=options)   
# 위에서 설정한 서비스와 옵션을 사용하여 Chrome 드라이버 객체 생성

try:
    # 웹사이트 이동
    driver.get("https://n.news.naver.com/mnews/article/023/0003893596")  # 지정된 뉴스 기사 페이지로 이동
    time.sleep(2)  # 자바스크립트 렌더링 대기용 지연 시간 설정

    # 뉴스 제목 가져오기
    news_title = driver.find_element("css selector", "h2.media_end_head_headline").text 
    # CSS 선택자를 사용하여 제목 요소를 찾고 텍스트를 추출
    print("뉴스 제목:", news_title)                                                       
    # 추출한 제목을 콘솔에 출력

except Exception as e:
    print("에러 발생:", e)   # 예외 발생 시 에러 메시지 출력

finally:
    driver.quit() # 브라우저와 WebDriver 프로세스를 종료하여 리소스 누수 방지
```
