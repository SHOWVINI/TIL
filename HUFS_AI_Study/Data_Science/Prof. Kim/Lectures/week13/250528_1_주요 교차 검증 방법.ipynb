```python
from sklearn.model_selection import train_test_split
# 전체 데이터를 학습 데이터와 테스트 데이터로 나누기 위한 함수 임포트 수행
# 모델의 일반화 성능 평가를 위한 데이터 분리 기능 수행
# 무작위 샘플링 방식으로 입력 데이터를 특정 비율로 분할하는 방식 사용

from sklearn.model_selection import KFold, LeaveOneOut, StratifiedKFold
# 다양한 형태의 교차 검증 방식(K-Fold, LOOCV, 계층적 K-Fold)을 지원하는 클래스 임포트 수행
# 모델의 과적합 방지 및 일반화 성능 향상을 위한 검증 전략 제공
# KFold는 데이터를 균등하게 나누는 방식, LeaveOneOut은 하나씩 제외하며 반복, StratifiedKFold는 클래스 비율을 유지하는 방식 사용

from sklearn.model_selection import cross_val_score, GridSearchCV
# 모델의 성능을 반복적으로 평가하는 함수와 하이퍼파라미터 최적화를 위한 탐색 클래스 임포트 수행
# cross_val_score는 교차 검증을 통한 모델 성능 평균 산출 기능 수행
# GridSearchCV는 모든 하이퍼파라미터 조합에 대해 교차 검증을 수행하여 최적 조합을 선택하는 방식 사용

from sklearn.metrics import accuracy_score
# 분류 모델의 정확도를 측정하는 함수 임포트 수행
# 예측값과 실제 정답 간 일치 비율을 계산하는 지표 기능 수행
# 전체 샘플 중 맞춘 샘플의 비율을 단일 수치로 출력하는 방식 사용

from sklearn.ensemble import RandomForestClassifier
# 앙상블 학습 기법 중 하나인 랜덤 포레스트 분류기 클래스 임포트 수행
# 여러 개의 결정 트리를 랜덤하게 생성하여 투표 기반으로 분류 결과를 출력하는 모델 구성 방식 제공
# 각 결정 트리의 다양성과 다수결 원리를 통해 높은 정확도와 과적합 방지 기능 제공

import pandas as pd
# 테이블 형태의 데이터 구조인 DataFrame을 생성하고 조작하기 위한 pandas 라이브러리 임포트 수행
# CSV, Excel 등 다양한 외부 데이터 포맷 불러오기 및 통계 분석 기능 지원
# 라벨 기반의 인덱싱, 정렬, 결측값 처리 등의 고수준 데이터 처리 기능 제공

import numpy as np
# 수치 해석 및 배열 계산 기능을 위한 numpy 라이브러리 임포트 수행
# 벡터, 행렬 등의 수학 연산과 선형대수, 확률 분포, 난수 생성 등 수학적 처리 기능 제공
# 다차원 배열(ndarray)을 중심으로 빠른 계산과 메모리 효율성 확보 기능 제공
```

```python
# 데이터 불러오기
# https://www.kaggle.com/datasets/uciml/glass
df = pd.read_csv("../../../main/datasets/glass.csv")

# 데이터 샘플 확인
df.head()
```

```python
# k-fold 방법 1

# 종속변수 문자형 변환
df["Type_str"] = df["Type"].apply(str)
# 숫자형 타입의 종속변수 'Type'을 문자열로 변환하여 새로운 열 'Type_str'로 추가하는 처리 수행
# scikit-learn의 분류기에서 레이블 인코딩 또는 문자열 분류 처리가 가능하도록 준비하는 전처리 절차 수행
# 범주형 데이터 처리나 시각화를 위한 사전 준비 작업으로 활용 가능

# 독립변수, 종속변수 분리
df_x1 = df[['RI','Na','Mg','Al','Si','K','Ca','Ba','Fe']]
df_y1 = df[['Type_str']]
# 입력 특성(feature)인 9개의 화학 성분(RI~Fe)만 선택하여 독립변수 df_x1로 분리하는 작업 수행
# 타겟 클래스(Type_str)만 선택하여 종속변수 df_y1로 지정하는 작업 수행
# 모델 학습 시 입력과 출력 역할을 명확히 구분하기 위한 전처리 단계 수행

# 기본 모델 설정
rnf_model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=0)
# 랜덤 포레스트 분류기를 기본 모델로 정의하고, 트리 개수는 100개, 최대 깊이는 5로 설정하는 구성 수행
# 과적합 방지를 위해 깊이를 제한하며, 예측 결과의 일관성을 위해 난수 시드를 고정하는 방식 사용
# 트리 기반 앙상블 모델을 통해 높은 분류 성능과 견고한 일반화 능력 확보 기능 수행

# 학습셋과 테스트셋 분리하여 생성(6:4)
x_train, x_test, y_train, y_test = train_test_split(
    df_x1, df_y1, test_size=0.4, random_state=10)
# 전체 데이터의 60%는 학습용, 40%는 테스트용으로 무작위 분할하여 훈련/평가용 데이터셋 생성 작업 수행
# random_state 설정을 통해 분할의 재현성을 확보하고, 교차 검증 이전의 기준 모델 훈련을 위한 준비 작업 수행
# 모델 평가 시 훈련 데이터와 무관한 테스트 데이터를 확보함으로써 과적합 여부를 판별할 수 있는 기반 마련

# cross_val_score 함수로 k-fold 성능 측정
kfold_scores_1 = cross_val_score(rnf_model, x_train, y_train, cv = 7)
# 학습용 데이터에 대해 7-fold 교차 검증을 수행하여 각 분할별 모델 성능을 측정하고 결과를 배열로 반환하는 처리 수행
# 전체 학습 데이터를 7개 조각으로 나누고, 각 조각을 한 번씩 검증용으로 사용하며 7회 반복 학습 및 평가 수행
# 모델의 평균적 성능을 보다 신뢰성 있게 확인하고 데이터셋 의존성을 줄이는 평가 방식 적용

print('k-fold 교차 검증 SCORE : ', kfold_scores_1)
# 각 fold 별 검증 정확도를 배열 형태로 출력하여 fold 간 편차 확인 기능 수행

print("k-fold 교차 검증 SCORE 평균 : {:.2f}".format(kfold_scores_1.mean()))
# 전체 fold 결과의 평균값을 소수점 둘째 자리까지 포맷팅하여 출력하는 처리 수행
# 모델 전반적인 성능에 대한 요약 지표 제공 및 향후 모델 튜닝 방향 설정에 활용 가능
```

```python
# k-fold 방법 2

# 종속변수 문자형 변환 및 분리
df["Type_str"] = df["Type"].apply(str)
# 정수형으로 되어 있는 'Type' 값을 문자열로 변환하여 새로운 열 'Type_str'로 생성하는 처리 수행
# 문자열 레이블을 사용하는 분류 모델에 적합한 형태로 종속변수를 변환하는 전처리 작업 수행

df_x2 = np.array(df[['RI','Na','Mg','Al','Si','K','Ca','Ba','Fe']])
# 독립변수로 사용할 9개의 화학 성분 특성을 NumPy 배열로 변환하여 df_x2에 저장하는 처리 수행
# scikit-learn에서 지원하는 배열 기반 모델 학습 방식에 적합하도록 데이터 포맷을 변환하는 작업 수행

df_y2 = df['Type_str']
# 문자열로 변환된 종속변수 'Type_str'을 레이블 배열로 분리하여 df_y2에 저장하는 처리 수행
# 모델 학습 시 입력 특성과 정답 레이블을 구분하여 준비하는 작업 수행

# k-fold 설정
kfold = KFold(n_splits=7, shuffle=True, random_state=37)
# 7개의 fold로 데이터를 분할하고, 매 fold마다 무작위로 섞은 뒤 고정된 순서를 유지하기 위해 시드값 설정 수행
# 반복적인 검증에서 데이터 분포에 의한 편향을 줄이기 위한 무작위 분할 방식 적용
# 모델 성능의 신뢰도 확보와 일반화 검증 목적의 교차 검증 구조 정의

kfold_scores_2_mean = []
# fold 별 정확도 값을 저장할 빈 리스트 생성

fold_idx = 0
# fold 인덱스를 추적하기 위한 초기값 설정

# k-fold 수행
for train_idx, test_idx in kfold.split(df_x2):
    # 현재 fold에서 사용할 훈련 인덱스와 테스트 인덱스를 kfold 객체로부터 생성하는 처리 수행
    # 전체 데이터를 훈련용과 검증용으로 나누는 과정 반복 수행

    train_x, train_y = df_x2[train_idx], df_y2[train_idx]
    # 현재 fold의 훈련용 독립변수 및 종속변수 데이터를 추출하는 작업 수행
    # 배열 인덱싱을 활용하여 학습 입력과 정답 레이블을 분리하는 처리 수행

    test_x, test_y = df_x2[test_idx], df_y2[test_idx]
    # 현재 fold의 테스트용 독립변수 및 종속변수 데이터를 추출하는 작업 수행
    # 모델 검증을 위한 평가용 데이터셋 구성 작업 수행

    # 기본 모델 설정
    rnf_model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=0)
    # 최대 깊이 5, 트리 개수 100개인 랜덤 포레스트 분류기 객체를 생성하여 모델 정의 수행
    # 반복 fold 사이에서 동일한 조건으로 훈련할 수 있도록 시드 고정을 통해 일관성 확보

    rnf_model.fit(train_x, train_y)
    # 현재 fold의 훈련 데이터를 기반으로 랜덤 포레스트 모델을 학습하는 과정 수행
    # 다수의 결정 트리를 구성하고 각 트리의 학습 결과를 결합하여 최종 모델 생성

    pred_y = rnf_model.predict(test_x)
    # 테스트 입력 데이터를 기반으로 학습된 모델이 예측한 분류 결과 생성 작업 수행
    # 앙상블 모델의 투표 결과를 기반으로 각 샘플에 대한 클래스 레이블 산출

    kfold_scores_2 = accuracy_score(test_y, pred_y)
    # 실제 정답(test_y)과 예측값(pred_y) 사이의 정확도를 계산하여 현재 fold의 성능을 평가하는 작업 수행
    # 분류 모델의 성능을 단일 수치로 정량화하는 평가 지표로 활용

    fold_idx += 1
    # fold 횟수를 1 증가시켜 다음 루프를 위한 인덱스 갱신 수행

    kfold_scores_2_mean.append(kfold_scores_2)
    # 현재 fold의 정확도 값을 리스트에 저장하여 평균 계산을 위한 누적 데이터 구성

print(f"k-fold 교차 검증 SCORE 평균 : {np.mean(kfold_scores_2_mean)}")
# 7개 fold에 대해 계산된 정확도 값들의 평균을 산출하고 출력하는 작업 수행
# 모델 성능의 전반적인 수준을 요약하여 평가하고 튜닝 결과와 비교 가능한 기준 제공
```

```python
# LOOCV (Leave-One-Out Cross-Validation)
# 전체 샘플 수만큼 교차 검증을 수행하여 각 샘플을 한 번씩 테스트셋으로 사용하고 나머지는 학습셋으로 사용하는 검증 방식임
# 매우 정밀한 성능 평가가 가능하나 샘플 수가 많을 경우 연산 비용이 높아지는 단점이 존재함
# 소규모 데이터셋에 적합하며 모델 일반화 성능을 미세하게 측정하는 데 활용 가능함

# 기본 모델 설정
rnf_model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=0)
# 랜덤 포레스트 분류기 객체를 생성하고, 트리 개수는 100개, 최대 깊이는 5로 제한하는 설정 수행
# 일관된 결과 재현을 위한 난수 시드 고정 처리 수행
# 이후 교차 검증 시 반복적으로 사용될 모델 구조 정의

rnf_model.fit(train_x, train_y)
# 주어진 훈련 데이터를 기반으로 랜덤 포레스트 모델을 학습시키는 작업 수행
# 앙상블 방식으로 다수의 결정 트리를 구성하여 복합적인 분류 기준 학습

pred_y = rnf_model.predict(test_x)
# 학습된 모델을 이용하여 테스트 데이터에 대한 예측값을 생성하는 처리 수행
# 각 샘플에 대해 투표 기반으로 최종 클래스 라벨 결정 수행

loocv = LeaveOneOut()
# LOOCV(Leave-One-Out Cross-Validation) 객체 생성 작업 수행
# 전체 데이터 중 단 한 개의 샘플을 검증용으로 사용하고 나머지를 훈련에 사용하는 방식으로 분할 설정
# 관측치 수만큼 분할이 일어나므로 매우 세밀한 평가가 가능한 검증 전략 구성

loocv_scores = cross_val_score(rnf_model, test_x, test_y, cv=loocv)
# 테스트 데이터를 대상으로 Leave-One-Out 교차 검증을 수행하고, 각 분할에서의 정확도를 측정하여 배열로 저장하는 처리 수행
# 각 관측치가 테스트셋으로 1회씩 사용되며, 나머지 샘플로 모델을 학습하는 작업 반복 수행
# 매우 세밀한 평가가 가능하나 계산 비용이 큰 방식 적용

print("테스트 셋 전체 관측치 수 : ", len(test_y))
# 테스트셋에 포함된 총 관측치 수를 출력하여 전체 분할 횟수와 비교하기 위한 기준 제공

print("LOOCV 검증 분할 횟수 : ", len(loocv_scores))
# LOOCV 결과로 생성된 교차 검증 횟수를 출력하여 검증 반복 횟수 확인 기능 수행
# 테스트셋 관측치 수와 일치하는지 확인하여 LOOCV 적용 여부 검증 가능

print("LOOCV 교차 검증 SCORE 평균 : {:.2f}".format(loocv_scores.mean()))
# 모든 LOOCV 분할 결과의 정확도 평균을 산출하고 소수점 둘째 자리까지 포맷팅하여 출력하는 처리 수행
# 모델의 미세한 일반화 성능을 평가할 수 있는 정밀 지표 제공
```

```python
# Stratified K-fold Cross Validation

# Stratified K-fold 교차검증 설정
strati = StratifiedKFold(n_splits=3)
# 전체 데이터셋을 계층적(stratified) 방식으로 3개의 fold로 분할하는 StratifiedKFold 객체 생성 수행
# 각 fold마다 클래스 비율이 전체 데이터와 유사하도록 유지하는 분할 방식 설정
# 클래스 불균형 데이터에서 평가 편향을 줄이기 위한 교차 검증 전략 구성

n_iter = 0
# 반복 횟수를 저장할 카운터 변수 초기화 작업 수행

print("전체 데이터셋 범주 별 관측치 수 :\n", df_y2.value_counts())
# 전체 레이블 데이터(df_y2)에 포함된 각 클래스별 관측치 개수를 출력하여 원래 클래스 분포 확인 작업 수행

for train_index, test_index in strati.split(df_x2, df_y2):
    # StratifiedKFold 객체의 split 메서드를 통해 학습용 인덱스와 검증용 인덱스를 생성하는 반복문 수행
    # 각 반복마다 클래스 비율을 유지한 상태로 데이터를 학습셋과 검증셋으로 분할하는 처리 수행

    n_iter += 1
    # 반복 카운터를 1 증가시켜 현재 분할 횟수를 추적하는 처리 수행

    strati_train_y = df_y2.iloc[train_index]
    # 현재 fold에서 선택된 학습용 인덱스를 기반으로 종속변수 레이블을 추출하여 strati_train_y에 저장하는 작업 수행

    strati_test_y = df_y2.iloc[test_index]
    # 현재 fold에서 선택된 검증용 인덱스를 기반으로 종속변수 레이블을 추출하여 strati_test_y에 저장하는 작업 수행

    print('분할 {0}'.format(n_iter))
    # 현재 fold의 번호를 출력하여 반복 상황을 명시하는 정보 제공

    print('학습 셋 범주 별 관측치 수:\n', strati_train_y.value_counts())
    # 학습셋 내 클래스별 관측치 수를 출력하여 클래스 비율 유지 여부 확인 작업 수행

    print('검증 셋 범주 별 관측치 수:\n', strati_test_y.value_counts())
    # 검증셋 내 클래스별 관측치 수를 출력하여 클래스 비율 유지 여부 확인 작업 수행

strati_scores = cross_val_score(rnf_model, test_x, test_y, cv=strati)
# 학습된 랜덤 포레스트 모델을 사용하여 테스트 데이터에 대해 Stratified K-Fold 교차 검증을 수행하는 처리 수행
# 각 fold에서 모델 훈련 및 평가를 반복하여 정확도 점수를 배열 형태로 반환하는 방식 사용

print("Stratified K-fold 교차 검증 SCORE 평균 : {:.2f}".format(strati_scores.mean()))
# 3개 fold의 정확도 점수 평균을 계산하여 소수점 둘째 자리까지 포맷팅 후 출력하는 작업 수행
# 클래스 불균형에 대응한 교차 검증 결과를 정량적으로 요약하여 모델 성능 평가를 보다 신뢰성 있게 수행
```

```python
# Nested Cross Validation 및 Grid Search Cross Validation

# ① Grid Search Cross Validation:
#    - 사전에 정의된 하이퍼파라미터 후보 집합을 기반으로
#      각 조합에 대해 교차 검증을 수행하고 최적의 조합을 선택하는 방식임
#    - 내부적으로 지정된 cv 파라미터에 따라 교차 검증이 반복되며,
#      각 모델의 평균 성능을 기반으로 최고의 조합을 탐색함
#    - 모델 선택 자동화 및 하이퍼파라미터 튜닝을 위한 핵심 기법으로 활용됨
#
# ② Nested Cross Validation:
#    - 하이퍼파라미터 탐색 과정에서 발생할 수 있는 데이터 누수(data leakage)를 방지하기 위한 구조화된 검증 절차임
#    - 내부 루프(inner CV)에서는 하이퍼파라미터를 최적화하고,
#      외부 루프(outer CV)에서는 내부에서 선택된 모델의 일반화 성능을 평가함
#    - 모델 평가와 튜닝 과정의 분리를 통해 과적합 방지 및 신뢰성 높은 성능 추정치를 제공함
#    - 특히 소규모 데이터셋이나 복잡한 하이퍼파라미터 공간을 다룰 때 효과적인 평가 방식임

# 학습셋과 테스트셋 분리하여 생성(6:4)
x_train, x_test, y_train, y_test = train_test_split(
    df_x2, df_y2, test_size=0.4, random_state=10)
# 전체 데이터의 60%를 학습용, 40%를 테스트용으로 무작위 분할하여 x/y 학습셋과 테스트셋 구성
# 모델 성능 검증과 최종 테스트를 명확히 분리하기 위한 데이터 분할 작업 수행
# 재현 가능한 분할을 위해 난수 시드를 고정하는 설정 적용

# 모델 시도 횟수 설정
NUM_TRIALS = 20
# Nested Cross Validation을 총 20번 반복 수행하기 위한 반복 횟수 설정
# 반복을 통해 평균적이고 안정적인 모델 성능 추정값 확보 목적 수행

# 그리드 하이퍼 파라미터 설정
hp_para = {'max_depth': [2, 3, 4], 'min_samples_split': [2, 3]}
# 하이퍼파라미터 후보군으로 결정 트리 최대 깊이와 최소 분할 샘플 수를 사전 정의하는 작업 수행
# GridSearchCV가 각 조합을 실험적으로 탐색할 수 있도록 파라미터 공간 제공

# 스코어를 array로 저장
nested_scores = np.zeros(NUM_TRIALS)
# 각 trial에서 얻은 교차 검증 평균 점수를 저장할 배열 초기화 작업 수행

# 교차 검증 반복 수행
for i in range(NUM_TRIALS):
    # trial 인덱스를 기반으로 매 반복마다 다른 난수 시드로 교차 검증 실행

    inner_loop = KFold(n_splits=3, shuffle=True, random_state=i)
    # 내부 교차 검증 루프 설정: 하이퍼파라미터 튜닝용 3-fold CV 구성
    # 데이터를 무작위로 섞어 각 fold 간 편향 최소화 처리 수행

    outer_loop = KFold(n_splits=3, shuffle=True, random_state=i)
    # 외부 교차 검증 루프 설정: 내부 루프에서 선택된 최적 모델을 검증할 3-fold CV 구성
    # 모델 선택과 성능 검증을 완전히 분리하여 과적합 방지 및 일반화 성능 추정 기능 수행

    # Nested Cross Validation 하이퍼 파라미터 최적화
    gs_cv = GridSearchCV(rnf_model, param_grid=hp_para, cv=inner_loop)
    # 내부 루프를 이용하여 각 하이퍼파라미터 조합에 대해 최적 모델을 탐색하는 GridSearchCV 객체 정의
    # 설정된 파라미터 그리드를 기준으로 교차 검증을 반복하며 최고 성능 조합을 선택하는 기능 수행

    nested_score = cross_val_score(gs_cv, X=x_train, y=y_train, cv=outer_loop)
    # 외부 루프를 통해 내부에서 선택된 최적 모델을 검증하고 각 fold의 성능 점수를 산출하는 작업 수행
    # 내부 루프와 외부 루프를 분리하여 하이퍼파라미터 탐색으로 인한 데이터 누수 방지 기능 수행

    nested_scores[i] = nested_score.mean()
    # 현재 trial에서 얻은 평균 성능 점수를 배열에 저장하는 작업 수행

# 테스트셋에 모델 적용
gs_cv.fit(x_test, y_test)
# 테스트 데이터를 기반으로 GridSearchCV 객체를 훈련시켜 최종적으로 최적 하이퍼파라미터와 모델 적합 수행
# 테스트셋을 이용하여 선택된 모델의 성능 확인 및 하이퍼파라미터 적용 효과 확인 가능

print("각 TRIAL 별 SCORE 평균 : \n", nested_scores)
# 각 반복 trial마다 산출된 교차 검증 평균 정확도를 배열 형태로 출력하여 반복 성능 확인 기능 제공

print("전체 TRIAL SCORE 평균 : {:.2f}".format(nested_scores.mean()))
# 20개 trial에서 얻은 성능의 평균을 출력하여 모델의 전반적인 일반화 성능 요약 정보 제공

print("최대 TRIAL SCORE : {:.2f}".format(nested_scores.max()))
# 20개 trial 중 가장 높은 정확도를 출력하여 최고 성능 모델에 대한 기준 제공

print("최적 하이퍼 파라미터 : \n", gs_cv.best_params_)
# GridSearchCV 객체에 저장된 최적 하이퍼파라미터 조합을 출력하여 성능 향상에 기여한 설정값 제시
```
