```python
# 필요한 패키지 설치

import pandas as pd
# 표 형태의 데이터 처리와 분석을 위한 pandas 라이브러리 임포트 수행
# 데이터프레임을 기반으로 정렬, 그룹화, 결측값 처리 등의 고수준 데이터 조작 기능 제공

import numpy as np
# 수치 계산 및 배열 처리를 위한 numpy 라이브러리 임포트 수행
# 벡터 및 행렬 연산, 통계적 계산, 선형대수 연산 등의 기반 기능 제공

from sklearn.linear_model import LinearRegression
# 선형 회귀 모델 클래스를 sklearn에서 임포트 수행
# 연속형 종속변수를 예측하는 선형 모델 학습 및 추론 기능 제공

from sklearn.model_selection import train_test_split
# 전체 데이터를 학습용과 테스트용으로 분할하는 함수 임포트 수행
# 모델 학습 후 일반화 성능을 평가하기 위한 데이터 분할 기능 제공

from sklearn.metrics import mean_squared_error # MSE
# 예측값과 실제값 간의 오차 제곱의 평균을 계산하는 함수 임포트 수행
# 회귀 모델의 평균적인 오차 크기를 제곱 기준으로 측정하는 지표 제공

from sklearn.metrics import mean_absolute_error # MAE
# 예측값과 실제값 간의 절대 오차의 평균을 계산하는 함수 임포트 수행
# 회귀 모델의 예측 오차 크기를 직관적으로 평가할 수 있는 지표 제공

from sklearn.metrics import mean_absolute_percentage_error # MAPE
# 실제값 대비 예측 오차 비율의 평균을 계산하는 함수 임포트 수행
# 단위에 무관하게 상대적인 예측 정확도를 평가할 수 있는 지표 제공

from sklearn.metrics import mean_squared_log_error # MSLE
# 로그 변환된 예측값과 실제값 간의 오차 제곱 평균을 계산하는 함수 임포트 수행
# 큰 오차보다 작은 오차에 더 민감하게 반응하며, 과소 예측에 중점을 두는 지표 제공
```

```python
# 데이터 불러오기
# https://www.kaggle.com/datasets/harlfoxem/housesalesprediction
df = pd.read_csv("../../../main/datasets/kc_house_data.csv")

# 데이터 샘플 확인
df.head()
```

```python
# 독립변수와 종속변수 분리하여 생성
x = df[[ 'bedrooms', 'bathrooms', 'sqft_living',
       'sqft_lot', 'floors', 'waterfront', 'view', 'condition', 'grade',
       'sqft_above', 'sqft_basement', 'yr_built', 'yr_renovated', 
       'sqft_living15', 'sqft_lot15']]
# 회귀 모델의 입력값으로 사용할 독립변수(feature)들을 선택하여 변수 x에 저장하는 처리 수행
# 주택 가격에 영향을 미칠 수 있는 주요 구조적/환경적 특성들로 구성된 다차원 특성 벡터 정의
# 'id'와 'date'는 고유 식별자 및 시계열적 정보로 예측에 직접 활용하지 않으므로 제외함

# 'id', 'date'는 키값에 해당하므로 변수에서 제외 해준다.
# 'id'는 각 행의 고유 식별자 역할이며, 'date'는 거래일 정보이므로 모델 입력 변수에서 제외함이 타당함
# 예측 성능에 기여하지 않거나 오히려 모델 왜곡을 초래할 수 있는 열은 분석에서 배제하는 전처리 원칙 적용

y = df[['price']]
# 회귀 모델의 예측 대상인 종속변수 'price' 열을 선택하여 변수 y에 저장하는 처리 수행
# 모델이 학습을 통해 예측하려는 목표 변수로, 주택의 최종 거래 가격 정보 포함

# 학습셋과 테스트셋 분리하여 생성(7:3)
# df_train, df_test = train_test_split(df, test_size = 0.4) 
# 원래 DataFrame 자체를 나누는 방식은 주석 처리함

x_train, x_test, y_train, y_test = train_test_split(x, y, train_size=0.6, test_size=0.4)
# 전체 데이터 중 60%를 학습용, 40%를 테스트용으로 분리하여 각 입력(x)과 출력(y)에 대해 훈련/테스트 셋을 구성하는 작업 수행
# 무작위 분할을 통해 학습에 사용되지 않은 데이터로 모델의 일반화 성능을 측정할 수 있는 기반 마련
# 명시적으로 train_size와 test_size를 모두 지정하여 분할 비율을 제어하는 설정 적용
```

```python
# 다중회귀모델 생성
mreg = LinearRegression(fit_intercept=True)
# sklearn의 LinearRegression 클래스를 이용하여 다중 선형 회귀 모델 객체를 생성하는 작업 수행
# fit_intercept=True 설정을 통해 절편(편향항, bias term)을 포함한 회귀 모델을 구성함
# 입력 특성들과 목표 변수 간의 선형 관계를 학습하여 가중치(회귀계수)를 추정하는 목적 수행

mreg.fit(x_train, y_train)
# 학습용 입력 데이터(x_train)와 출력 데이터(y_train)를 이용하여 회귀 모델을 학습하는 작업 수행
# 최소제곱법(Ordinary Least Squares)에 기반하여 각 특성에 대한 회귀계수를 계산하고 모델을 적합함
# 주어진 입력 변수들이 주택 가격에 미치는 영향을 수치적으로 추정하는 과정 수행

# 테스트셋에 모델 적용
y_predict = mreg.predict(x_test)
# 학습된 회귀 모델을 이용하여 테스트셋 입력 데이터(x_test)에 대한 예측값을 계산하는 작업 수행
# 각 테스트 샘플의 특성을 이용하여 선형 회귀식을 기반으로 주택 가격을 추정하는 처리 수행
# 예측 결과는 y_predict 배열에 저장되며, 이후 평가 지표 계산에 활용 가능
```

```python
    # RMSE 산출 (MSE에 루트 적용)

MSE = mean_squared_error(y_test, y_predict)
# 실제값(y_test)과 예측값(y_predict) 사이의 오차를 제곱한 후 평균을 구하여 평균제곱오차(MSE)를 계산하는 처리 수행
# 예측값이 실제값과 얼마나 떨어져 있는지를 제곱 기반으로 평가하는 지표 산출

RMSE = np.sqrt(MSE)
# 평균제곱오차(MSE)에 제곱근을 적용하여 루트 평균 제곱 오차(RMSE)를 계산하는 처리 수행
# MSE보다 해석이 직관적이며 단위가 실제 예측 단위와 같기 때문에 비교 및 설명에 유리한 지표 제공

print(("RMSE : {:.2f}".format(RMSE)))
# RMSE 값을 소수점 둘째 자리까지 포맷팅하여 출력하는 작업 수행

    # MAE 산출
    
MAE = mean_absolute_error(y_test, y_predict)
# 실제값과 예측값 사이의 절대값 차이를 평균하여 평균절대오차(MAE)를 계산하는 처리 수행
# 오차를 그대로 측정하므로 이상치(outlier)에 덜 민감한 지표 제공

print(("MAE : {:.2f}".format(MAE)))
# MAE 값을 소수점 둘째 자리까지 포맷팅하여 출력하는 작업 수행

    # MAPE 산출

MAPE = mean_absolute_percentage_error(y_test, y_predict)
# 실제값 대비 예측 오차의 비율을 계산하고 그 평균을 구하여 평균 절대 백분율 오차(MAPE)를 산출하는 처리 수행
# 단위와 무관하게 예측 성능을 백분율로 해석할 수 있는 지표 제공

print(("MAPE : {:.2f}".format(MAPE)))
# MAPE 값을 소수점 둘째 자리까지 포맷팅하여 출력하는 작업 수행

    # RMSLE 산출 (MSLE에 루트 적용)

# 음수값 전처리
y_predict_df = pd.DataFrame(y_predict, columns=['price2'])
# 예측값 배열을 DataFrame으로 변환하고 'price2'라는 열 이름을 부여하여 저장하는 처리 수행

y_predict_df2 = y_predict_df.copy()
# RMSLE 계산 전 처리를 위해 원본 예측값 DataFrame을 복사하여 y_predict_df2에 저장하는 작업 수행

y_predict_df2.loc[y_predict_df2['price2'] < 0, 'price2'] = 0
# 로그 계산에서 음수가 입력되면 오류가 발생하므로, 예측값이 0보다 작은 경우 0으로 수정하는 전처리 수행
# RMSLE는 로그 기반 지표이므로 음수 예측값은 허용되지 않음

y_predict_rmsle = y_predict_df2.to_numpy()
# 수정된 예측값 DataFrame을 넘파이 배열로 변환하여 RMSLE 계산에 사용할 수 있도록 형식을 맞추는 처리 수행

MSLE = mean_squared_log_error(y_test, y_predict_rmsle)
# 실제값과 로그 예측값 사이의 제곱 평균을 구하여 평균제곱로그오차(MSLE)를 계산하는 처리 수행
# 과소 예측에 더 민감하고 이상치의 영향은 줄이는 특성을 지닌 지표 산출

RMSLE = np.sqrt(MSLE)
# MSLE에 제곱근을 적용하여 루트 평균제곱로그오차(RMSLE)를 계산하는 작업 수행
# 예측값이 실제값보다 작을 때 손실이 크고, 과대 예측에 관대함

print(("RMSLE : {:.2f}".format(RMSLE)))
# RMSLE 값을 소수점 둘째 자리까지 포맷팅하여 출력하는 작업 수행
```

```python
# RMSLE ver. 2 (rmsle 결과값이 다름)

def rmsle(predicted_values, actual_values):
    # RMSLE(Root Mean Squared Logarithmic Error)를 수식 기반으로 직접 구현한 함수 정의
    # 로그 변환을 통해 과소 예측에 민감하게 작동하고, 큰 값의 오차 영향을 완화하는 회귀 성능 평가 지표 계산 목적 수행

    # 테스트셋 y 값과 예측값에 +1 및 로그
    log_y_test = np.log(y_test + 1)
    log_y_predict = np.log(y_predict + 1)
    # 실제값과 예측값 모두에 log(1 + x) 변환을 적용하여 로그 기반 RMSLE 계산 준비 수행
    # 0 이하 값에서 로그 계산이 불가능하므로 +1 처리를 통해 안정성을 확보하는 전처리 절차 적용

    # 테스트셋 y 값 - 예측값 및 제곱
    diff = log_y_predict - log_y_test
    # 로그 변환된 예측값과 실제값 간의 차이를 계산하여 차이 벡터 생성

    diff_square = np.square(diff)
    # 로그 차이를 제곱하여 오차의 방향성을 제거하고 제곱 오차로 변환하는 처리 수행

    # 차이값 평균 및 루트
    mean_diff = diff_square.mean()
    # 제곱 오차들의 평균을 계산하여 평균 로그 제곱 오차(MSLE)를 산출

    final_rmsle = np.sqrt(mean_diff)
    # 평균 로그 제곱 오차(MSLE)에 제곱근을 적용하여 RMSLE를 최종 계산

    return final_rmsle
    # 계산된 RMSLE 값을 반환하여 외부에서 호출 시 사용 가능하도록 설정

rmsle(y_test, y_predict)
# 테스트셋의 실제값(y_test)과 예측값(y_predict)을 전달하여 RMSLE 계산 함수 호출
```

```python
# RMSLE ver. 2 수정 본
# predicted는 y_predict이고, 이 값은 일부 음수 값을 포함
# 로그 함수는 음수를 받을 수 없으므로 NaN이 발생하고, 결과가 왜곡

def rmsle_custom(actual, predicted):
    predicted = np.maximum(predicted, 0)  # 음수 제거 추가
    # 예측값 배열 중 0보다 작은 값을 모두 0으로 바꾸는 전처리 수행
    # 로그 함수는 입력값이 음수일 경우 NaN을 반환하므로, RMSLE 계산 안정성을 확보하기 위한 필수 처리

    # 테스트셋 y 값과 예측값에 +1 및 로그
    log_actual = np.log(actual + 1)
    log_predicted = np.log(predicted + 1)
    # 실제값과 예측값 각각에 대해 log(1 + x)를 적용하여 로그 기반 RMSLE 계산 준비 수행
    # 값이 0일 경우도 고려하여 +1 처리를 통해 로그 정의역 조건을 만족시키는 방식 적용

    # 테스트셋 y 값 - 예측값 및 제곱
    diff = log_predicted - log_actual
    # 로그 변환된 실제값과 예측값 간의 차이를 계산하여 오차 벡터 생성

    # 차이값 평균 및 루트
    mean_diff = np.mean(np.square(diff))
    # 차이를 제곱하여 방향성을 제거하고 평균을 구함으로써 평균 로그 제곱 오차(MSLE)를 계산

    return np.sqrt(mean_diff)
    # MSLE에 제곱근을 적용하여 최종 RMSLE(Root Mean Squared Logarithmic Error) 값을 반환하는 처리 수행

rmsle_custom(y_test, y_predict)
# 테스트셋 실제값(y_test)과 예측값(y_predict)을 입력으로 하여 RMSLE 값을 계산하고 반환하는 함수 호출 수행
```
