### CV 분야에서는 왜 정적인 이미지를 영상이라고 칭하는걸까?


#### 1. 영상(Image) vs. 비디오(Video)

일반적으로 우리는 "이미지(Image)"와 "비디오(Video)"를 구분해서 사용

- 이미지(Image): 단일 프레임으로 존재하는 정적인 그림 (예: JPG, PNG 파일)
- 비디오(Video): 여러 개의 연속된 이미지를 시간 순서대로 배열하여 움직임을 표현하는 것 (예: MP4, AVI 파일)

그러나 **영상(Image)**이라는 용어는 컴퓨터 비전 및 영상처리 분야에서 단순한 이미지뿐만 아니라 다양한 방식으로 표현된 시각적 데이터를 포함하는 개념으로 사용

#### 2. 영상(Image)이라는 용어의 의미

컴퓨터 비전에서는 디지털 이미지가 2D 배열(행렬)로 저장되고 처리되며, 이때 "영상"이라는 용어를 사용

이는 다음과 같은 이유 때문

    (1) 신호 처리 개념에서 유래

컴퓨터 비전 및 신호 처리에서는 **영상(Signal Processing Image)**이라는 용어가 널리 사용

영상은 픽셀(pixel) 단위로 구성되며, 이는 특정 신호 데이터를 의미

우리가 보는 이미지는 디지털 신호로 표현된 데이터이며, 이 데이터의 집합을 "영상"이라고 함

    (2) 처리 대상이 되는 데이터의 범위를 확장

영상처리는 단순히 정적인 이미지를 다루는 것뿐만 아니라, 여러 형태의 시각적 데이터를 포함함

따라서 단순한 사진(Image)뿐만 아니라 열화상, 의료 영상(MRI, CT Scan), 인공위성 사진, X-ray, 심지어 

3D Depth Map 같은 다양한 데이터를 포괄하기 위해 **영상(Image)**이라는 표현을 사용

    (3) OpenCV 및 컴퓨터 비전에서의 관례

OpenCV 및 컴퓨터 비전 기술에서는 이미지 처리(Image Processing)와 영상 처리(Video Processing)를 함께 다룸

많은 연구 논문 및 교재에서도 디지털 이미지를 "영상"이라고 칭하는 경우가 많음

특히, 카메라에서 입력된 데이터는 단일 이미지라도 비디오 스트림의 일부로 간주될 수 있기 때문에 "영상"이라는 용어가 일반적으로 사용됨

#### 3. 정리

"영상"이라고 표현하는 이유는 단순한 정적인 이미지(사진)뿐만 아니라, 디지털 신호로 구성된 2D 데이터 전체를 포괄하는 개념으로 사용하기 때문

즉, "영상"이라는 용어는 컴퓨터 비전 및 영상처리 분야에서의 전문 용어이며, 단순한 그림을 넘어 데이터를 의미하는 포괄적인 표현으로 사용됨

### 픽셀이 뭐지?

##### 📌 픽셀(Pixel)의 의미와 역할

###### 🔹 픽셀(Pixel)이란?
- 픽셀(Pixel, **Picture Element**)은 **디지털 이미지에서 가장 작은 단위**.  
- 디지털 이미지는 수많은 픽셀들이 모여 형성되는 격자(Grid) 형태의 데이터.  
- **픽셀 하나는 특정한 색상 정보를 저장하며, 여러 픽셀이 조합되어 하나의 이미지를 표현**.  
- **컴퓨터 디스플레이, 디지털 이미지, 동영상 등 모든 시각적 데이터는 픽셀의 조합으로 이루어짐**.

###### 🔹 픽셀의 역할
1. **이미지를 구성하는 요소**
   - 픽셀은 디지털 이미지의 기본 단위로, 수많은 픽셀들이 모여 하나의 그림을 형성.
   - 해상도가 높을수록 더 많은 픽셀이 포함되며, 이미지가 더욱 선명하게 보이는 특징.

2. **색상 정보를 저장하는 역할**
   - **그레이스케일(흑백) 이미지**: 각 픽셀은 `0~255` 범위의 밝기(Gray Scale) 값만 저장.  
     - `0 = 검은색`, `255 = 흰색`, `중간 값 = 회색`.  
   - **컬러 이미지(RGB/BGR)**: 각 픽셀은 **빨강(R), 초록(G), 파랑(B) 값(0~255, 8비트 × 3)을 저장**.  
     - 예: `[255, 0, 0] = 빨간색`, `[0, 255, 0] = 초록색`, `[0, 0, 255] = 파란색`.

3. **디지털 디스플레이를 구성하는 역할**
   - **모니터, TV, 스마트폰 화면**은 수많은 픽셀로 구성되며, 픽셀 하나하나가 특정한 색을 표시하는 역할.
   - 픽셀 밀도(PPI, Pixel Per Inch)가 높을수록 더 선명한 화면을 표현하는 특징.

4. **이미지 및 영상 처리에서 분석 단위로 사용되는 역할**
   - 컴퓨터 비전(Computer Vision) 및 이미지 프로세싱에서 **픽셀 단위의 연산을 통해 이미지 필터링, 변형, 분석이 가능**.  
   - 예: 얼굴 인식, 물체 탐지, 색상 보정, 노이즈 제거 등.

###### 🔹 픽셀 단위로 이미지 조작 (Python 예제)
```python
import cv2 as cv
import numpy as np

# 3x3 픽셀의 작은 컬러 이미지 생성
img = np.array([
    [[255, 0, 0], [0, 255, 0], [0, 0, 255]],   # 빨강, 초록, 파랑
    [[255, 255, 0], [255, 0, 255], [0, 255, 255]],  # 노랑, 자홍, 청록
    [[0, 0, 0], [127, 127, 127], [255, 255, 255]]   # 검정, 회색, 흰색
], dtype=np.uint8)

print("픽셀 값:", img[1, 1])  # (1,1) 위치의 픽셀 출력 (자홍색)
```
**출력:**
```
픽셀 값: [255   0 255]  # 자홍색 (Magenta)
```

##### 📌 하나의 픽셀이 표현할 수 있는 색상의 개수

###### 🔹 8비트(R) × 8비트(G) × 8비트(B)의 조합
- **각 채널(R, G, B)은 8비트(2⁸ = 256) 개의 값**을 가질 수 있음.
- 픽셀 하나는 **세 개의 색상 채널(R, G, B)을 조합하여 색상을 표현**.
- 따라서, 픽셀 하나가 표현할 수 있는 색상 개수는 2^8 X 2^8 X 2^8  
  **즉, 하나의 픽셀은 16,777,216가지 색상을 표현할 수 있음.** 🎨  

###### 🔹 각 비트 수에 따른 표현 가능한 색상 수 비교
| 비트 수 | 표현 가능한 색상 수 |
|---------|------------------|
| 1비트 (흑백) | `2^1 = 2` (검은색, 흰색) |
| 8비트 (그레이스케일) | `2^8 = 256` (흑백 + 254단계의 회색) |
| 16비트 (High Color) | `2^{16} = 65,536` |
| 24비트 (True Color, RGB) | `2^{24} = 16,777,216` |
| 32비트 (RGBA, 투명도 포함) | `2^{32} = 4,294,967,296` |

###### 🔹 32비트(RGBA) 컬러의 경우
- 32비트 컬러에서는 **RGB(24비트) + A(Alpha, 투명도 8비트)까지 포함**됨.
- 32비트 색상 표현 가능 개수:
  
  \[
  2^{32} = 4,294,967,296
  \]
  **즉, 42억 개 이상의 색상을 표현 가능**하며, 투명도를 조절할 수 있음.

##### 📌 결론
- 픽셀은 **디지털 이미지의 최소 단위**로, **색상 정보를 저장하고, 모여서 화면을 구성하는 역할**.
- 픽셀의 조합을 통해 **디지털 디스플레이, 이미지, 동영상이 표현되는 원리**.
- 하나의 픽셀은 **RGB(8비트 × 3)를 통해 16,777,216가지 색상을 표현 가능**.
- 픽셀의 해상도와 밀도가 높을수록 **더 선명한 이미지와 화면을 구현할 수 있는 특징**.

🚀 **즉, 픽셀은 디지털 영상과 이미지를 구성하고 표현하는 가장 중요한 기본 단위!** 🎨

### 왜 RGB 각각 채널의 데이터 값이 2^8 (0~255) 범위를 가지는 걸까? 

##### 📌 1. 비트(Bit)는 무엇인가?
🔹 **"비트(Bit)"는 컴퓨터가 정보를 저장하고 처리하는 가장 작은 단위**  
컴퓨터는 **0과 1(이진수, Binary)** 만 이해할 수 있음  
따라서, 모든 데이터(숫자, 문자, 이미지)는 **0과 1로 이루어진 비트의 조합**으로 저장

✅ **비트 예시**
- `1비트(bit)` → `0` 또는 `1` (총 **2개 값** 저장 가능)
- `2비트(bit)` → `00`, `01`, `10`, `11` (총 **4개 값** 저장 가능)
- `3비트(bit)` → `000`, `001`, `010`, `011`, `100`, `101`, `110`, `111` (총 **8개 값** 저장 가능)

💡 **즉, `n` 비트라면 `2^n` 개의 값을 저장할 수 있음!**

##### 📌 2. 8비트(1바이트)로 저장하는 이유
🔹 **8비트(1바이트)의 경우**
```python
8비트 = 2^8 = 256개의 값 표현 가능 (0~255)
```
- `8비트`는 `00000000 ~ 11111111` (이진수)로 표현 가능
- 이를 **십진수로 변환하면 `0~255`까지 총 256개 값**이 됨

✅ **8비트 예시**
| **이진수(2진법)** | **십진수(10진법)** |
|----|----|
| `00000000` | `0` (검은색) |
| `00000001` | `1` |
| `00000010` | `2` |
| ... | ... |
| `11111110` | `254` |
| `11111111` | `255` (최대값, 가장 밝은 색) |

💡 **즉, 8비트로 저장하면 한 채널당 `0~255` 범위의 색상을 표현할 수 있음!**  
🔹 `0 = 어두운 색`, `255 = 밝은 색` 으로 설정하면 이미지의 색상을 효과적으로 표현할 수 있음

##### 📌 3. RGB 이미지에서 8비트를 사용하는 이유
디지털 이미지에서 **각각의 색상(R, G, B)** 은 **독립적인 8비트 값(0~255)로 표현**됨
즉, **각 픽셀은 3개의 8비트 값을 갖고(R, G, B), 이를 합쳐서 색상을 표현**함

✅ **예제:**
- `(255, 0, 0)` → **빨간색 (Red)**  
  - R(빨강) = 255 (최대), G(초록) = 0, B(파랑) = 0  
- `(0, 255, 0)` → **초록색 (Green)**
  - R = 0, G = 255, B = 0  
- `(0, 0, 255)` → **파란색 (Blue)**
  - R = 0, G = 0, B = 255  
- `(255, 255, 255)` → **흰색 (White)**
  - R = 255, G = 255, B = 255 (모든 색상 채널 최댓값)

🔹 **결과적으로, RGB 색상을 8비트씩 저장하면**
```python
8비트(R) + 8비트(G) + 8비트(B) = 24비트
```
총 `2^(8+8+8) = 16,777,216` 가지 색상 표현 가능! 🎨

💡 **즉, 8비트씩 사용하면 메모리를 효율적으로 사용하면서도 사람이 구별할 수 있는 충분한 색상을 표현할 수 있음**

##### 📌 4. 8비트가 표준이 된 이유
###### **1️⃣ 256단계(0~255)로도 충분한 색상 표현 가능**
- 인간의 눈은 8비트(256단계) 수준에서도 대부분의 색상을 자연스럽게 인식 가능
- 16비트(65,536단계)처럼 높은 비트를 사용하면 더 정밀하지만, 일반적인 디스플레이에서는 차이가 크지 않음

###### **2️⃣ 메모리와 속도 효율성**
- `8비트(uint8)`은 메모리 사용이 적고, 연산 속도가 빠름
- `16비트(uint16)` 또는 `32비트(float32)`는 더 많은 메모리를 차지하고, 연산이 느림

###### **3️⃣ 기본적인 디지털 이미지 표준**
- 대부분의 이미지 포맷(JPG, PNG, BMP 등)은 8비트 컬러를 기본으로 사용
- 디지털 카메라, 모니터, 그래픽 카드에서도 **8비트 컬러(24비트 RGB)를 표준**으로 지원

##### 📌 5. 결론: 왜 RGB 각각 8비트(0~255)인가?
✅ **1바이트(8비트)로 표현 가능하므로 각 채널은 `0~255` 값 저장 가능**  
✅ **RGB 각각 8비트씩 할당하면 총 24비트(16,777,216개 색상) 표현 가능**  
✅ **256단계 색상 변화는 인간이 충분히 부드럽게 인식할 수 있음**  
✅ **메모리와 속도 효율성이 뛰어나므로 디지털 이미지의 표준으로 사용됨**  

---

🚀 **결론:**  
**RGB 각각 8비트(0~255)를 사용하는 것은 "컴퓨터 메모리 효율성 + 색상 표현력 + 인간의 시각 감지 능력"을 고려한 최적의 선택!** 🎨

#### 크기 조정

영상의 크기는 resize 함수를 이용하여 조정할 수 있음. Interpolation 방식을 option으로 지정

- cv.INTER_AREA
- cv.INTER_CUBIC		빈 부분을 3차 다항식으로 보간
- cv.INTER_LINEAR		빈 부분을 양선형식으로 보간

---

##### 📌 보간(Interpolation)의 개념과 활용

###### 🔹 보간(Interpolation)이란?
보간법은 **두 개 이상의 값이 주어졌을 때, 그 사이의 값을 추정하는 방법**을 의미함.  
예를 들어, 아래와 같은 숫자가 있다고 가정.

```python
x = [0, 1, 2, 3]
y = [10, 20, ?, 40]
```
여기서 `?`에 들어갈 값을 찾는 것이 **보간(Interpolation)**.  
간단한 방법으로 **평균**을 사용하면 `y[2] = 30`이 됨.

---

##### 📌 보간법의 종류

###### 🔹 (1) 선형 보간 (Linear Interpolation)
- **두 점을 잇는 직선의 방정식을 사용하여 중간 값을 구하는 방식**.  
- **가장 간단한 보간 방법**.  
- 두 개의 점 \((x_1, y_1)\)과 \((x_2, y_2)\)가 주어졌을 때, 중간 값 \(y\)는 다음과 같이 계산됨.
  
  $$
  y = y_1 + \frac{x - x_1}{x_2 - x_1} \times (y_2 - y_1)
  $$
  
✅ **예제**  
주어진 점: \((1,10)\), \((3,30)\)  
`x = 2`일 때의 값을 구하면,

$$
  y = 10 + \frac{2 - 1}{3 - 1} \times (30 - 10) = 10 + \frac{1}{2} \times 20 = 20
$$

즉, \( y(2) = 20 \)이 됨.

---

###### 🔹 (2) 양선형 보간 (Bilinear Interpolation)
- **2차원 평면에서 사용되는 보간법으로, 두 방향(가로 & 세로)으로 선형 보간을 수행하는 방식**.  
- 주로 **이미지 크기 조정**에서 사용됨.  
- 한 축에서 먼저 보간한 후, 다른 축에서 보간 수행.  

✅ **예제**
- 이미지 크기를 확대할 때 **주변 픽셀 값을 이용하여 새로운 픽셀 값을 계산**하는 데 사용됨.

---

###### 🔹 (3) 다항식 보간 (Polynomial Interpolation)
- **직선이 아닌 다항식(곡선)을 사용하여 보간하는 방법**.  
- 대표적인 방법:
  - **라그랑주 보간법(Lagrange Interpolation)**
  - **뉴턴 보간법(Newton Interpolation)**
- **더 정밀한 결과를 얻을 수 있지만, 계산량이 많아질 수 있음**.

---

##### 📌 이미지 처리에서 보간법 활용
이미지에서 보간법은 **픽셀 값을 예측하여 이미지 크기를 조정하거나 변형할 때 사용**됨.

###### 🔹 OpenCV에서 제공하는 보간법

| **보간법** | **특징** |
|------------------|--------------------------------------------------|
| `cv.INTER_NEAREST` | 최근접 이웃 보간 (**가장 단순**, 블록 형태 유지) |
| `cv.INTER_LINEAR` | 선형 보간 (**기본값**, 부드러운 변환) |
| `cv.INTER_CUBIC` | 3차 다항식 보간 (**더 부드러운 변환**, 연산량 증가) |
| `cv.INTER_LANCZOS4` | Lanczos 보간 (**고해상도에서 선명한 결과 제공**) |

✅ **예제: 이미지 크기 조정**
```python
import cv2 as cv
import matplotlib.pyplot as plt

# 이미지 불러오기
img = cv.imread("china.jpg")

# 크기 조정 (선형 보간 사용)
resized = cv.resize(img, (img.shape[1] * 2, img.shape[0] * 2), interpolation=cv.INTER_LINEAR)

# 결과 표시
plt.imshow(cv.cvtColor(resized, cv.COLOR_BGR2RGB))
plt.show()
```

---

##### 📌 결론
✅ **보간(Interpolation)은 두 점 사이의 값을 예측하는 방법**.  
✅ **이미지 처리에서는 크기 조정, 회전, 변형 등에 필수적으로 사용됨**.  
✅ **실습을 진행할 때 가장 많이 사용되는 보간법은 선형 보간(Linear Interpolation)과 양선형 보간(Bilinear Interpolation)**.  

🚀 **즉, 보간법은 이미지 및 데이터 처리에서 필수적인 기법으로, 가장 적절한 보간 방법을 선택하는 것이 중요함!** 🎨




````python
# 크기 조정
resized_img = cv.resize(img, (c//3, r//3), interpolation=cv.INTER_LINEAR)

# cv.resize() 함수
    # OpenCV에서 이미지를 특정 크기로 조정하는 함수
        # 기본 구조:
        # cv.resize(src, dsize, fx, fy, interpolation)
            # src → 원본 이미지
            # dsize → 변경할 이미지 크기 (너비, 높이) (width, height)
            # fx, fy → 너비 및 높이 비율(생략 가능)
            # interpolation → 크기 조정 방식

# (c//3, r//3) 부분
    # c//3, r//3 → 가로, 세로 크기를 1/3로 축소
        # c → 원본 이미지의 너비 (width)
        # r → 원본 이미지의 높이 (height)
        # 주의! cv.resize()의 dsize는 (width, height) 순서이므로 **(c//3, r//3)**를 사용해야 올바르게 작동.

# interpolation=cv.INTER_LINEAR
    # 보간법(Interpolation): 이미지를 축소하거나 확대할 때 픽셀 값을 새롭게 계산하는 방식.
    # cv.INTER_LINEAR → 선형 보간법 (기본값, 보통 축소 시 사용)
        # 축소 시 이미지가 부드럽게 유지됨.
        # 확대할 때는 cv.INTER_CUBIC이나 cv.INTER_LANCZOS4를 사용하면 더 나은 품질 제공.
        
plt.imshow(cv.cvtColor(resized_img, cv.COLOR_BGR2RGB))
plt.show()
````

#### 이미지 이동

````python
M = np.float32(
    [
        [1, 0 ,100], 
        [0, 1, 100]
    ]
)
# M은 2×3 크기의 변환 행렬(affine transformation matrix)
# Affine 변환(아핀 변환)은 선형 변환(회전, 크기 변경, 이동 등)을 포함하는 변환을 의미
# 이 행렬의 역할은 이미지를 x축과 y축 방향으로 이동하는 것

            # 요소	        의미
        # ----------------------------------
            # 1	        x 좌표 유지 (scale=1)
            # 0	        x 방향으로 회전 없음
            # 100	    x축으로 100픽셀 이동
        # ----------------------------------
            # 0	        y 방향으로 회전 없음
            # 1	        y 좌표 유지 (scale=1)
            # 100	    y축으로 100픽셀 이동
# 즉, 이미지를 x축(오른쪽)으로 100픽셀, y축(아래쪽)으로 100픽셀 이동하는 역할

moved_img = cv.warpAffine(img, M, (c, r))
# cv.warpAffine() 함수는 2D 아핀 변환을 적용하여 이미지를 변형하는 함수
#(src → 변환할 원본 이미지, M -> 2×3 변환 행렬 (np.float32 타입), dsize -> 변환된 이미지의 크기 (너비, 높이) = (c, r) (원본 이미지 크기 유지))

plt.imshow(cv.cvtColor(moved_img, cv.COLOR_BGR2RGB))
plt.show()

````

#### 이미지 회전

````python
print(r,c) # r = 427, c = 647 

M = cv.getRotationMatrix2D((c/2, r/2), 60, 1)
print(M) # 2D 회전 변환을 위한 Affine 변환 행렬(2×3 행렬)을 생성
# c = 647, r = 427을 기준으로 나누기 2해서 중심 설정하고, 해당 중심을 기준으로 60도 회전
# 60 -> 이미지를 중심을 기준으로 반시계 방향 60도 회전(양수 값(+) → 반시계 방향, 음수 값(-) → 시계 방향 회전을 의미: -60 = 시계 방향으로 60도 회전)
# 1 -> 스케일 유지(크기 변환 없음)

rotated_img = cv.warpAffine(img, M, (c, r))
# cv.warpAffine()을 사용하여 이미지에 회전 변환 적용
# (c, r) → 원본 크기(640x427)를 유지하면서 회전된 이미지 출력


plt.imshow(cv.cvtColor(rotated_img, cv.COLOR_BGR2RGB))
plt.show()

# 이미지가 반시계 방향으로 60도 회전
# 원본 크기를 유지했으므로, 회전 후 잘려 보임
    # 잘리는 문제 해결 방법
    # 이미지를 회전할 때 일부 영역이 잘리는 문제를 방지하려면 출력 이미지 크기를 더 크게 설정하는 방법이 있음
# 회전 후 빈 공간은 기본적으로 검정색(픽셀 값 0)으로 채워짐
````

````python
g = np.zeros((256, 256), np.uint8)
# 이미지 크기: 256 × 256
# np.uint8은 NumPy 데이터 타입을 지정하는 부분
# 각 픽셀의 데이터 유형이 uint8 (Unsigned 8-bit Integer, 부호 없는 8비트 정수형) 이라는 의미
# 이미지 처리에서 가장 많이 사용하는 데이터 타입 (픽셀 값 저장에 최적화됨)

# 데이터 타입 종류
# 데이터 타입	        크기	      값 범위
# --------------------------------------
# np.uint8    8비트(1바이트)	    0 ~ 255
# np.int8	    8비트(1바이트)	    -128 ~ 127
# np.uint16	16비트(2바이트)	        0 ~ 65,535
# np.float32	32비트(4바이트)	    실수 표현

print(g)

for i in range(256):
    g[i, :] = i 
    # g[i, :] → i번째 행(가로줄)의 모든 열을 선택
    # g[0, :] = 밝기 0 → 맨 위는 검은색 (0)
    # g[255, :] = 밝기 255 → 맨 아래는 흰색 (255)

plt.imshow(g, cmap="gray")
plt.show()
````

#### 이미지 임계값 처리 (Thresholding)

````python
# 이미지 임계값 처리는 특정 픽셀 값을 기준으로 픽셀을 이진화하는 기법. 
# OpenCV에서는 cv.threshold() 함수를 사용하여 임계값을 적용할 수 있습니다.

# 기본적인 Thresholding (OpenCV에서 지원하는 임계값 처리 방법 중 대표적인 두 가지)
#     cv.THRESH_BINARY: 픽셀 값이 임계값보다 크면 최대값(255), 작으면 0으로 변환
#     cv.THRESH_BINARY_INV: THRESH_BINARY와 반대로 동작

# 임계값 적용
ret, h1 = cv.threshold(g, 127, 255, cv.THRESH_BINARY)
ret, h2 = cv.threshold(g, 127, 255, cv.THRESH_BINARY_INV)

# 결과 출력
fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(g, cmap="gray")
ax[0].set_title("Original Grayscale Image")

ax[1].imshow(h1, cmap="gray")
ax[1].set_title("Threshold Binary")

ax[2].imshow(h2, cmap="gray")
ax[2].set_title("Threshold Binary INV")

plt.show()
````

````python
# 이미지 불러오기 (그레이스케일로 변환)
img = cv.imread("china.jpg", cv.IMREAD_GRAYSCALE)
# 임계값 적용 (Threshold Binary)
ret, binary = cv.threshold(img, 127, 255, cv.THRESH_BINARY)
# 임계값 적용 (Threshold Binary Inverted)
ret, binary_inv = cv.threshold(img, 127, 255, cv.THRESH_BINARY_INV)

# 결과 출력
fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(img, cmap="gray")
ax[0].set_title("Original Grayscale Image")

ax[1].imshow(binary, cmap="gray")
ax[1].set_title("Threshold Binary")

ax[2].imshow(binary_inv, cmap="gray")
ax[2].set_title("Threshold Binary INV")

plt.show()
````

````python
# 이미지 불러오기 (컬러 이미지)
img = cv.imread("china.jpg")  # BGR 형식

# BGR 채널 분리
b, g, r = cv.split(img)

# 각각의 채널에 대해 임계값 적용
ret, b_thresh = cv.threshold(b, 127, 255, cv.THRESH_BINARY)
ret, g_thresh = cv.threshold(g, 127, 255, cv.THRESH_BINARY)
ret, r_thresh = cv.threshold(r, 127, 255, cv.THRESH_BINARY)

# BGR 채널을 다시 합치기
img_thresh = cv.merge([b_thresh, g_thresh, r_thresh])

# 결과 출력
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
ax[0].imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))
ax[0].set_title("Original Image")

ax[1].imshow(cv.cvtColor(img_thresh, cv.COLOR_BGR2RGB))
ax[1].set_title("Thresholded Image (Per Channel)")

plt.show()
````

````python
# 이미지 불러오기
img = cv.imread("china.jpg")

# BGR → HSV 변환
hsv = cv.cvtColor(img, cv.COLOR_BGR2HSV)
h, s, v = cv.split(hsv)  # Hue(색상), Saturation(채도), Value(명도) 채널 분리

# 명도(V) 채널에 임계값 적용
ret, v_thresh = cv.threshold(v, 127, 255, cv.THRESH_BINARY)

# 새로운 HSV 영상 만들기 (H, S는 원본 유지, V만 변경)
hsv_thresh = cv.merge([h, s, v_thresh])

# 다시 BGR로 변환
img_hsv_thresh = cv.cvtColor(hsv_thresh, cv.COLOR_HSV2BGR)

# 결과 출력
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
ax[0].imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))
ax[0].set_title("Original Image")

ax[1].imshow(cv.cvtColor(img_hsv_thresh, cv.COLOR_BGR2RGB))
ax[1].set_title("Thresholded Image (Using V Channel)")

plt.show()
````

2. Bilinear(양선형) interpolation

color_rect 함수를 수정하여 오른쪽 그림과 같이 256x256 크기의 color 영상을 bilinear interpolation에 의해 생성하는 프로그램을 작성.

방법:

a. 네 모서리의 색을 지정: (255, 0, 0), (0, 0, 255), (0, 255, 0), (255, 255, 0) 등

b. 나머지 화소들의 색상을 bilinear interpolation 에 의해 계산함.

오른쪽 그림과 동일한 결과가 얻어지는지 확인할 것.

````python
import numpy as np
import matplotlib.pyplot as plt

def bilinear_interpolation(top_left, top_right, bottom_left, bottom_right):
    # 256x256 크기의 컬러 이미지 생성
    img = np.zeros((256, 256, 3), dtype=np.uint8)

    for y in range(256): # y 방향(세로 방향)으로 반복
        for x in range(256): # x 방향(가로 방향)으로 반복
            alpha_x = x / 255.0  # x 방향 보간 비율
            alpha_y = y / 255.0  # y 방향 보간 비율

            # x 좌표 기준으로 1차 보간 (가로 방향 값 계산)
            top_interp = (1 - alpha_x) * top_left + alpha_x * top_right
            bottom_interp = (1 - alpha_x) * bottom_left + alpha_x * bottom_right

            # 2차 보간 (위에서 구한 가로 방향 보간 값을 이용하여 세로 방향 보간을 적용 후 최종 색상을 계산)
            pixel_value = (1 - alpha_y) * top_interp + alpha_y * bottom_interp

            img[y, x] = pixel_value  # 최종 계산된 색상 할당

    return img

# 모서리 색상 지정 RGB
top_left = np.array([255, 0, 0])        # 빨강
top_right = np.array([0, 0, 255])       # 파랑
bottom_left = np.array([0, 255, 0])     # 초록
bottom_right = np.array([255, 255, 0])  # 노랑

# Bilinear Interpolation을 통해 계산
result = bilinear_interpolation(top_left, top_right, bottom_left, bottom_right)

# 결과 이미지 출력
plt.imshow(result)
plt.show()
````
