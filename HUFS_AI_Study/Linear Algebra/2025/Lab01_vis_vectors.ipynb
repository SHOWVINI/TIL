- Normalization 후, 본인과의 dot product의 절대값이 가장 낮은 학생은 누구인가? (자기자신은 제외)
- 본인, 본인과의 dot product의 절대값이 가장 높은 학생, 본인과의 dot product의 절대값이 가장 낮은 학생의 vector를 시각화를 해보자 (5개의 차원 중 관심있는 2개 또는 3개의 차원을 선택하여 시각화).
- 모든 학생들의 vector를 본인이 가장 관심있는 2개의 차원에 대해 시각화를 해보자.

```python
# [설정] 관심 차원 정의
INTEREST_DIMS_2D = ['NS', 'TF'] # 2차원 시각화 관심 축 정의
INTEREST_DIMS_3D = ['EI', 'TF', 'JP'] # 3차원 시각화 관심 축 정의
ALL_DIMS = ['EI', 'NS', 'TF', 'JP'] # 전체 분석 대상 축 정의
SELF_NAME = 'Hyunbin' # 본인 이름 정의
```

```python
# [1] 데이터 로드 및 기본 전처리
xlsx_path = 'BME_MBTI_2025.xlsx' # 엑셀 파일 경로 정의

df = pd.read_excel(xlsx_path) # 엑셀 파일 로드 수행
assert set(['Name','EI','NS','TF','JP']).issubset(df.columns), "필수 열(Name, EI, NS, TF, JP) 누락" # 열 검증

df_full = df.copy() # 전체 데이터프레임 복사본 생성

df_full
```

```python
# [2] 본인 인덱스 탐색 및 벡터 준비
self_rows = df_full.index[df_full['Name'] == SELF_NAME].tolist() # 본인 이름 행 인덱스 리스트 추출
assert len(self_rows) > 0, "데이터셋에서 본인(Hyunbin)을 찾을 수 없음" # 본인 존재성 검증
self_idx = self_rows[0] # 중복 시 첫 번째 인덱스 사용


X_all = df_full[ALL_DIMS].to_numpy(dtype=float) # 전체 4차원 성향 행렬 추출
names_all = df_full['Name'].astype(str).to_numpy() # 이름 배열 추출


self_vec = X_all[self_idx] # 본인 벡터 추출

self_vec
```

```python
# [3] 정규화 유틸리티 정의
def row_l2_normalize(X: np.ndarray, eps: float = 1e-12) -> np.ndarray:
    """행 벡터 L2 정규화 수행 함수 정의
    - 각 행에 대해 ||x||_2로 나누어 단위 벡터화 수행
    - 영벡터 방지를 위해 분모에 epsilon 적용
    """
    norms = np.linalg.norm(X, axis=1, keepdims=True) # 각 행 L2 노름 계산
    norms = np.maximum(norms, eps) # 영노름 방지 처치
    return X / norms # 정규화 결과 반환

X_all_norm = row_l2_normalize(X_all) # 전체 4차원 행 벡터 정규화 수행
self_vec_norm = X_all_norm[self_idx] # 본인 정규화 벡터 추출

self_vec_norm
```

```python
# [4] 내적 기반 유사도 계산(원본/정규화)
# 4-1) 원본 내적 벡터 계산
raw_dots = X_all @ self_vec # 모든 학생과 본인 간 원본 내적 계산
# 4-2) 정규화(=코사인) 내적 벡터 계산
cos_dots = X_all_norm @ self_vec_norm # 모든 학생과 본인 간 코사인 유사도 계산
```

```python
cos_dots
```

```python
# [5] 자기 자신 제외 후, 절댓값 최대/최소 인덱스 탐색
idx_all = np.arange(len(names_all)) # 전체 인덱스 배열 생성
mask_others = idx_all != self_idx # 자기 자신 제외 마스크 생성
```

```python
# 5-1) 원본 내적(|dot|) 기준
abs_raw = np.abs(raw_dots) # 원본 내적 절댓값 계산
abs_raw_others = abs_raw[mask_others] # 자신 제외 절댓값 배열 추출
idx_others = idx_all[mask_others] # 자신 제외 인덱스 배열 추출

raw_argmax_idx = idx_others[np.argmax(abs_raw_others)] # 절댓값 최대 인덱스 산출
raw_argmin_idx = idx_others[np.argmin(abs_raw_others)] # 절댓값 최소 인덱스 산출
```

```python
# 5-2) 정규화(코사인) 내적(|dot|) 기준
abs_cos = np.abs(cos_dots) # 코사인 유사도 절댓값 계산
abs_cos_others = abs_cos[mask_others] # 자신 제외 절댓값 배열 추출


cos_argmax_idx = idx_others[np.argmax(abs_cos_others)] # 절댓값 최대 인덱스 산출
cos_argmin_idx = idx_others[np.argmin(abs_cos_others)] # 절댓값 최소 인덱스 산출
```

```python
# [6] 결과 요약 출력
print('[결과 요약]')
print(f"본인: {names_all[self_idx]}")
print('\n[원본 내적 |dot| 기준]')
print(f"최대 유사: {names_all[raw_argmax_idx]} (dot={raw_dots[raw_argmax_idx]:.3f}, |dot|={abs_raw[raw_argmax_idx]:.3f})")
print(f"최소 유사: {names_all[raw_argmin_idx]} (dot={raw_dots[raw_argmin_idx]:.3f}, |dot|={abs_raw[raw_argmin_idx]:.3f})")


print('\n[정규화(코사인) |dot| 기준]')
print(f"최대 유사: {names_all[cos_argmax_idx]} (cos={cos_dots[cos_argmax_idx]:.3f}, |cos|={abs_cos[cos_argmax_idx]:.3f})")
print(f"최소 유사: {names_all[cos_argmin_idx]} (cos={cos_dots[cos_argmin_idx]:.3f}, |cos|={abs_cos[cos_argmin_idx]:.3f})")
```
